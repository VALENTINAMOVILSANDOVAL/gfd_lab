# Baroclinic Instability

## Eady Problem

We are going to analize flows on an f-plane with uniform zonal shear.

The zonally symmetric part of the Montgomery potential is defined as:

$$M_{Z}=-f\Lambda(\theta-\theta_{M})y \tag{1}$$

The ambient zonal wind is defined from the Montgomery potential:

$$U(\theta)=-\frac{1}{f}\frac{\partial M_{Z}}{\partial y}=\Lambda(\theta-\theta_{M}) \tag{2}$$

The potential vorticity associated with the mean sheared flow is:

$$q_{Z}=q_{0}\left(\frac{1}{f^{2}}\nabla_{h}^{2}M_{Z}+\frac{\Gamma_{R}^{2}}{N^{2}}\frac{\partial^{2}M_{Z}}{\partial\theta^{2}}\right)=0 \tag{3}$$

The part of the Montgomery potential associated with a developing disturbance, $M^{\prime}$, satisfies:

$$\frac{1}{f^{2}}\nabla_{h}^{2}M^{\prime}+\frac{\Gamma_{R}^{2}}{N_{R}^{2}}\frac{\partial^{2}M^{\prime}}{\partial\theta^{2}}=0 \tag{4}$$

Upper and lower boundary conditions on $M^{*}$ (assuming no terrain) are evaluated on reference levels $\theta_{T0}$ and $\theta_{B0}$:

$$\left(\frac{\partial M^{*}}{\partial\theta}\right)_{T0}=\frac{N_{R}^{2}}{\Gamma_{R}^{2}}\theta_{T}^{*} \quad \left(\frac{\partial M^{*}}{\partial\theta}\right)_{B0}=\frac{N_{R}^{2}}{\Gamma_{R}^{2}}\theta_{B}^{*} \tag{5}$$

The governing equations for potential temperature anomalies $\theta_{T}^{*}$ and $\theta_{B}^{*}$ at the boundaries are:

$$\frac{\partial\theta_{T}^{*}}{\partial t}+u_{gT0}\frac{\partial\theta_{T}^{*}}{\partial x}+v_{gT0}\frac{\partial\theta_{T}^{*}}{\partial y}=0 \tag{6}$$

$$\frac{\partial\theta_{B}^{*}}{\partial t}+u_{gB0}\frac{\partial\theta_{B}^{*}}{\partial x}+v_{gB0}\frac{\partial\theta_{B}^{*}}{\partial y}=0 \tag{7}$$

By inserting geostrophic winds, eliminating $\theta_{T}^{*}$ and $\theta_{B}^{*}$, and linearizing, the boundary conditions for the disturbance $M^{\prime}$ become:

$$\frac{\partial}{\partial t}\left(\frac{\partial M^{\prime}}{\partial\theta}\right)_{T0}+\Lambda\Delta\theta\frac{\partial}{\partial x}\left(\frac{\partial M^{\prime}}{\partial\theta}\right)_{T0}-\Lambda\frac{\partial M_{T0}^{\prime}}{\partial x}=0 \tag{8}$$

$$\frac{\partial}{\partial t}\left(\frac{\partial M^{\prime}}{\partial\theta}\right)_{B0}-\Lambda\Delta\theta\frac{\partial}{\partial x}\left(\frac{\partial M^{\prime}}{\partial\theta}\right)_{B0}-\Lambda\frac{\partial M_{B0}^{\prime}}{\partial x}=0 \tag{9}$$

We try a solution of the form:

$$M^{\prime}=A\exp[i(kx-\omega t)+m(\theta-\theta_{M})]+B\exp[i(kx-\omega t)-m(\theta-\theta_{M})] \tag{10}$$

Substituting into equation (4) gives

$$m=\frac{kN_{R}}{f\Gamma_{R}} \tag{11}$$

Substituting into the boundary equations (8) and (9) results in the matrix:

$$\begin{pmatrix}(c+\Lambda\Delta\theta+\Lambda/m)e^{-m\Delta\theta}&-(c+\Lambda\Delta\theta-\Lambda/m)e^{m\Delta\theta}\\(c-\Lambda\Delta\theta+\Lambda/m)e^{m\Delta\theta}&-(c-\Lambda\Delta\theta-\Lambda/m)e^{-m\Delta\theta}\end{pmatrix}\begin{pmatrix}A\\B\end{pmatrix}=0 \tag{12}$$

Setting the determinant to zero yields the dispersion relation:

$${\color{blue}\boxed{c=\pm(U_{T}-U_{B})\left[\left(\frac{1}{\kappa}-\frac{1}{2}\right)^{2}-\frac{\coth(\kappa)-1}{\kappa}\right]^{1/2}}} \tag{13}$$

Where the dimensionless horizontal wavenumber is defined as:

$$\kappa=2m\Delta\theta=Lk \tag{14}$$

The dimensionless growth rate $\nu$ is defined as:

$$\nu=\frac{\kappa c_{I}}{U_{T}-U_{B}}=\left\{-(1-\kappa/2)^{2}+\kappa[\coth(\kappa)-1]\right\}^{1/2} \tag{15}$$

Based on symmetry ($|A|=|B|$), setting $A=E\exp(i\phi)$ and $B=E\exp(-i\phi)$ informs us that:

$$\frac{A}{B}=\exp(2i\phi)=-\frac{\kappa-2+i\nu}{\kappa+2+i\nu}\exp(\kappa) \tag{16}$$

From which we conclude:

$$\phi=\frac{1}{2}\arctan\left(\frac{4\nu}{\kappa^{2}+\nu^{2}-4}\right) \tag{17}$$

## Alternate Interpretation of Boundary Conditions

We want to show how a thin layer of potential vorticity can mimic a potential temperature perturbation at a bounding surface.

Adding a thin layer of potential vorticity just above $\theta_{B0}$ yields a total flow-associated potential vorticity of:

$$q_{T}^{*}=q_{0}I_{B}(x,y,t)\delta(\theta-\theta_{B0}-\epsilon)+q^{*} \tag{18}$$

The diagnostic equation for $M^{*}$ becomes:

$$\frac{1}{f_{0}^{2}}\nabla_{h}^{2}M^{*}+\frac{\Gamma_{R}^{2}}{N_{R}^{2}}\frac{\partial^{2}M^{*}}{\partial\theta^{2}}=\frac{q_{T}^{*}}{q_{0}} \tag{19}$$

Integrating over a thin layer surrounding $\theta_{B0}+\epsilon$ results in:

$$\frac{\Gamma_{R}^{2}}{N_{R}^{2}}\left[\left(\frac{\partial M^{*}}{\partial\theta}\right)_{+}-\left(\frac{\partial M^{*}}{\partial\theta}\right)_{-}\right]=I_{B} \tag{20}$$

Setting the layer equal to the temperature perturbation:

$$I_{B}=\theta_{B}^{*} \tag{21}$$

Provides the new surface boundary condition:

$${\color{red}\boxed{\left(\frac{\partial M^{*}}{\partial\theta}\right)_{B0}=\frac{N_{R}^{2}}{\Gamma_{R}^{2}}I_{B}}} \tag{22}$$

The potential vorticity layer is subject to the same governing equation as the surface perturbation:

$$\frac{\partial I_{B}}{\partial t}+v_{gB0}\cdot\nabla_{h}I_{B}=0 \tag{23}$$

For an upper lid, the relationship between potential temperature perturbation and layer potential vorticity has a minus sign:

$$I_{T}=-\theta_{T}^{*} \tag{24}$$

Leading to the upper boundary condition:

$$\left(\frac{\partial M^{*}}{\partial\theta}\right)_{T0}=-\frac{N_{R}^{2}}{\Gamma_{R}^{2}}I_{T} \tag{25}$$

## Rossby Waves and the Mean Flow

Here we develop the linearized quasi-geostrophic equations in isentropic coordinates for a zonal-mean flow pattern.

We assume variables have base state, zonally symmetric, and perturbation components:

$$M=M_{0}(\theta)+M_{Z}(y,\theta)+M^{\prime}=M_{0}(\theta)+M^{*}$$
$$\sigma=\sigma_{0}+\sigma_{Z}+\sigma^{\prime}=\sigma_{0}+\sigma^{*}$$
$$q_{g}=q_{0}(1+\beta y/f_{0})+q_{Z}(y,\theta)+q^{\prime}=q_{0}(1+\beta y/f_{0})+q^{*}$$
$$\Phi=\Phi_{0}(\theta)+\Phi_{Z}(y,\theta)+\Phi^{\prime}=\Phi_{0}(\theta)+\Phi^{*}$$
$$u_{g}=U(y,\theta)+u_{g}^{\prime}$$

$$v_{g}=v_{g}^{\prime} \tag{26}$$

Geostrophic balance dictates:

$$U=-\frac{1}{f_{0}}\frac{\partial M_{Z}}{\partial y} \tag{27}$$

The relationship between Montgomery potential and geopotential yields:

$$\Phi_{Z}=-\theta_{R}\frac{\partial M_{Z}}{\partial\theta} \tag{28}$$

Cross-differentiation gives the thermal wind relationship:

$$\frac{1}{\theta_{R}}\frac{\partial\Phi_{Z}}{\partial y}=f_{0}\frac{\partial U}{\partial\theta} \tag{29}$$

The definition of potential vorticity provides:

$$\frac{q_{Z}}{q_{0}}=-\frac{1}{f_{0}}\frac{\partial U}{\partial y}-\frac{\Gamma_{R}^{2}}{\theta_{R}N_{R}^{2}}\frac{\partial\Phi_{Z}}{\partial\theta}=\frac{1}{f_{0}^{2}}\frac{\partial^{2}M_{Z}}{\partial y^{2}}+\frac{\Gamma_{R}^{2}}{N_{R}^{2}}\frac{\partial^{2}M_{Z}}{\partial\theta^{2}} \tag{30}$$

The disturbance Montgomery potential satisfies:

$$\frac{1}{f_{0}^{2}}\nabla_{h}^{2}M^{\prime}+\frac{\Gamma_{R}^{2}}{N_{R}^{2}}\frac{\partial^{2}M^{\prime}}{\partial\theta^{2}}=\frac{q^{\prime}}{q_{0}} \tag{31}$$

The linearized potential vorticity advection equation is:

$$\left(\frac{\partial}{\partial t}+U\frac{\partial}{\partial x}\right)\left(\nabla_{h}^{2}M^{\prime}+\frac{\partial^{2}M^{\prime}}{\partial\xi^{2}}\right)+\left(\frac{\partial Q}{\partial y}\right)\left(\frac{\partial M^{\prime}}{\partial x}\right)=0 \tag{32}$$

Where $Q$ represents the beta effect and zonal mean flow:

$$Q=f_{0}(\beta y/f_{0}+q_{z}/q_{0}) \tag{33}$$

And $\xi$ is a scaled potential temperature:

$$d\theta=\frac{f_{0}\Gamma_{R}}{N_{R}}d\xi \tag{34}$$

Assuming a plane wave structure gives the analog to the Taylor-Goldstein equation:

$${\color{green}\boxed{\frac{\partial^{2}M^{\prime}}{\partial y^{2}}+\frac{\partial^{2}M^{\prime}}{\partial\xi^{2}}-\mu^{2}M^{\prime}=0}} \tag{35}$$

Where:

$$\mu^{2}=\frac{1}{c-U}\frac{\partial Q}{\partial y}+k^{2} \tag{36}$$

The relationship between $M^{\prime}$ and isentropic density $\sigma^{\prime}$ is:

$$\frac{\partial^{2}M^{\prime}}{\partial\xi^{2}}=-f_{0}^{2}\frac{\sigma^{\prime}}{\sigma_{0}} \tag{37}$$

The relationship between $M^{\prime}$ and perturbation geopotential $\Phi^{\prime}$ is:

$$\frac{\partial M^{\prime}}{\partial\xi}=-\frac{f_{0}\Gamma_{R}}{N_{R}\theta_{R}}\Phi^{\prime} \tag{38}$$

## Conditions for Instability

We obtain the Charney-Stern necessary condition for baroclinic instability.

Multiplying equation (35) by the complex conjugate of the Montgomery potential perturbation $M^{\prime*}$ and integrating yields two components:

$$\int_{-\infty}^{\infty}M^{\prime*}\frac{\partial^{2}M^{\prime}}{\partial y^{2}}dy=\int_{-\infty}^{\infty}\left[\frac{\partial}{\partial y}\left(M^{\prime*}\frac{\partial M^{\prime}}{\partial y}\right)-\left|\frac{\partial M^{\prime}}{\partial y}\right|^{2}\right]dy \tag{39}$$

$$\int_{B}^{T}M^{\prime*}\frac{\partial^{2}M^{\prime}}{\partial\xi^{2}}d\xi=\int_{B}^{T}\left[\frac{\partial}{\partial\xi}\left(M^{\prime*}\frac{\partial M^{\prime}}{\partial\xi}\right)-\left|\frac{\partial M^{\prime}}{\partial\xi}\right|^{2}\right]d\xi \tag{40}$$

Setting boundaries causes the first terms on the right sides to vanish, giving the result:

$$\int_{-\infty}^{\infty}dy\int_{B}^{T}d\xi\left[\left|\frac{\partial M^{\prime}}{\partial y}\right|^{2}+\left|\frac{\partial M^{\prime}}{\partial\xi}\right|^{2}+\left(\frac{1}{c-U}\frac{\partial Q}{\partial y}+k^{2}\right)|M^{\prime}|^{2}\right]=0 \tag{41}$$

Which has the imaginary part:

$${\color{purple}\boxed{c_{I}\int_{-\infty}^{\infty}dy\int_{B}^{T}d\xi\left(\frac{1}{|c-U|^{2}}\frac{\partial Q}{\partial y}\right)|M^{\prime}|^{2}=0}} \tag{42}$$

## Non-interaction Theorem

This section explores the non-interaction theorem for low Rossby number disturbances by examining the meridional flux of potential vorticity.

Assuming a single mode, the meridional flux of potential vorticity is:

$$F_{q}=\frac{1}{L}\int_{0}^{L}Re(v_{g})Re(q^{\prime})dx=\frac{1}{4}[v_{g}q^{\prime*}+v_{g}^{*}q^{\prime}] \tag{43}$$

Using previous relations, the potential vorticity perturbation is:

$$q^{\prime}=\frac{q_{0}}{f_{0}^{2}}\left(\frac{\partial^{2}M^{\prime}}{\partial y^{2}}+\frac{\partial^{2}M^{\prime}}{\partial\xi^{2}}-k^{2}M^{\prime}\right)=\frac{q_{0}M^{\prime}}{f_{0}^{2}(c-U)} \tag{44}$$

Coupled with the definition of geostrophic velocity:

$$v_{g}=\frac{1}{f_{0}}\frac{\partial M^{\prime}}{\partial x} \tag{45}$$

The single-mode flux evaluates to:

$$F_{q}=\frac{ikq_{0}}{f_{0}^{3}}\left[\frac{1}{c^{*}-U}-\frac{1}{c-U}\right]|M^{\prime}|^{2}=-\frac{kq_{0}c_{I}|M^{\prime}|^{2}}{f_{0}^{3}|c-U|^{2}} \tag{46}$$

Expanding this by superimposing multiple modes over the globe yields the final non-interaction result:

$${\color{orange}\boxed{F_{q}=-\frac{q_{0}}{4f_{0}^{3}}\sum_{j}\frac{k_{j}c_{jI}|M_{j}^{\prime}|^{2}}{|c_{j}-U|^{2}}}} \tag{47}$$

### 1. Compute the meridional flux of potential vorticity for the Eady mode. Hint: Treat the upper and lower surface potential temperature distributions as potential vorticity distributions. Comment on the meridional heat transport of the Eady mode

In the Eady mode, the ambient potential vorticity in the interior of the flow is constant ($q_Z = q_0$), meaning no interior potential vorticity anomalies exist ($q'=0$). Because the meridional flux of potential vorticity is defined as $F_q = \overline{v'_g q'}$, the flux in the interior of the domain is exactly zero. 

However, based on the hint, we treat the boundaries as containing thin sheets of potential vorticity.

- At the lower boundary, the potential vorticity sheet is defined as $I_B = \theta_B^*$. The flux here is $F_{q,B} = \overline{v'_g I_B} = \overline{v'_g \theta_B^*}$

- At the upper boundary, the potential vorticity sheet is defined with a minus sign as $I_T = -\theta_T^*$. The flux here is $F_{q,T} = \overline{v'_g I_T} = -\overline{v'_g \theta_T^*}$

Comment on meridional heat transport:

Baroclinic instability is driven by the release of available potential energy, which requires a net poleward (positive) heat transport ($\overline{v'_g \theta'} > 0$). Because the upper wave lags to the west of the lower wave in the Eady mode, the phase shift ensures that $v'_g$ and $\theta'$ are positively correlated at both boundaries. Consequently, the meridional heat transport is positive (poleward) throughout the fluid. Applying this to our boundary PV fluxes:  

- The lower boundary PV flux is positive ($F_{q,B} > 0$).
- The upper boundary PV flux is negative ($F_{q,T} < 0$).

### 2. Using the theory developed in section 10.3, find the dispersion relation and the spatial form of $u_g'$, $v_g$, and $M'$ for a Rossby wave with assumed structure $\exp[i(kx+ly+m\xi-\omega t)]$. Assume that $U=0$ but that $\beta \ne 0$ Do not apply upper or lower boundary conditions

We start with the governing equation for the disturbance Montgomery potential (the analog to the Taylor-Goldstein equation):

$$\frac{\partial^2 M'}{\partial y^2} + \frac{\partial^2 M'}{\partial \xi^2} - \mu^2 M' = 0$$

Where $\mu^2$ is defined as:

$$\mu^2 = \frac{1}{c-U}\frac{\partial Q}{\partial y} + k^2$$

Given $U=0$ and $\beta \ne 0$, the base state parameter $Q$ is simplified. From the definition $Q=f_0(\beta y/f_0 + q_z/q_0)$, we find $\frac{\partial Q}{\partial y} = \beta$. Substituting this into the $\mu^2$ equation yields:

$$\mu^2 = \frac{\beta}{c} + k^2$$

Now, assume the wave structure $M' = M_0 \exp[i(kx+ly+m\xi-\omega t)]$. Taking the second derivatives with respect to $y$ and $\xi$:

$$\frac{\partial^2 M'}{\partial y^2} = -l^2 M' \quad \text{and} \quad \frac{\partial^2 M'}{\partial \xi^2} = -m^2 M'$$

Substitute these back into the governing equation:

$$(-l^2 - m^2 - \mu^2) M' = 0 \implies \mu^2 = -(l^2 + m^2)$$

Equating our two expressions for $\mu^2$:

$$\frac{\beta}{c} + k^2 = -(l^2 + m^2)$$

$$\frac{\beta}{c} = -(k^2 + l^2 + m^2) \implies c = -\frac{\beta}{k^2 + l^2 + m^2}$$

Since $c = \omega/k$, the **dispersion relation** is:

$$\omega = -\frac{\beta k}{k^2 + l^2 + m^2}$$

For the spatial form of the velocities, we use geostrophic balance definitions $u'_g = -\frac{1}{f_0}\frac{\partial M'}{\partial y}$ and $v'_g = \frac{1}{f_0}\frac{\partial M'}{\partial x}$:

- $M' = M_0 \exp[i(kx+ly+m\xi-\omega t)]$
- $u'_g = -\frac{il}{f_0} M_0 \exp[i(kx+ly+m\xi-\omega t)]$
- $v'_g = \frac{ik}{f_0} M_0 \exp[i(kx+ly+m\xi-\omega t)]$

### 4. Redo the non-interaction theorem without substituting equations (10.31) and (10.34) into equation (10.44) for the potential vorticity perturbation.

#### (a) Show that $F_q$ can be written in the form $F_q = -\frac{\partial E_y}{\partial y} - \frac{\partial E_\xi}{\partial \xi}$ and determine the components of the vector $\mathbf{E} = (E_y, E_\xi)$ This vector is called the Eliassen-Palm flux. The Eliassen-Palm flux will only be non-zero when there is wave growth or decay or nonlinear behavior.

We start with the definition of meridional PV flux, $F_q = \overline{v'_g q'}$, and the un-substituted PV perturbation $q' = \frac{q_0}{f_0^2} \left( \frac{\partial^2 M'}{\partial x^2} + \frac{\partial^2 M'}{\partial y^2} + \frac{\partial^2 M'}{\partial \xi^2} \right)$. Using $v'_g = \frac{1}{f_0}\frac{\partial M'}{\partial x}$, we can write the zonal average as:

$$F_q = \frac{q_0}{f_0^3} \overline{ \frac{\partial M'}{\partial x} \left( \frac{\partial^2 M'}{\partial x^2} + \frac{\partial^2 M'}{\partial y^2} + \frac{\partial^2 M'}{\partial \xi^2} \right) }$$

Expanding the terms inside the average:

- $\overline{\frac{\partial M'}{\partial x}\frac{\partial^2 M'}{\partial x^2}} = \frac{1}{2} \overline{\frac{\partial}{\partial x} \left( \frac{\partial M'}{\partial x} \right)^2} = 0$ (The average of a perfect zonal derivative over a wave period is zero).

- $\overline{\frac{\partial M'}{\partial x}\frac{\partial^2 M'}{\partial y^2}} = \frac{\partial}{\partial y} \overline{ \left( \frac{\partial M'}{\partial x}\frac{\partial M'}{\partial y} \right) } - \overline{ \frac{\partial M'}{\partial y}\frac{\partial^2 M'}{\partial x \partial y} }$. The second term is $\frac{1}{2}\frac{\partial}{\partial x}(\frac{\partial M'}{\partial y})^2$, which averages to zero, leaving just the first term.

- $\overline{\frac{\partial M'}{\partial x}\frac{\partial^2 M'}{\partial \xi^2}} = \frac{\partial}{\partial \xi} \overline{ \left( \frac{\partial M'}{\partial x}\frac{\partial M'}{\partial \xi} \right) }$ (using the same product rule logic).

Putting it all together:

$$F_q = \frac{q_0}{f_0^3} \left[ \frac{\partial}{\partial y} \overline{ \left( \frac{\partial M'}{\partial x}\frac{\partial M'}{\partial y} \right) } + \frac{\partial}{\partial \xi} \overline{ \left( \frac{\partial M'}{\partial x}\frac{\partial M'}{\partial \xi} \right) } \right]$$

By defining this to match $F_q = -\frac{\partial E_y}{\partial y} - \frac{\partial E_\xi}{\partial \xi}$, we extract the components of the Eliassen-Palm flux vector $\mathbf{E}$:

$$E_y = -\frac{q_0}{f_0^3} \overline{ \frac{\partial M'}{\partial x}\frac{\partial M'}{\partial y} } \quad \text{and} \quad E_\xi = -\frac{q_0}{f_0^3} \overline{ \frac{\partial M'}{\partial x}\frac{\partial M'}{\partial \xi} }$$

#### (b) Further show that $E_y$ is proportional to the meridional flux of zonal momentum and that $E_\xi$ is proportional to the meridional flux of geopotential.

- Meridional flux of zonal momentum: We know $u'_g = -\frac{1}{f_0}\frac{\partial M'}{\partial y}$ and $v'_g = \frac{1}{f_0}\frac{\partial M'}{\partial x}$. The momentum flux is $\overline{u'_g v'_g} = -\frac{1}{f_0^2} \overline{ \frac{\partial M'}{\partial x}\frac{\partial M'}{\partial y} }$. Comparing this to $E_y$, we see $E_y = \frac{q_0}{f_0} \overline{u'_g v'_g}$, proving it is proportional to the meridional flux of zonal momentum

- Meridional flux of geopotential: Equation (10.38) gives $\frac{\partial M'}{\partial \xi} = -\frac{f_0 \Gamma_R}{N_R \theta_R} \Phi'$. Substituting this into $E_\xi$, we get $E_\xi = -\frac{q_0}{f_0^3} \overline{ (f_0 v'_g) \left( -\frac{f_0 \Gamma_R}{N_R \theta_R} \Phi' \right) } = \frac{q_0 \Gamma_R}{f_0 N_R \theta_R} \overline{v'_g \Phi'}$. This confirms $E_\xi$ is proportional to the meridional flux of geopotential $\overline{v'_g \Phi'}$. 

#### (c) Compute the Eliassen-Palm flux for the Eady mode.

The basic Eady model explicitly assumes a uniform zonal shear on an f-plane, and its mode has no variations in the $y$-direction ($\partial M' / \partial y = 0$).

- Because $\partial M'/\partial y = 0$, the $y$-component vanishes entirely: $E_y = 0$.
- Since the interior PV flux is zero ($F_q = 0$), the divergence of the EP flux must be zero: $\nabla \cdot \mathbf{E} = 0$. Since $E_y = 0$, it follows that $\frac{\partial E_\xi}{\partial \xi} = 0$.
- Therefore, $E_\xi$ is a non-zero constant with height throughout the fluid, driven purely by the poleward heat transport generated at the boundaries.
