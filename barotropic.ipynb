# Barotropic Instability

## Linearized Governing Equations

We begin by decomposing the fractional thickness perturbation, zonal wind, and potential vorticity into zonally symmetric base states and disturbance (primed) components:

- $\eta^{*} = \eta_{Z}(y) + \eta'$
- $u_{g} = U + u_{g}'$
- $q^{*} = q_{Z}(y) + q'$

The base zonal wind $U$ is in geostrophic balance with the base thickness perturbation:

$$U(y) = -f_{0}L_{R}^{2}\frac{d\eta_{Z}}{dy} \tag{1}$$

Geostrophic balance for the primed perturbation quantities is given by:

$$u_{g}' = -f_{0}L_{R}^{2}\frac{\partial\eta'}{\partial y} \tag{2}$$

$$v_{g}' = f_{0}L_{R}^{2}\frac{\partial\eta'}{\partial x} \tag{3}$$

The zonal flow part of the potential vorticity (PV) is defined as:

$$q_{Z} = q_{0}\left(L_{R}^{2}\frac{d^{2}\eta_{Z}}{dy^{2}} - \eta_{Z}\right) \tag{4}$$

The total geostrophic PV is $q_{g} = q_{A} + q_{Z} + q'$ , where the geographic PV (ignoring terrain) is:

$$q_{A} = q_{0}(1 + \beta y/f_{0}) \tag{5}$$

The perturbation PV is diagnosed using the relationship:

$$L_{R}^{2}\nabla^{2}\eta' - \eta' = q'/q_{0} \tag{6}$$

The full, non-linear potential vorticity evolution equation is:

$$\frac{\partial q'}{\partial t} + \left(U - f_{0}L_{R}^{2}\frac{\partial\eta'}{\partial y}\right)\frac{\partial q'}{\partial x} + f_{0}L_{R}^{2}\frac{\partial\eta'}{\partial x}\left(\frac{\partial(q_{A} + q_{Z} + q')}{\partial y}\right) = 0 \tag{7}$$

To linearize this equation, we assume the primed disturbance quantities are small and drop all products of primed terms (i.e., $-f_{0}L_{R}^{2}\frac{\partial\eta'}{\partial y}\frac{\partial q'}{\partial x}$ and $f_{0}L_{R}^{2}\frac{\partial\eta'}{\partial x}\frac{\partial q'}{\partial y}$). This yields:

$$\frac{\partial q'}{\partial t} + U\frac{\partial q'}{\partial x} + f_{0}L_{R}^{2}\left(\frac{d(q_{A}+q_{Z})}{dy}\right)\frac{\partial\eta'}{\partial x} = 0 \tag{8}$$

Substituting the diagnosed PV from Equation 6 into Equation 8 gives the linearized governing equation in terms of $\eta'$:

$${\color{red}\boxed{ \left(\frac{\partial}{\partial t} + U\frac{\partial}{\partial x}\right)(L_{R}^{2}\nabla^{2}\eta' - \eta') + f_{0}L_{R}^{2}\frac{dQ}{dy}\frac{\partial\eta'}{\partial x} = 0 }} \tag{9}$$

where the geographic plus zonal wind PV, scaled by $q_{0}$, is defined as:

$$Q(y) = \frac{q_{A} + q_{Z}}{q_{0}} = 1 + \frac{\beta y}{f_{0}} + L_{R}^{2}\frac{d^{2}\eta_{Z}}{dy^{2}} - \eta_{Z} \tag{10}$$

# Non-dimensionalization

To reduce clutter, we scale all lengths by the Rossby radius $L_{R}$ and all times by the inverse of the base Coriolis parameter $f_{0}^{-1}$. Applying this scaling to Equations 1, 2, and 3 yields the dimensionless geostrophic relations:

$$U(y) = -\frac{d\eta_{Z}}{dy}, \quad u_{g}' = -\frac{\partial\eta'}{\partial y}, \quad v_{g}' = \frac{\partial\eta'}{\partial x} \tag{11}$$

The dimensionless PV diagnostic equation simplifies to:

$$\nabla^{2}\eta' - \eta' = q'/q_{0} \tag{12}$$

Applying the temporal ($f_0^{-1}$) and spatial ($L_R$) scaling to Equation 9, the constants cancel out, yielding the dimensionless evolution equation:

$${\color{blue}\boxed{ \left(\frac{\partial}{\partial t} + U\frac{\partial}{\partial x}\right)(\nabla^{2}\eta' - \eta') + \frac{dQ}{dy}\frac{\partial\eta'}{\partial x} = 0 }} \tag{13}$$

Correspondingly, the ambient PV profile $Q(y)$ non-dimensionalizes to:

$$Q(y) = 1 + \beta y + \frac{d^{2}\eta_{Z}}{dy^{2}} - \eta_{Z} \tag{14}$$

# Interfacial Rossby Waves

We now consider an idealization where the PV gradient is large but confined to a tiny meridional interval at $y=0$. We express this gradient using the Dirac delta function $\delta(y)$:

$$\frac{dQ}{dy} = \Delta Q\delta(y) \tag{15}$$

where $\Delta Q$ is the discontinuity in $Q$ across the interface. Assuming plane wave solutions of the form $\exp[i(kx - \omega t)]$, the differential operators in Equation 13 become $\frac{\partial}{\partial t} \to -i\omega$ and $\frac{\partial}{\partial x} \to ik$. Substituting these and dividing by $ik$ gives:

$$-(c - U)\left(\frac{\partial^{2}}{\partial y^{2}} - k^{2} - 1\right)\eta' + \Delta Q\delta(y)\eta' = 0 \tag{16}$$

where $c = \omega/k$ is the wave's phase speed. For regions away from the interface ($y \neq 0$), the delta function term is zero, leaving:

$$\frac{\partial^{2}\eta'}{\partial y^{2}} - l^{2}\eta' = 0 \tag{17}$$

where we have defined the meridional decay scale:

$$l = (k^{2} + 1)^{1/2} \tag{18}$$

This second-order ordinary differential equation has exponential solutions:

$$\eta' \propto \exp(\pm ly) \tag{19}$$

To ensure the perturbation vanishes as $|y|\to \infty$, we construct a piecewise solution:

$$\eta' = \begin{cases} A_{+}\exp(-ly), & y > 0 \\ A_{-}\exp(ly), & y < 0 \end{cases} \tag{20}$$

To avoid an infinite zonal wind $u_{g}'$ at $y=0$, $\eta'$ must be continuous, meaning $A_{+} = A_{-} \equiv A$. 

To find the dispersion relation, we integrate Equation 16 across the interface from $-\epsilon$ to $\epsilon$. Approximating $U \approx U_0 = U(0)$ near the interface gives:

$$-(c - U_{0})\Delta\left(\frac{\partial\eta'}{\partial y}\right) + \Delta Q A = 0 \tag{21}$$

The jump in the derivative across $y=0$ is:

$$\Delta\left(\frac{\partial\eta'}{\partial y}\right) = \left(\frac{\partial\eta'}{\partial y}\right)_{\epsilon} - \left(\frac{\partial\eta'}{\partial y}\right)_{-\epsilon} = (-lA) - (lA) = -2lA \tag{22}$$

Substituting Equation 22 into Equation 21 and solving for $c$ yields the dispersion relation:

$${\color{green}\boxed{ c = \frac{\omega}{k} = U_{0} - \frac{\Delta Q}{2(k^{2} + 1)^{1/2}} }} \tag{23}$$

The physical solutions (taking the real part) are:

$$\eta^{*} = \eta_{Z}(y) + A\cos(kx - \omega t)\exp(-l|y|) \tag{24}$$
$$u_{g} = U(y) + Al\text{sgn}(y)\cos(kx - \omega t)\exp(-l|y|) \tag{25}$$
$$v_{g} = -Ak\sin(kx - \omega t)\exp(-l|y|) \tag{26}$$

To determine a consistent ambient flow, we require $dQ/dy = 0$ for $y \neq 0$. Applying this to $Q(y) = 1 + \beta y - dU/dy + U$ gives a differential equation for $U$:

$$\frac{d^{2}U}{dy^{2}} - U - \beta = 0 \tag{27}$$

Solving this while demanding boundedness at infinity and setting $U(0) = U_0$ gives:

$$U = (U_{0} + \beta)\exp(-|y|) - \beta \tag{28}$$

Integrating $U = -d\eta_Z/dy$ yields the base thickness (assuming continuity at $y=0$):

$$\eta_{Z}(y) = (U_{0} + \beta)\text{sgn}(y)[\exp(-|y|) - 1] + \beta y \tag{29}$$

Substituting these back into Equation 14 gives the ambient PV:

$$Q = 1 + (U_{0} + \beta)\text{sgn}(y) \tag{30}$$

### 1. Charney-Stern theorem for shallow water equations

#### (a) Starting with equation (9.12), assume $x-t$ dependence of the form $\exp[ik(x - ct)]$, divide by $c-U$, and then multiply the equation by $\eta'^*$, the complex conjugate of $\eta'$, and integrate in $y$ from $-\infty$ to $\infty$. 

Let $\eta' = \hat{\eta}(y)\exp[ik(x - ct)]$. The differential operators become $\frac{\partial}{\partial t} \to -ikc$ and $\frac{\partial}{\partial x} \to ik$. Substituting these into equation 9.12 gives:$$(-ikc + ikU)(\frac{d^2\hat{\eta}}{dy^2} - k^2\hat{\eta} - \hat{\eta}) + ik\frac{dQ}{dy}\hat{\eta} = 0$$Divide the entire equation by $ik(U - c)$:$$-(\frac{d^2\hat{\eta}}{dy^2} - k^2\hat{\eta} - \hat{\eta}) + \frac{dQ/dy}{U - c}\hat{\eta} = 0$$Next, multiply by the complex conjugate $\hat{\eta}^*$ and integrate over $y$ from $-\infty$ to $\infty$:$$\int_{-\infty}^{\infty} \left[ -\hat{\eta}^*\frac{d^2\hat{\eta}}{dy^2} + (k^2 + 1)|\hat{\eta}|^2 + \frac{dQ/dy}{U - c}|\hat{\eta}|^2 \right] dy = 0$$

#### (b) Do an integration by parts on the first term and assume that $\eta' \to 0$ as $y \to \pm\infty$.

Integrating the first term by parts yields:$$\int_{-\infty}^{\infty} -\hat{\eta}^*\frac{d^2\hat{\eta}}{dy^2} dy = \left[ -\hat{\eta}^* \frac{d\hat{\eta}}{dy} \right]_{-\infty}^{\infty} + \int_{-\infty}^{\infty} \frac{d\hat{\eta}^*}{dy} \frac{d\hat{\eta}}{dy} dy$$Since we assume the perturbation vanishes at infinity ($\hat{\eta} \to 0$ as $y \to \pm\infty$), the boundary evaluation term becomes zero. The integral simplifies to $\int_{-\infty}^{\infty} \left| \frac{d\hat{\eta}}{dy} \right|^2 dy$. Substituting this back gives:$$\int_{-\infty}^{\infty} \left[ \left| \frac{d\hat{\eta}}{dy} \right|^2 + (k^2 + 1)|\hat{\eta}|^2 + \frac{dQ/dy}{U - c}|\hat{\eta}|^2 \right] dy = 0$$

#### (c) Take the imaginary part of the equation and determine a necessary condition for the imaginary part of $c$ to be non-zero.

Let the phase speed be complex: $c = c_R + ic_I$. We can rewrite the denominator of the last term as:$$\frac{1}{U - c} = \frac{1}{(U - c_R) - ic_I} = \frac{(U - c_R) + ic_I}{(U - c_R)^2 + c_I^2}$$Taking only the imaginary part of the full integrated equation leaves:$$\int_{-\infty}^{\infty} \frac{c_I \frac{dQ}{dy}}{(U - c_R)^2 + c_I^2} |\hat{\eta}|^2 dy = 0$$If we assume the mode is unstable, then $c_I \neq 0$, and we can divide it out of the integral:$$\int_{-\infty}^{\infty} \frac{\frac{dQ}{dy}}{(U - c_R)^2 + c_I^2} |\hat{\eta}|^2 dy = 0$$For this integral to sum to zero, the integrand must take on both positive and negative values. Because both the denominator and $|\hat{\eta}|^2$ are strictly positive, the necessary condition for instability is that the ambient potential vorticity gradient $\frac{dQ}{dy}$ must change sign somewhere within the domain.

#### (d) Determine whether the modes discussed in sections 9.3 and 9.4 satisfy this condition.

- Section 9.3 (Interfacial Rossby waves): The potential vorticity gradient is $\frac{dQ}{dy} = \Delta Q\delta(y)$. Since $\Delta Q$ is a single constant jump across the interface, it does not change sign. The necessary condition is not met, meaning $c_I = 0$ and the mode is stable

- Section 9.4 (Barotropic instability): The potential vorticity gradient is $\frac{dQ}{dy} = \Delta Q[\delta(y + \xi) - \delta(y - \xi)]$. Because the delta functions are subtracted, $\frac{dQ}{dy}$ is positive at the $y = -\xi$ interface and negative at the $y = \xi$ interface. Since it changes sign, the Charney-Stern necessary condition is satisfied, allowing for the calculated instability.

### 4. Show that the potential vorticity flux in the y direction can be written in terms of the fluxes of momentum and fractional thickness perturbation... From the results of the last two problems, compute the flux of potential vorticity in the y direction in the case of the barotropic instability mode.

  $$L_R^2 \nabla^2 \eta' = \frac{1}{f_0}\frac{\partial v_g}{\partial x} - \frac{1}{f_0}\frac{\partial u_g}{\partial y}$$Substituting this into the flux equation gives:$$\overline{q' v_g} = q_0 \left( \frac{1}{f_0}\overline{v_g \frac{\partial v_g}{\partial x}} - \frac{1}{f_0}\overline{v_g \frac{\partial u_g}{\partial y}} - \overline{\eta' v_g} \right)$$Because we are averaging over a closed wavelength, $\overline{v_g \frac{\partial v_g}{\partial x}} = \frac{1}{2}\frac{\partial}{\partial x}(\overline{v_g^2}) = 0$.Furthermore, the geostrophic wind is non-divergent ($\frac{\partial u_g}{\partial x} + \frac{\partial v_g}{\partial y} = 0$).
