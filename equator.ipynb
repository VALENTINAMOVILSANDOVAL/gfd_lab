# Equatorial Dynamics

## Weak Temperature Gradient Approximation

To illustrate the reduced coupling between atmospheric levels near the equator, we begin with a quasi-geostrophic potential vorticity anomaly on an f-plane given by $q^*/q_0=\epsilon\delta(\theta-\theta_0)\sin(kx)$. The Montgomery potential anomaly, $M^*$, generated by this is found by solving the inversion equation:

$$\frac{\partial^2M^*}{\partial\theta^2}-m^2M^*=\frac{\epsilon N_R^2}{\Gamma_R^2}\delta(\theta-\theta_0)\sin(kx)\tag{1}$$

where $m$ is defined as:

$$m=\frac{kN_R}{f_0\Gamma_R}\tag{2}$$

Solving this differential equation yields an $M^*$ that decreases exponentially away from the $\theta=\theta_0$ level:

$$M^*=-\frac{\epsilon N_R^2}{2m\Gamma_R^2}\exp(-m|\theta-\theta_0|)\sin(kx)\tag{3}$$

The potential vorticity anomaly thus has a penetration depth equivalent to the decay length $m^{-1}$:

$${\color{blue}\boxed{\delta\theta_p=\frac{1}{m}=\frac{f_0\Gamma_R}{kN_R}}}\tag{4}$$

Because this penetration depth is proportional to $f_0$, the depth shrinks as one nears the equator. Another effect is the weakening of horizontal temperature gradients. The Boussinesq thermal wind equation in geometric coordinates is:

$$\nabla_hb=-f\mathbf{k}\times\frac{\partial\mathbf{v}_g}{\partial z}\tag{5}$$

For a given vertical shear, the horizontal buoyancy gradient gets smaller as $f$ drops. In the limit where these gradients are zero and the mean profile $\theta=\theta_0(z)$ is steady, the standard thermodynamic equation:

$$\frac{\partial\theta}{\partial t}+\mathbf{v}_h\cdot\nabla_h\theta+w\frac{\partial\theta}{\partial z}=S\tag{6}$$

reduces to a direct one-to-one balance between vertical velocity and convective heating:

$$w\frac{d\theta_0}{dz}=S\tag{7}$$

## Hadley and Monsoon Circulations

On an equatorial beta-plane, the zonally symmetric zonal momentum equation expressing the conservation of angular momentum is:

$$\frac{d(u-\beta y^2/2)}{dt}=\frac{dm}{dt}=F_x\tag{8}$$

where the related angular momentum parameter $m$ is:

$$m=u-\beta y^2/2\tag{9}$$

To see how this connects to actual angular momentum, $L=ru+r^2\Omega$, we substitute $r=a\cos\phi$ for an air parcel:

$$L=au\cos\phi+a^2\Omega-a^2\Omega\sin^2\phi\tag{10}$$

Applying the beta plane approximation ignoring Earth's curvature (except for the Coriolis parameter), we set $\cos\phi\approx1$ and $y=\phi a$. By defining the equatorial value of beta:

$$\beta=\left(\frac{df}{dy}\right)_{\phi=0}=2\Omega\left(\frac{d\sin\phi}{d\phi}\right)_{\phi=0}\frac{d\phi}{dy}=\frac{2\Omega}{a}\tag{11}$$

we can linearize the relationship to show how actual angular momentum relates to $m$:

$${\color{red}\boxed{L\approx a(u-\beta y^2/2)+a^2\Omega=am+a^2\Omega}}\tag{12}$$

## Matsuno Modes (Shallow Water Equations)

Analyzing small perturbations from rest in the tropics utilizes the linearized shallow water equations on an equatorial beta plane:

$$\frac{\partial u}{\partial t}-\beta yv+c^2\frac{\partial\eta}{\partial x}=0\tag{13}$$$$\frac{\partial v}{\partial t}+\beta yu+c^2\frac{\partial\eta}{\partial y}=0\tag{14}$$$$\frac{\partial\eta}{\partial t}+\frac{\partial u}{\partial x}+\frac{\partial v}{\partial y}=0\tag{15}$$

To non-dimensionalize, length is measured in units of the equatorial Rossby radius $L_R=(c/\beta)^{1/2}$ and time in units of $T=(\beta c)^{-1/2}$, enforcing $\beta\rightarrow1$ and $c\rightarrow1$. Assuming a plane wave in the $x$ direction of the form $\exp[i(kx-\omega t)]$ results in the core non-dimensional system:

$${\color{green}\boxed{-i\omega u-yv+ik\eta=0}}\tag{16}$$$$-i\omega v+yu+\frac{\partial\eta}{\partial y}=0\tag{17}$$$$-i\omega\eta+iku+\frac{\partial v}{\partial y}=0\tag{18}$$

## Kelvin Waves

For solutions where $v=0$, equations 16 and 18 become purely algebraic:

$$-i\omega u+ik\eta=0\tag{19}$$
$$-i\omega\eta+iku=0\tag{20}$$

This provides the dispersion relation for equatorial Kelvin waves:

$$\omega=\pm k\tag{21}$$

and the corresponding polarization relation:

$$u=\pm\eta\tag{22}$$

Substituting $v=0$ into equation 17 leaves:

$$\frac{\partial\eta}{\partial y}\pm y\eta=0\tag{23}$$

To prevent $\eta$ from blowing up as $y\rightarrow\pm\infty$, only the minus sign is physically acceptable, yielding a Gaussian form about the equator:

$$\eta=\eta_0\exp(-y^2/2)\tag{24}$$

Compiling these results yields the complete functional form for the Kelvin wave:

$${\color{purple}\boxed{\begin{aligned}u&=\eta_0\exp(-y^2/2)\cos(kx-\omega t)\v &=0\\\eta&=\eta_0\exp(-y^2/2)\cos(kx-\omega t)\end{aligned}}}\tag{25}$$

## Equatorial Rossby and Gravity Waves

When $v\neq0$, equations 16 and 18 can be solved algebraically for $u$ and $\eta$ in terms of $v$:

$$u=\frac{-i\omega yv+ik(\partial v/\partial y)}{k^2-\omega^2}\tag{26}$$

$$\eta=\frac{i\omega(\partial v/\partial y)-ikyv}{k^2-\omega^2}\tag{27}$$

Substituting these back into equation 17 yields a second-order differential equation:

$$\frac{\partial^2v}{\partial y^2}+(\omega^2-k^2-k/\omega-y^2)v=0\tag{28}$$

Because this shares the form of the Schrödinger equation for a harmonic oscillator, it only approaches zero at $y\rightarrow\pm\infty$ if:

$$\omega^2-k^2-k/\omega=2n+1,\quad n=0,1,2,...\tag{29}$$

Multiplying by $\omega$ reorganizes this into a cubic dispersion relation:

$$\omega^3-(k^2+2n+1)\omega-k=0\tag{30}$$

The solutions involve Hermite polynomials:

$$v=v_0\exp(-y^2/2)H_n(y)\tag{31}$$

$$H_0(y)=1,\quad H_1(y)=2y,\quad H_2(y)=4y^2-2\tag{32}$$

For $k^2\gg1$, we can extract approximate solutions. If $\omega^2$ is comparable to $k^2$, we can drop the $-k$ term:

$$\omega^3-(k^2+2n+1)\omega\approx0\tag{33}$$

This gives the inertia-gravity wave approximation:

$$\omega\approx(k^2+2n+1)^{1/2}\tag{34}$$

Alternatively, if $|\omega|\ll|k|$, dropping the $\omega^3$ term gives the Rossby wave approximation:

$$\omega\approx-\frac{k}{k^2+2n+1}\tag{35}$$

For the specific case of an $n=1$ equatorial Rossby wave, the functional form is:

$${\color{orange}\boxed{\begin{aligned}u&=\eta_0\left[\frac{\omega y^2-k(1-y^2)}{k^2-\omega^2}\right]\exp(-y^2/2)\sin(kx-\omega t)\v&=\eta_0y\exp(-y^2/2)\cos(kx-\omega t)\\\eta&=\eta_0\left[\frac{ky^2-\omega(1-y^2)}{k^2-\omega^2}\right]\exp(-y^2/2)\sin(kx-\omega t)\end{aligned}}}\tag{36}$$

## Mixed Rossby-Gravity Wave

For the distinct $n=0$ case, the exact dispersion relation is:

$$\omega^3-(k^2+1)\omega-k=0\tag{37}$$

This naturally factors into:

$$(\omega+k)(\omega^2-k\omega-1)=0\tag{38}$$

This produces the following mathematical roots:

$$\omega=-k\tag{39}$$

$$\omega=\frac{k}{2}\pm\left[\left(\frac{k}{2}\right)^2+1\right]^{1/2}\tag{40}$$

The $\omega=-k$ solution is physically spurious because it corresponds to an unphysical westward-moving Kelvin mode. Additionally, the negative sign on the quadratic root simply corresponds to the positive sign evaluated at $-k$. Thus, the definitive dispersion relation for the mixed Rossby-gravity mode uses the positive branch:

$$\omega=\frac{k}{2}+\left[\left(\frac{k}{2}\right)^2+1\right]^{1/2}\tag{41}$$

The full functional form for this mixed mode is:

$${\color{teal}\boxed{\begin{aligned}u&=-\eta_0\omega y\exp(-y^2/2)\sin(kx-\omega t)\v&=\eta_0\exp(-y^2/2)\cos(kx-\omega t)\\\eta&=-\eta_0\omega y\exp(-y^2/2)\sin(kx-\omega t)\end{aligned}}}\tag{42}$$

## Matsuno Modes and Convection

To couple Matsuno modes with deep convection, we recall that shallow water modes traveling at speed $c$ are isomorphic to hydrostatic gravity waves with a vertical wavenumber defined by the Brunt-Väisälä frequency $N$:

$${\color{brown}\boxed{m=\frac{N}{c}}}\tag{43}$$

### 1. Compute the penetration depth $\delta\theta_{p}$ for a disturbance of horizontal wavelength 3000 km at latitudes $5^{\circ}$ and $50^{\circ}$. Assume typical tropical atmospheric values $\Gamma_{R}=d\theta/dz=5\times10^{-3}\text{ K m}^{-1}$ and $N=10^{-2}\text{ s}^{-1}$.

The penetration depth $\delta\theta_{p}$ is given by $\delta\theta_{p}=\frac{f_{0}\Gamma_{R}}{kN_{R}}$.
First, we compute the Coriolis parameter $f_{0}=2\Omega\sin\phi$ for both latitudes, using Earth's rotation rate $\Omega\approx7.292\times10^{-5}\text{ s}^{-1}$:

- At $\phi=5^{\circ}$: $f_{0}\approx1.271\times10^{-5}\text{ s}^{-1}$
- At $\phi=50^{\circ}$: $f_{0}\approx1.117\times10^{-4}\text{ s}^{-1}$

Next, we compute the horizontal wavenumber $k=2\pi/\lambda$:

$$k=\frac{2\pi}{3\times10^{6}\text{ m}}\approx2.094\times10^{-6}\text{ m}^{-1}$$

Substituting these parameters yields the penetration depths for both latitudes:

- At $5^{\circ}$: $\delta\theta_{p}=\frac{(1.271\times10^{-5})(5\times10^{-3})}{(2.094\times10^{-6})(10^{-2})}\approx3.03\text{ K}$

- At $50^{\circ}$: $\delta\theta_{p}=\frac{(1.117\times10^{-4})(5\times10^{-3})}{(2.094\times10^{-6})(10^{-2})}\approx26.67\text{ K}$

### 2. Consider steady, zonally symmetric meridional flow with linear friction $F_{x}=-\lambda u$ on a beta-plane. The meridional velocity $v$ is assumed to be constant. 

#### (a) Show that the angular momentum obeys the equation $v\frac{dm}{dy}+\lambda m=-\lambda\beta y^{2}/2$. 

The zonal momentum equation with linear friction is $\frac{dm}{dt}=-\lambda u$. For a steady, zonally symmetric flow, the total time derivative reduces to $\frac{d}{dt}=v\frac{d}{dy}$. Angular momentum is defined as $m=u-\beta y^{2}/2$, meaning $u=m+\beta y^{2}/2$. Substituting this into the momentum equation:

$$v\frac{dm}{dy}=-\lambda(m+\beta y^{2}/2)$$

$$v\frac{dm}{dy}+\lambda m=-\lambda\beta y^{2}/2$$

#### (b) Show that if $y$ is measured in units of $v/\lambda$ and if $m$ is measured in units of $\beta v^{2}/\lambda^{2}$, this equation simplifies to $\frac{dm}{dy}+m=-y^{2}/2$.

Let $y=Y(v/\lambda)$ and $m=M(\beta v^{2}/\lambda^{2})$. Substituting these into the dimensional equation:

$$v\frac{d}{d(Y v/\lambda)}\left(M\frac{\beta v^{2}}{\lambda^{2}}\right)+\lambda\left(M\frac{\beta v^{2}}{\lambda^{2}}\right)=-\lambda\beta\frac{(Y v/\lambda)^{2}}{2}$$

$$\lambda\frac{\beta v^{2}}{\lambda^{2}}\frac{dM}{dY}+\lambda\frac{\beta v^{2}}{\lambda^{2}}M=-\frac{\beta v^{2}}{2\lambda}Y^{2}$$

Dividing both sides by the scaling factor $\frac{\beta v^{2}}{\lambda}$ leaves the dimensionless equation:

$$\frac{dM}{dY}+M=-Y^{2}/2$$

#### (c) Solve the above equation, noting that it has homogeneous and inhomogeneous parts to the solution. Hint: Try $m_{I}=Ay^{2}+By+C$ for the inhomogeneous part. (You know what the homogeneous solution is!)

The homogeneous solution is $M_{H}=C_{0}e^{-Y}$. Using the hint, we assume an inhomogeneous solution $M_{I}=AY^{2}+BY+C$, meaning $\frac{dM_{I}}{dY}=2AY+B$. Substituting into the dimensionless differential equation: 

$$(2AY+B)+(AY^{2}+BY+C)=-Y^{2}/2$$

$$AY^{2}+(2A+B)Y+(B+C)=-Y^{2}/2$$

Matching polynomial coefficients gives $A=-1/2$, $2A+B=0\implies B=1$, and $B+C=0\implies C=-1$. Therefore, the complete general solution in dimensionless form is:

$$M(Y)=C_{0}e^{-Y}-\frac{Y^{2}}{2}+Y-1$$

#### (d) Use this result to plot the zonal wind in steady, cross-equatorial flow in the boundary layer. Compare in particular the behavior of the zonal wind to that which would occur if there were no friction. As an example, assume an initial $y=-1500\text{ km}$, an initial $u=-10\text{ m s}^{-1}$, $v=5\text{ ms}^{-1}$, and $\lambda=10^{-5}\text{ s}^{-1}$. At what approximate value of $y$ does the zonal wind switch from easterly to westerly?

First, we compute the dimensionless initial conditions. We assume $\beta\approx2.29\times10^{-11}\text{ m}^{-1}\text{s}^{-1}$.
The relevant scales are $L_{y}=v/\lambda=500\text{ km}$ and $M_{0}=\beta v^{2}/\lambda^{2}\approx5.725\text{ m s}^{-1}$.
The initial condition is set at $y_{0}=-1500\text{ km}$, yielding a dimensionless initial position $Y_{0}=-3$.
The initial angular momentum is $m_{0}=u_{0}-\beta y_{0}^{2}/2=-10-(2.29\times10^{-11})(-1.5\times10^{6})^{2}/2=-35.76\text{ m s}^{-1}$.
The dimensionless initial momentum is $M(-3)=-35.76/5.725=-6.246$.
Solving for the integration constant $C_{0}$:  

$$C_{0}e^{3}-(-3)^{2}/2+(-3)-1=-6.246$$

$$C_{0}e^{3}-8.5=-6.246\implies C_{0}=2.254e^{-3}\approx0.112$$

The dimensionless zonal wind is derived from $U(Y)=M(Y)+Y^{2}/2=C_{0}e^{-Y}+Y-1$. To find where the wind switches from easterly ($U<0$) to westerly ($U>0$), we set $U(Y)=0$:

$$0.112e^{-Y}+Y-1=0$$

Numerically evaluating this root yields $Y\approx0.957$.
Converting this to dimensional space, the switch occurs at $y=Y\cdot L_{y}\approx0.957\times500\text{ km}\approx478\text{ km}$.

- Comparison: Without friction, $m$ is perfectly conserved at $m=-35.76\text{ m s}^{-1}$. The wind would switch directions when $u=m+\beta y^{2}/2=0$, meaning $y=\sqrt{2|m|/\beta}\approx1767\text{ km}$. Friction bleeds off the westerly momentum much faster, causing the zonal wind to reverse significantly closer to the equator (478 km vs 1767 km).

### 3. Compute the group velocities for the Rossby and mixed Rossby-gravity modes. Determine the range of dimensionless wavenumbers for which each has eastward and westward group velocities.

Group velocity is defined as the partial derivative of the frequency with respect to the wavenumber: $c_{g}=\frac{\partial\omega}{\partial k}$.


- Rossby Mode:

Using the approximate dispersion relation $\omega\approx-\frac{k}{k^{2}+2n+1}$:

$$c_{g,R}=\frac{\partial}{\partial k}\left(-\frac{k}{k^{2}+2n+1}\right)=-\frac{(k^{2}+2n+1)(1)-k(2k)}{(k^{2}+2n+1)^{2}}=\frac{k^{2}-(2n+1)}{(k^{2}+2n+1)^{2}}$$

Eastward group velocity ($c_{g,R}>0$): This occurs when the numerator is positive, $k^{2}>2n+1$, or $|k|>\sqrt{2n+1}$.

Westward group velocity ($c_{g,R}<0$): This occurs when the numerator is negative, $k^{2}<2n+1$, or $|k|<\sqrt{2n+1}$.

- Mixed Rossby-Gravity Mode:

Using the exact dispersion relation $\omega=\frac{k}{2}+\left[\left(\frac{k}{2}\right)^{2}+1\right]^{1/2}$:

$$c_{g,MRG}=\frac{\partial\omega}{\partial k}=\frac{1}{2}+\frac{1}{2\sqrt{\frac{k^{2}}{4}+1}}\left(\frac{2k}{4}\right)=\frac{1}{2}\left(1+\frac{k}{\sqrt{k^{2}+4}}\right)$$

Because $\sqrt{k^{2}+4}>|k|$ for all real numbers $k$, the fractional term $\frac{k}{\sqrt{k^{2}+4}}$ is strictly bounded between $-1$ and $1$. Consequently, the entire expression in the parentheses is always positive.

Eastward group velocity ($c_{g,MRG}>0$): Occurs for all $k$

Westward group velocity: Never occurs for this mode.

### 4. For the Kelvin wave compute the relative vorticity and divergence $\zeta=\frac{\partial v}{\partial x}-\frac{\partial u}{\partial y}$, $\Delta=\frac{\partial u}{\partial x}+\frac{\partial v}{\partial y}$. Compute the ratio $|\Delta|/|\zeta|$ where the absolute value sign indicates the magnitude of the quantity at fixed $y$ and $k$, obtained by setting the multiplying cosine or sine to unity. Sketch in the $k-y$ plane where this ratio is greater and less than unity.

For the equatorial Kelvin wave, the functional form is $u=\eta_{0}\exp(-y^{2}/2)\cos(kx-\omega t)$, $v=0$, with $\omega=k$.
Relative vorticity ($\zeta$):

$$\zeta=\frac{\partial(0)}{\partial x}-\frac{\partial}{\partial y}\left[\eta_{0}\exp(-y^{2}/2)\cos(kx-\omega t)\right]=\eta_{0}y\exp(-y^{2}/2)\cos(kx-\omega t)$$

Divergence ($\Delta$):

$$\Delta=\frac{\partial}{\partial x}\left[\eta_{0}\exp(-y^{2}/2)\cos(kx-\omega t)\right]+\frac{\partial(0)}{\partial y}=-\eta_{0}k\exp(-y^{2}/2)\sin(kx-\omega t)$$

Magnitudes at fixed $y$ and $k$ (setting trig functions to $1$):

$$|\zeta|=|\eta_{0}y\exp(-y^{2}/2)|$$

$$|\Delta|=|-\eta_{0}k\exp(-y^{2}/2)|$$

The divergence-vorticity ratio is:

$$\frac{|\Delta|}{|\zeta|}=\frac{|\eta_{0}k\exp(-y^{2}/2)|}{|\eta_{0}y\exp(-y^{2}/2)|}=\left|\frac{k}{y}\right|$$

Regions in the $k-y$ plane:

- The ratio is greater than unity ($|\Delta|/|\zeta|>1$) in the domains where $|k|>|y|$. In a sketch, this represents the wedges bounded by the lines $k=y$ and $k=-y$ that contain the horizontal $k$-axis.

- The ratio is less than unity ($|\Delta|/|\zeta|<1$) in the domains where $|k|<|y|$. In a sketch, this represents the wedges containing the vertical $y$-axis.

### 5. Repeat problem 4 for the case of the mixed Rossby-gravity wave, except plot the divergence-vorticity ratio as a function of $k$ only for $y=1/2$. Explicitly use the dispersion relation to eliminate $\omega$ in favor of $k$. Pay particular attention to the limits where $k\gg1$ and $k\ll-1$.

For the mixed Rossby-gravity wave: $u=-\eta_{0}\omega y\exp(-y^{2}/2)\sin(kx-\omega t)$ and $v=\eta_{0}\exp(-y^{2}/2)\cos(kx-\omega t)$.
Relative vorticity ($\zeta$):

$$\frac{\partial v}{\partial x}=-\eta_{0}k\exp(-y^{2}/2)\sin(kx-\omega t)$$

$$\frac{\partial u}{\partial y}=-\eta_{0}\omega(1-y^{2})\exp(-y^{2}/2)\sin(kx-\omega t)$$

$$\zeta=\eta_{0}[\omega(1-y^{2})-k]\exp(-y^{2}/2)\sin(kx-\omega t)$$

Divergence ($\Delta$):

$$\frac{\partial u}{\partial x}=-\eta_{0}\omega k y\exp(-y^{2}/2)\cos(kx-\omega t)$$

$$\frac{\partial v}{\partial y}=-\eta_{0}y\exp(-y^{2}/2)\cos(kx-\omega t)$$

$$\Delta=-\eta_{0}y(\omega k+1)\exp(-y^{2}/2)\cos(kx-\omega t)$$

Evaluating magnitudes exclusively at $y=1/2$:

$$|\zeta|_{y=1/2}=\left|\eta_{0}\left(\frac{3}{4}\omega-k\right)e^{-1/8}\right|$$

$$|\Delta|_{y=1/2}=\left|\eta_{0}\left(\frac{1}{2}(\omega k+1)\right)e^{-1/8}\right|$$

The divergence-vorticity ratio evaluated at $y=1/2$ is:

$$R(k)=\frac{|\Delta|}{|\zeta|}=\frac{\left|\frac{1}{2}(\omega k+1)\right|}{\left|\frac{3}{4}\omega-k\right|}=\frac{|2(\omega k+1)|}{|3\omega-4k|}$$

We substitute the dispersion relation $\omega=\frac{k}{2}+\left[\left(\frac{k}{2}\right)^{2}+1\right]^{1/2}$  to evaluate the asymptotic limits

- Limit as $k\gg1$: The high-frequency mode behaves like an inertia-gravity wave, so $\omega\approx k$. Substituting this approximation:

$$R\approx\frac{|2(k^{2}+1)|}{|3k-4k|}\approx\frac{2k^{2}}{|-k|}=2|k|$$

As $k\rightarrow\infty$, the ratio $R\rightarrow\infty$. Divergence completely dominates relative vorticity.

- Limit as $k\ll-1$: The low-frequency mode acts like a Rossby wave. Using the Taylor expansion for large magnitude negative $k$, the root approximates to $-\frac{k}{2}\left(1+2/k^{2}\right)$. This simplifies the dispersion relation to $\omega\approx-\frac{1}{k}$. Substituting $\omega=-1/k$ into the ratio numerator:

$$2(\omega k+1)=2\left(-\frac{1}{k}k+1\right)=2(-1+1)=0$$

The denominator approaches $|3(-1/k)-4k|\approx4|k|$, which is non-zero. Because the numerator collapses, as $k\rightarrow-\infty$, the ratio $R\rightarrow0$. Relative vorticity completely dominates divergence.
