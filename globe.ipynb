# Global Circulations

## Simple Model of Global Radiation Balance

To determine the earth's global radiative temperature, we must equate the incoming solar energy with the outgoing thermal radiation.

- Incoming Energy:

The solar flux at the earth's orbit is $F_{s}$. A fraction of this, $A$, is reflected, meaning the earth absorbs $F_{s}(1-A)$. We multiply this by the earth's projected area, $\pi R^{2}$, to get the total absorbed energy.

- Outgoing Energy:

The outflow of thermal radiation is modeled as a black body, which radiates over the earth's entire surface area, $4\pi R^{2}$, yielding $4\pi R^{2}\sigma T_{rad}^{4}$.

- Equilibrium:

Equating the incoming and outgoing energy gives the global radiation balance:

$$\pi R^{2}F_{s}(1-A)=4\pi R^{2}\sigma T_{rad}^{4} \tag{1}$$

Solving for Temperature: By canceling out $\pi R^{2}$ and solving for $T_{rad}$, we arrive at the earth's radiative temperature:

$${\color{blue}\boxed{T_{rad}=\left[\frac{F_{s}(1-A)}{4\sigma}\right]^{1/4}}} \tag{2}$$

## Latitudinal Energy Transfer

Because solar radiation is not deposited uniformly, we must also calculate the local equilibrium temperature at a specific latitude $\phi$ during the equinox.

- Area Adjustments: The actual surface area of a latitudinal strip is $\delta S=2\pi R\cos\phi\cdot R\delta\phi$. However, the projected area as seen from the sun is $\delta S_{p}=2R\cos\phi\cdot R\cos\phi\delta\phi$.

- Energy Balance by Latitude: Assuming an albedo $A$, the energy balance for this specific strip is $F_{s}(1-A)\delta S_{p}=\sigma T_{rad}^{4}\delta S$.

Substituting the area formulas into the balance equation and solving for $T_{rad}(\phi)$ gives:

$${\color{teal}\boxed{T_{rad}(\phi)=\left[\frac{F_{s}(1-A)\cos\phi}{\pi\sigma}\right]^{1/4}}} \tag{3}$$

## Mass Continuity and Streamfunction

To define the mean meridional circulation in the troposphere (like the Hadley cell), we utilize a streamfunction derived from mass continuity.

- Continuity Equation: In pressure coordinates, the zonally symmetric mass continuity equation for the atmosphere is:

$$\frac{1}{a}\frac{\partial v\cos\phi}{\partial\phi}+\frac{\partial\omega}{\partial p}=0 \tag{4}$$

To satisfy this continuity equation automatically, we can define a streamfunction $\psi$ such that the velocity components $v$ and $\omega$ are given by:

$${\color{purple}\boxed{v=\frac{a}{\cos\phi}\frac{\partial\psi}{\partial p}, \quad \omega=-\frac{\partial\psi}{\partial\phi}}} \tag{5}$$

## The Ferrel Cell and Vorticity Transport

We need to analyze the flux form of the vorticity equation in the presence of friction.

In spherical and pressure coordinates, the vorticity equation is:

$$a\cos\phi\frac{\partial\zeta_{a}}{\partial t}+\frac{\partial}{\partial\lambda}\left[u\zeta_{a}+\frac{\partial v}{\partial p}\omega-F_{y}\right]+\frac{\partial}{\partial\phi}\left[\left(v\zeta_{a}-\frac{\partial u}{\partial p}\omega+F_{x}\right)\cos\phi\right]=0 \tag{6}$$

Taking the zonal and time average (denoted by an overbar), the derivatives with respect to $t$ and $\lambda$ vanish. This simplifies the equation to:

$$\frac{\partial}{\partial\phi}\left[\left(\overline{v\zeta_{a}}-\overline{\frac{\partial u}{\partial p}\omega}+\overline{F_{x}}\right)\cos\phi\right]=0 \tag{7}$$

Integrating this equation yields a constant, which must be zero to prevent a delta function vorticity source (a non-zero meridional flux) at the poles. Therefore:

$$\overline{v\zeta_{a}}-\overline{\frac{\partial u}{\partial p}\omega}+\overline{F_{x}}=0 \tag{8}$$

We split the fields into zonal means and eddy deviations (e.g., $v=\overline{v}+v'$), keeping in mind that the average of a deviation is zero. Substituting this decomposition into equation 8 gives the final partitioned balance of inviscid symmetric ($\overline{F_{s}}$), inviscid eddy ($\overline{F_{e}}$), and viscous ($\overline{F_{x}}$) transports:

$${\color{brown}\boxed{\left\{\overline{v}\overline{\zeta_{g}}-\frac{\partial\overline{u}}{\partial p}\overline{\omega}\right\}+\left\{\overline{v'\zeta_{a}'}-\overline{\frac{\partial u'}{\partial p}\omega'}\right\}+\overline{F_{x}}=\overline{F_{s}}+\overline{F_{e}}+\overline{F_{x}}=0}} \tag{9}$$

## Rossby Waves and Momentum Flux

The maintenance of the polar jet is driven by the meridional convergence of momentum due to radiated Rossby waves.

### Dispersion Relation: 

For Boussinesq Rossby waves on a beta-plane with a constant mean wind $U$, the dispersion relation is:

$$\omega=kU-\frac{\beta k}{k^{2}+l^{2}+m^{2}} \tag{10}$$

Differentiating the dispersion relation with respect to $l$ yields the meridional group velocity:

$$\frac{\partial\omega}{\partial l}=\frac{2\beta kl}{(k^{2}+l^{2}+m^{2})^{2}} \tag{11}$$

The resulting meridional convergence of zonal momentum translates to an effective force accelerating the jet. This relationship between the momentum flux and the eddy-induced vorticity flux is approximated by:

$${\color{magenta}\boxed{\overline{v'\zeta_{a}'}\approx-\frac{\partial\overline{u'v'}}{\partial y}}} \tag{12}$$
