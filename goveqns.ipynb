# Geophysical Fluid Dynamics

When we study fluid dynamics, we are interested in understanding how fluids move and evolve over time. But what exactly is Geophysical Fluid Dynamics (GFD)? GFD is the application of fluid dynamics to naturally occurring fluids on Earth and other planets, such as the atmosphere and the oceans. Since these fluids are part of the Earth system, they are called *geophysical fluids*. By studying their motion, we can better understand important phenomena such as weather patterns, ocean currents, and climate. Two key features that make geophysical fluids unique are the rotation of the Earth, which affects fluid motion, and stratification, which refers to the layering of fluids with different properties. In this series of notebooks, we will introduce the main concepts of GFD, combining physical intuition, mathematical derivations, and simple examples to help build a solid understanding of the subject.

## 1. Governing Equations

<div style="border-left: 5px solid #1f77b4; background-color: #eaf4ff; padding: 12px; border-radius: 5px;">

To study the motion of the atmosphere and oceans, we use a set of equations called the **governing equations** (or **constituent equations**). These equations describe how **mass**, **momentum**, and **other properties of a fluid**, such as temperature and pressure, change over time. Unlike classical mechanics, where we follow the motion of a particular object, in fluid dynamics we usually study a fixed region in space, called a **control volume**, and analyze how fluid moves through it. Because fluid can enter and leave this region, carrying mass, energy, and momentum across its boundaries, fluid dynamics is treated as an **open-system problem**. 

In this notebook, we will explore both the physical meaning and the mathematical derivation of the governing equations used in GFD.

</div>

<hr style="border:none; border-top:1px solid #888;">

## Mass Continuity

<div style="border-left: 5px solid #1f77b4; background-color: #eaf4ff; padding: 12px; border-radius: 5px;">

The mass continuity equation is based on the **principle of mass conservation**, which states that the rate of change of mass within a control volume is equal to the net mass flux into or out of that control volume.
</div>

The most common forms are outlined below:

### Flux Form

Let $\rho = \rho(x, t)$ be the fluid mass density and $v = v(x, t)$ be the fluid velocity. For a stationary control volume $V$, the total mass $M$ contained within that volume is the integral of the density:

$$
M = \int \rho dV \tag{1.1}
$$

Because the control volume is fixed in space, the rate at which the total mass changes over time depends entirely on how the density changes locally within that volume. This is expressed using the partial derivative with respect to time:

$$
\frac{dM}{dt} = \int \frac{\partial\rho}{\partial t} dV \tag{1.2}
$$

<br>

Using the surface area element $dA$ and an outward-pointing unit vector normal $n$, the mass flux ($\dot{m}$) across a surface is given, according to the principle of mass conservation, by:

$$
\frac{dM}{dt} = -\oint \rho v \cdot n dA \tag{1.3}
$$

To relate the surface flux back to the volume, we apply the divergence theorem to the closed surface integral. This transforms the flux across the surface into the divergence of the mass flux vector field over $V$:

$$
\oint \rho v \cdot n dA = \int \nabla \cdot (\rho v) dV \tag{1.4}
$$

Equating the two expressions for the time rate of change of mass and replacing the surface integral with the divergence theorem result yields:

$$
\int \frac{\partial\rho}{\partial t} dV = -\int \nabla \cdot (\rho v) dV \tag{1.5}
$$

By bringing all terms to the left side, we can combine everything under a single volume integral:

$$
\int \left( \frac{\partial\rho}{\partial t} + \nabla \cdot (\rho v) \right) dV = 0 \tag{1.6}
$$

Since this result is true for all control volumes, the integrand must be zero, resulting in the flux form of the mass continuity equation:

$$
\boxed{\frac{\partial\rho}{\partial t} + \nabla \cdot (\rho v) = 0} \tag{1.7}
$$

### Advective Form

Using the product rule to expand the divergence of the scalar-vector product in the flux form, we obtain:

$$
\nabla \cdot (\rho v) = v \cdot \nabla\rho + \rho \nabla \cdot v \tag{1.8}
$$

Substituting this expansion back into the flux equation gives:

$$
\frac{\partial\rho}{\partial t} + v \cdot \nabla\rho + \rho \nabla \cdot v = 0 \tag{1.9}
$$

To interpret the first two terms physically, we shift our perspective from a stationary control volume to a **fluid parcel**, a continuously identifiable volume element moving with the fluid flow. If we define the position of this moving parcel as $x$, we can observe the total rate of change of its density over time. Using the chain rule, this total derivative (often called the material or parcel derivative) is:

$$
\frac{d\rho}{dt} = \frac{\partial\rho}{\partial t} + \frac{dx}{dt} \cdot \nabla\rho \tag{1.10}
$$

Because the rate of change of the parcel's position, $dx/dt$, is exactly the fluid velocity $v$, we can rewrite this as:

$$
\frac{d\rho}{dt} = \frac{\partial\rho}{\partial t} + v \cdot \nabla\rho \tag{1.11}
$$

Substituting the material derivative back into our expanded continuity equation gives:

$$
\frac{d\rho}{dt} + \rho \nabla \cdot v = 0 \tag{1.12}
$$

Finally, dividing the entire equation by $\rho$ yields the standard advective form:

$$
\boxed{\frac{1}{\rho}\frac{d\rho}{dt} + \nabla \cdot v = 0} \tag{1.13}
$$

## Momentum Equation

<div style="border-left: 5px solid #1f77b4; background-color: #eaf4ff; padding: 12px; border-radius: 5px;">

The momentum equation is based on the **principle of conservation of momentum**, which states that the rate of change of momentum in a control volume plus the net flux of momentum out of the control volume equals the sum of forces acting on the fluid. 
</div>

The most common forms are outlined below:

### Flux Form

First, the time rate of change of momentum in the control volume is:

$$
\frac{d}{dt}\int\rho vdV=\int\frac{\partial\rho v}{\partial t}dV \tag{1.14}
$$

Next, the bulk flow of momentum out of the control volume is the closed area integral of this over the surface of the control volume:

$$
\oint\rho vv\cdot ndA=\int\nabla\cdot(\rho vv)dV \tag{1.15}
$$

Momentum is not conserved in a non-isolated system; the source of momentum is simply the force. We divide the total force applied to a fluid into two parts, body forces such as gravity, and short-range, inter-molecular forces. The total body force on the control volume is:

$$
F_{B}=\int\rho BdV \tag{1.16}
$$

The short-range forces are represented in terms of a stress tensor $T$. The force on the control volume due to short-range forces acting across the surface of the volume is represented by the surface integral:

$$
F_{T}=\oint tdA=\oint T\cdot ndA=\int\nabla\cdot TdV \tag{1.17}
$$

where $n$ is the outward unit normal on the surface of the control volume.

<br>

Both air and water are Newtonian fluids, which means that the stress tensor has parts related to pressure $p$, fluid strain rate $D$, and the trace of the strain rate $Tr(D)$. The effects of the $Tr(D)I$ term are of little interest in geophysical fluid dynamics, and for this reason this term is generally ignored. 

Putting all of this together and applying the same arguments as in the discussion of the mass continuity equation results in the flux form of the Navier-Stokes equation:

$$
\boxed{ \frac{\partial\rho v}{\partial t}+\nabla\cdot(\rho vv)=\rho B-\nabla p+\mu\nabla^{2}v } \tag{1.18}
$$


### Advective Form

Doing a product rule expansion of the terms on the left side of this equation and invoking the mass continuity equation results in the advective form of the Navier-Stokes equation:

$$
\boxed{ \frac{dv}{dt}=B-\frac{1}{\rho}\nabla p+\nu\nabla^{2}v } \tag{1.19}
$$

The quantity $\nu=\mu/\rho$ is called the kinematic viscosity. Both $\mu$ and $\nu$ are functions of temperature and pressure, and are generally treated as constants.

#### Body Forces ($B$)

  </figcaption>
</figure>
<br>

The distinguishing factor in geophysical fluid dynamics is the inclusion of body forces arising from the non-inertial reference frame of the rotating Earth (see Figure 1.3). We can express the total body force ($B$) acting on a fluid parcel by breaking it down into its constituent forces:

$$
B = \underbrace{g^*}_{\text{true gravity}} - \underbrace{\Omega \times (\Omega \times r)}_{\text{centrifugal force}} - \underbrace{2\Omega \times v}_{\text{Coriolis force}} \tag{1.20}
$$

Because Earth's rotation creates an outward centrifugal pull, it is combined with true gravity to define a constant gravity vector ($-gk$). We can also express this using the geopotential $\Phi = gz$:

$$
B = \underbrace{-g k}_{-\nabla\Phi} - 2\Omega \times v \tag{1.21}
$$

In large-scale GFD, the vertical component of the Coriolis force is typically dwarfed by stronger vertical forces and is often neglected. Removing it yields the standard simplified body force equation using the Coriolis parameter ($f = 2\Omega \sin\theta$):

$$
B = -g k - \underbrace{f k \times v}_{\text{horizontal Coriolis}} \tag{1.22}
$$

## Boussinesq approximation
<div style="border-left: 5px solid #1f77b4; background-color: #eaf4ff; padding: 12px; border-radius: 5px;">
    
The Boussinesq approximation simplifies fluid equations when density perturbations are small relative to a constant reference state ($\rho' \ll \rho_0$). It treats density as constant in the inertial and mass continuity equations, retaining variations solely in the gravitational term to resolve **buoyancy forces**. While this elegantly models nearly **incompressible fluids** like the ocean, it breaks down in the highly compressible atmosphere. Consequently, atmospheric dynamics are reformulated using **potential temperature** ($\theta$) and the **Exner function** ($\Pi$) to accurately capture stratified, **compressible flows**.
    
Below, we derive the Boussinesq approximation for each system.

</div>

<hr style="border:none; border-top:1px solid #888;">

### The Ocean

#### Mass Continuity

We start with the advective form of the mass continuity equation, which was previously derived:

$$
\frac{1}{\rho}\frac{d\rho}{dt} + \nabla \cdot v = 0 \tag{1.13}
$$

For the ocean, we assume the density of sea water parcels does not change with time except possibly as a result of diffusion. This condition is expressed compactly in terms of the material derivative of density:

$$
\frac{d\rho}{dt} = \sigma \nabla^2 \rho \tag{1.23}
$$

where $\sigma$ is a defined diffusivity.

Replacing (1.23) into the mass continuity equation (1.13) yields:

$$
\frac{\sigma \nabla^2 \rho}{\rho} + \nabla \cdot v = 0 \tag{1.24}
$$

Because the effect of diffusivity on this equation is generally considered to be negligible, it is therefore omitted:

$$
\boxed{\nabla \cdot v = 0} \tag{1.25}
$$


#### Momentum Equation

Starting with the advective form of the momentum equation:

$$
\frac{dv}{dt} = B - \frac{1}{\rho}\nabla p + \nu \nabla^2 v \tag{1.19}
$$

For the ocean, we write the body force $B$ by including gravity and the Coriolis force (ignoring the vertical component of the Coriolis force). We use the simplified horizontal Coriolis expression:

$$
B = -gk - fk \times v \tag{1.22}
$$

Substituting (1.22) into (1.19) and bringing all terms to the left side gives the advective form of the momentum equation for the ocean:

$$
\frac{dv}{dt} + \frac{1}{\rho}\nabla p + gk + fk \times v - \nu \nabla^2 v = 0 \tag{1.26}
$$

Before introducing perturbations, we define the base hydrostatic state. If we consider the ocean to be at rest so that $v = 0$, the momentum equation reduces to:

$$
\frac{1}{\rho}\nabla p = -gk \tag{1.27}
$$

Looking at the horizontal components of this vector equation shows us that $\frac{\partial p}{\partial x} = \frac{\partial p}{\partial y} = 0$, meaning $p = p(z)$. The vertical component can therefore be written as the hydrostatic equation:

$$
\frac{dp}{dz} = -g\rho \tag{1.28}
$$

Now, let us assume that the pressure and density can be decomposed into a background state in hydrostatic balance—a state where the downward gravitational force acting on a fluid parcel is exactly balanced by the upward pressure gradient force—plus small perturbations:

$$
p = p_0(z) + p' \quad \text{and} \quad \rho = \rho_R + \rho' \tag{1.29}
$$

Here, the background pressure $p_0$ is a function of $z$, but we take the background density $\rho_R$ to be a constant reference value close to the mean density of the fluid. Because this background state is in hydrostatic balance, it satisfies:

$$
\nabla p_0 = -g\rho_R k \tag{1.30}
$$

Substituting our expansion (1.29) and (1.30) into the pressure gradient and gravity terms of equation (1.26):

$$
\frac{1}{\rho_R + \rho'}\nabla(p_0 + p') + gk = \frac{-g\rho_R k + \nabla p'}{\rho_R + \rho'} + gk \tag{1.31}
$$

Factoring out $\frac{1}{\rho_R}$ and applying the binomial expansion approximation $\frac{1}{1 + \rho'/\rho_R} \approx 1 - \frac{\rho'}{\rho_R}$:

$$
\approx \frac{1}{\rho_R}\left(1 - \frac{\rho'}{\rho_R}\right)(-g\rho_R k + \nabla p') + gk \tag{1.32}
$$

Simplifying this expression and dropping the terms that are quadratic in primed quantities yields:

$$
\frac{1}{\rho_R}\nabla p' + g\frac{\rho'}{\rho_R}k \tag{1.33}
$$

We define the kinematic pressure $\pi$ and the buoyancy $b$ as:

$$
\pi = \frac{p'}{\rho_R} \quad \text{and} \quad b = -g\frac{\rho'}{\rho_R} \tag{1.34}
$$

Substituting these definitions back into our simplified expression, and placing that result back into the full momentum equation (1.26), we get the Boussinesq momentum equation:

$$
\boxed{\frac{dv}{dt} + \nabla \pi - bk + fk \times v - \nu \nabla^2 v = 0} \tag{1.35}
$$

**We can further simplify (1.35) by exploring the following balance conditions:**

<div style="border-left: 5px solid #fbc02d; background-color: #fff9c4; padding: 12px; border-radius: 5px;">

**The Hydrostatic Approximation**

If the vertical parcel acceleration is small, then the hydrostatic approximation can be made. This comes from ignoring the vertical acceleration and viscosity terms in the vertical component of the momentum equation, resulting in:

$$
\frac{\partial \pi}{\partial z} - b = 0 \tag{1.36}
$$

**Note:** By making the Hydrostatic Approximation, we are mathematically assuming the fluid is close enough to Hydrostatic Balance that we can safely apply the Hydrostatic Equation defined above.
</div>

<br>

<div style="border-left: 5px solid #fbc02d; background-color: #fff9c4; padding: 12px; border-radius: 5px;">

**Geostrophic Balance**

This balance occurs when the flow is horizontal and does not depend on $x$, $y$, or $t$. By defining the horizontal velocity components as $(u, v)$ and ignoring horizontal viscosity, equation (1.35) reduces to a strict balance between the pressure gradient and the Coriolis force:

$$
\nabla\pi + fk \times v = 0 \tag{1.37}
$$

To solve for the individual velocity components, we expand the cross product of the vertical unit vector $k$ and the horizontal velocity vector $v = u\mathbf{i} + v\mathbf{j}$:

$$
k \times v = -v\mathbf{i} + u\mathbf{j} \tag{1.38}
$$

Substituting this expanded cross product back into (1.37) alongside the horizontal components of the kinematic pressure gradient:

$$
\left(\frac{\partial \pi}{\partial x} - fv\right)\mathbf{i} + \left(\frac{\partial \pi}{\partial y} + fu\right)\mathbf{j} = 0 \tag{1.39}
$$

Solving for $u$ and $v$ gives the geostrophic velocity $(u_G, v_G)$:

$$
\boxed{(u_G, v_G) = \left(-\frac{1}{f}\frac{\partial \pi}{\partial y}, \frac{1}{f}\frac{\partial \pi}{\partial x}\right)} \tag{1.40}
$$
</div>

<br>
<div style="border-left: 5px solid #fbc02d; background-color: #fff9c4; padding: 12px; border-radius: 5px;">

**The Thermal Wind Equation**

In hydrostatic balance with the Boussinesq approximation, the horizontal gradient of buoyancy is proportional to the vertical derivative of the geostrophic wind; this is called the thermal wind equation. We can prove this by taking the vertical derivative ($\frac{\partial}{\partial z}$) of the geostrophic velocity components from (1.40):

$$
\frac{\partial u_G}{\partial z} = -\frac{1}{f}\frac{\partial}{\partial z}\left(\frac{\partial \pi}{\partial y}\right) = -\frac{1}{f}\frac{\partial}{\partial y}\left(\frac{\partial \pi}{\partial z}\right) \tag{1.41}
$$

$$
\frac{\partial v_G}{\partial z} = \frac{1}{f}\frac{\partial}{\partial z}\left(\frac{\partial \pi}{\partial x}\right) = \frac{1}{f}\frac{\partial}{\partial x}\left(\frac{\partial \pi}{\partial z}\right) \tag{1.42}
$$

Using the hydrostatic approximation from (1.36), we can substitute $\frac{\partial \pi}{\partial z} = b$ into these equations to yield the thermal wind balance:

$$
\boxed{\left(\frac{\partial u_G}{\partial z}, \frac{\partial v_G}{\partial z}\right) = \left(-\frac{1}{f}\frac{\partial b}{\partial y}, \frac{1}{f}\frac{\partial b}{\partial x}\right)} \tag{1.43}
$$
</div>

#### Buoyancy Equation

We return to the incompressibility condition from (1.23) and expand it using the definition of the material derivative:

$$
\frac{\partial \rho}{\partial t} + v \cdot \nabla \rho = \sigma \nabla^2 \rho \tag{1.44}
$$

Substitute the density perturbation from (1.29). Because $\rho_R$ is a constant reference value, its spatial and temporal derivatives are zero:

$$
\frac{\partial \rho'}{\partial t} + v \cdot \nabla \rho' = \sigma \nabla^2 \rho' \tag{1.45}
$$

Multiply the entire equation by $-g/\rho_R$ to convert the density perturbation into buoyancy using (1.34):

$$
\frac{\partial b}{\partial t} + v \cdot \nabla b = \sigma \nabla^2 b \tag{1.46}
$$

Finally, substituting this into the equation and generalizing the right side into an arbitrary source of buoyancy, $S_b$, results in:

$$
\boxed{\frac{\partial b}{\partial t} + v \cdot \nabla b = S_b} \tag{1.47}
$$

### The Atmosphere

#### Momentum Equation

Unlike the ocean, the atmosphere is highly compressible, meaning density changes significantly with height and pressure. Therefore, the specific entropy of dry air ($s$) becomes a key variable, expressed in terms of temperature ($T$) and pressure ($p$):

$$
s = C_p \ln\left(\frac{T}{T_R}\right) - R \ln\left(\frac{p}{p_R}\right) \tag{1.48}
$$

Here, $C_p$ is the specific heat at constant pressure, $R$ is the gas constant for dry air, and $T_R$ and $p_R$ are constant reference values. Entropy is conserved in slow, adiabatic processes. If heating occurs, the governing equation for specific dry entropy is:

$$
\frac{ds}{dt} = \frac{Q}{T} \tag{1.49}
$$

where $Q$ is the heating rate per unit mass.

In atmospheric dynamics, an alternative variable to entropy is extensively used: **potential temperature** ($\theta$). This is the temperature a parcel of air would attain if it were compressed or expanded isentropically to the standard reference pressure $p_R$:

$$
\theta = T\left(\frac{p_R}{p}\right)^\kappa \tag{1.50}
$$

where $\kappa = R/C_p$. We can directly relate entropy to potential temperature:

$$
s = C_p \ln\left(\frac{\theta}{T_R}\right) \tag{1.51}
$$

By taking the material derivative of (1.51) and combining it with (1.49), we get the governing thermodynamic energy equation in terms of potential temperature:

$$
\frac{d\theta}{dt} = \frac{\theta Q}{C_p T} = \frac{Q}{\Pi} \tag{1.52}
$$

where $\Pi = C_p T/\theta = C_p(p/p_R)^\kappa$ is defined as the **Exner function**.

Now, we consider the advective form of the momentum equation, including gravity and the Coriolis force:

$$
\frac{dv}{dt} + \frac{1}{\rho}\nabla p + gk + fk \times v - \nu \nabla^2 v = 0
$$

To couple our new thermodynamic framework with the fluid dynamics, we must link the pressure gradient force to $\theta$ and $\Pi$. Using the ideal gas law ($p/\rho = RT$) and the definition of potential temperature, we can establish a differential relationship:

$$
\theta d\Pi = \theta R\left(\frac{p}{p_R}\right)^{\kappa-1}\frac{dp}{p_R} = \frac{RT}{p}dp = \frac{dp}{\rho} \tag{1.53}
$$

When applied to a spatial gradient, this relationship allows us to eliminate density and pressure entirely in favor of potential temperature and the Exner function ($\theta \nabla \Pi = \frac{1}{\rho} \nabla p$). Substituting this into the momentum equation yields:

$$
\boxed{\frac{dv}{dt} + \theta\nabla\Pi + gk + fk \times v - \nu \nabla^2 v = 0} \tag{1.54}
$$

#### Anelastic Mass Continuity

We start with the standard flux form of mass continuity:

$$
\frac{\partial\rho}{\partial t} + \nabla \cdot (\rho v) = 0
$$

Because tracking full density variations complicates the mathematics by introducing high-frequency sound waves (which are largely irrelevant for meteorological weather patterns), we apply the **anelastic approximation**. 

This approximation assumes that the local time derivative of density ($\partial\rho/\partial t$) is negligible. Furthermore, it replaces the fully variable density $\rho$ inside the divergence operator with a fixed, ambient background density profile that only depends on height, denoted as $\rho_0(z)$. This reduces the continuity equation to:

$$
\boxed{\nabla \cdot (\rho_0 v) = 0} \tag{1.55}
$$


#### Density

To mathematically close this system of governing equations, we must maintain an explicit relationship between the dynamical variables and the state of the gas. We want to express the density of air $\rho$ entirely in terms of pressure $p$ and potential temperature $\theta$. 

We begin with the ideal gas law:

$$
p = \rho R T \implies \rho = \frac{p}{RT} 
$$

From our earlier definition of potential temperature (1.50), we can rearrange to solve for the actual temperature $T$:

$$
T = \theta\left(\frac{p}{p_R}\right)^\kappa 
$$

Substituting this actual temperature back into the ideal gas law yields our final equation of state:

$$
\boxed{\rho = \frac{p}{R\theta}\left(\frac{p_R}{p}\right)^\kappa} \tag{1.56}
$$

**Note:** Applying oceanic assumptions, the atmospheric momentum equation matches (1.35). Ignoring heating reduces the thermodynamic equation (1.52) to the standard buoyancy equation (1.47). Finally, assuming a constant $\rho_0$, the least justifiable step, simplifies the anelastic mass continuity (1.55) into the incompressible form (1.25).

**1. Why the mass continuity equation has no diﬀusion term?** 

**A:** The mass continuity equation tracks the total amount of fluid. It states that any mass gained in a space must exactly equal the mass flowing into it.

Diffusion spreads specific properties, like salt or heat, from high to low concentrations within the fluid. While diffusion moves these internal components around, it does not create or destroy the total fluid mass.

Because diffusion only rearranges things internally without adding or removing total mass from the system, it has no effect on the overall mass balance. Therefore, a diffusion term is not needed in the mass continuity equation.

<hr style="border:none; border-top:1px solid #888;">

**2. Imagine a container with salt water in the lower half and fresh water in the upper half.
As this container sits on the shelf over the years, salt gradually diffuses upward until ultimately
the salinity is uniform through the container. In this process is there fluid motion?**

**A:** No, there is no bulk fluid motion. At a macroscopic level, the water remains entirely stationary.

The upward movement of salt happens purely through molecular diffusion, the random, microscopic jiggling of individual salt and water molecules. While the salt ions gradually spread into the fresh water to balance the concentration, this process does not create a physical current, wave, or flow in the tank.

In fluid dynamics, we strictly distinguish between this invisible molecular spreading (diffusion) and the actual physical movement of continuous fluid parcels (advection).

**3. In hydrostatic balance with the Boussinesq approximation, show that the horizontal
gradient of buoyancy is proportional to the vertical derivative of the geostrophic wind. This
is called the thermal wind equation. Investigate how this plays out in the exact atmospheric
equations.**

**A:** As detailed in the previous derivations for the oceanic Boussinesq approximation, the thermal wind equation was already computed. 

Investigating how this plays out in the exact atmospheric equations:

Because air is highly compressible, we cannot use the Boussinesq approximation as we did for the ocean. Instead, the atmospheric thermal wind relies on thermodynamic variables linked to potential temperature ($\theta$) and the Exner function ($\Pi$). To derive it, we first establish the exact atmospheric geostrophic balance. By equating the Coriolis force to our thermodynamic pressure gradient ($\theta\nabla\Pi + f\mathbf{k} \times \mathbf{v}_G = 0$), we get the geostrophic velocities:

$$f u_G = -\theta \frac{\partial \Pi}{\partial y}, \quad f v_G = \theta \frac{\partial \Pi}{\partial x}$$

We couple this with the exact vertical hydrostatic balance:

$$\frac{\partial \Pi}{\partial z} = -\frac{g}{\theta}$$

Differentiating the $u_G$ and $v_G$ equations with respect to $z$ yields:

$$f \frac{\partial u_G}{\partial z} = -\frac{\partial \theta}{\partial z} \frac{\partial \Pi}{\partial y} - \theta \frac{\partial^2 \Pi}{\partial y \partial z}$$

$$f \frac{\partial v_G}{\partial z} = \frac{\partial \theta}{\partial z} \frac{\partial \Pi}{\partial x} + \theta \frac{\partial^2 \Pi}{\partial x \partial z}$$

Taking the horizontal derivatives of the hydrostatic equation gives the necessary cross-derivatives:

$$\frac{\partial^2 \Pi}{\partial y \partial z} = \frac{g}{\theta^2} \frac{\partial \theta}{\partial y}, \quad \frac{\partial^2 \Pi}{\partial x \partial z} = \frac{g}{\theta^2} \frac{\partial \theta}{\partial x}$$

Substituting these cross-derivatives back into our velocity derivatives reveals the expanded atmospheric thermal wind equations:

$$\boxed{f \frac{\partial u_G}{\partial z} = -\frac{\partial \theta}{\partial z} \frac{\partial \Pi}{\partial y} - \frac{g}{\theta} \frac{\partial \theta}{\partial y}}$$

$$\boxed{f \frac{\partial v_G}{\partial z} = \frac{\partial \theta}{\partial z} \frac{\partial \Pi}{\partial x} + \frac{g}{\theta} \frac{\partial \theta}{\partial x}}$$

**4.Using the hydrostatic equation, compute the pressure as a function of depth for a constant-density ocean and as a function of height for an isothermal and an isentropic atmosphere.**

**A:** The formula for each case is the standard hydrostatic equation:

$$\frac{dp}{dz}=-g\rho$$

**a) Constant-Density Ocean**

For an ocean modeled as an incompressible fluid, the density is assumed to be constant. Let this constant density be $\rho_0$.

Substituting this into the hydrostatic equation yields:
$$\frac{dp}{dz}=-g\rho_0$$

To find the pressure as a function of depth $D$, we set $z = -D$ and integrate from the surface (where $D = 0$ and pressure is $p_0$) down to depth $D$:
$$\int_{p_0}^{p}dp=\int_{0}^{-D}-g\rho_0dz$$
$$p-p_0=-g\rho_0(-D-0)$$

Solving for $p$, we get the pressure as a function of depth:
$$\boxed{p(D)=p_0+g\rho_0D}$$

**b) Isothermal Atmosphere**

In an isothermal atmosphere, the temperature remains constant at $T_0$ regardless of height. 

First, we use the ideal gas law, $p/\rho=RT$, to express the density of air in terms of pressure:
$$\rho=\frac{p}{RT_0}$$

Substituting this density into the hydrostatic equation gives:
$$\frac{dp}{dz}=-g\frac{p}{RT_0}$$

Rearranging the equation to separate the variables $p$ and $z$:
$$\frac{dp}{p}=-\frac{g}{RT_0}dz$$

Integrating from the surface ($z=0, p=p_0$) to height $z$:
$$\int_{p_0}^{p}\frac{1}{p}dp=\int_{0}^{z}-\frac{g}{RT_0}dz$$
$$\ln\left(\frac{p}{p_0}\right)=-\frac{gz}{RT_0}$$

Taking the exponential of both sides provides the pressure as a function of height:
$$\boxed{p(z)=p_0\exp\left(-\frac{gz}{RT_0}\right)}$$

**c) Isentropic Atmosphere**

An isentropic atmosphere is one where the entropy $s$ is conserved, meaning the potential temperature $\theta$ is also constant throughout the column ($\theta = \theta_0$).

Using the exact vertical hydrostatic balance established in the previous derivation:
$$\frac{\partial \Pi}{\partial z} = -\frac{g}{\theta}$$

Since the potential temperature is constant ($\theta = \theta_0$), this vertical derivative becomes:
$$\frac{d\Pi}{dz}=-\frac{g}{\theta_0}$$

Integrating from the surface ($z=0, \Pi=\Pi_0$) to height $z$:
$$\int_{\Pi_0}^{\Pi}d\Pi=\int_{0}^{z}-\frac{g}{\theta_0}dz$$
$$\Pi(z)=\Pi_0-\frac{gz}{\theta_0}$$

The Exner function is defined as $\Pi=C_p(p/p_R)^\kappa$. If we assume the surface pressure $p_0$ is equal to the reference pressure $p_R$, then the surface Exner function is $\Pi_0 = C_p$, and the surface potential temperature equals the actual surface temperature ($\theta_0 = T_0$).

Substituting the definition of $\Pi$ back into our integrated equation:
$$C_p\left(\frac{p}{p_0}\right)^\kappa=C_p-\frac{gz}{T_0}$$

Dividing the entire equation by $C_p$:
$$\left(\frac{p}{p_0}\right)^\kappa=1-\frac{gz}{C_pT_0}$$

Finally, solving for $p$ gives the pressure as a function of height:
$$\boxed{p(z)=p_0\left(1-\frac{gz}{C_pT_0}\right)^{1/\kappa}}$$
