# Circulation and Potential Vorticity

## Circulation and Stokes' Theorem

The vorticity is defined as $\zeta=\nabla\times v$. The circulation is the closed line integral of the fluid velocity parallel to the path. By applying Stokes' theorem, where $n$ is the unit normal to the area element $dA$, we relate the circulation to the vorticity:

$${\color{blue}\boxed{\Gamma=\oint v\cdot dl=\int(\nabla\times v)\cdot ndA=\int\zeta\cdot ndA}} \tag{1}$$

## Kelvin Circulation Theorem

Taking the time derivative of $\Gamma$ is complex because the circulation loop evolves with time. We first write it in finite sum form, where $dl\rightarrow\Delta l_{i}$:

$$\frac{d\Gamma}{dt}=\frac{d}{dt}\sum v_{i}\cdot\Delta l_{i}=\sum\frac{dv_{i}}{dt}\cdot\Delta l_{i}+\sum v_{i}\cdot\Delta v_{i} \tag{2}$$

Reverting to an integral form and utilizing the relation $v\cdot dv=d(v^{2}/2)$, we get:

$$\frac{d\Gamma}{dt}=\oint\frac{dv}{dt}\cdot dl+\oint d(v^{2}/2) \tag{3}$$

The second integral is a perfect differential over a closed path and equals zero. The total time derivative of velocity is replaced using the momentum equation:

$$\frac{dv}{dt}+\theta\nabla\Pi+\nabla\Phi+2\Omega\times v=0 \tag{4}$$

Assuming an inertial reference frame where $\Omega=0$, equation (3) becomes:

$$\frac{d\Gamma}{dt}=-\oint(\theta\nabla\Pi+\nabla\Phi)\cdot dl=-\oint(\theta\nabla\Pi\cdot dl+d\Phi) \tag{5}$$

Since the integral of the perfect differential $d\Phi$ around a closed loop is zero, we obtain the Kelvin circulation theorem:

$${\color{green}\boxed{\frac{d\Gamma}{dt}=-\oint\theta\nabla\Pi\cdot dl}} \tag{6}$$

## Absolute Vorticity and Circulation

To compute the circulation in the rotating frame of the earth, we relate the inertial velocity $v_{I}$ to the rotating frame velocity $v$:

$$v_{I}=v+\Omega\times r \tag{7}$$

Using Stokes' theorem to convert the rotation line integral into an area integral, the circulation becomes:

$${\color{orange}\boxed{\Gamma=\oint v_{I}\cdot dl=\oint v\cdot dl+\int[\nabla\times(\Omega\times r)]\cdot ndA}} \tag{8}$$

Applying the vector identity $\nabla\times(\Omega\times r)=2\Omega$ reduces the equation:

$$\Gamma=\oint v\cdot dl+\int 2\Omega\cdot ndA \tag{9}$$

Recognizing absolute vorticity as $\zeta_{a}=\nabla\times v_{I}=\nabla\times v+2\Omega$, we obtain:

$${\color{orange}\boxed{\Gamma=\int(\nabla\times v+2\Omega)\cdot ndA=\int\zeta_{a}\cdot ndA}} \tag{10}$$

## Shallow Water Potential Vorticity

In the shallow water approximation, the circulation bounds a parcel with area $A$ and thickness $h$. The circulation is:

$${\color{purple}\boxed{\Gamma=\zeta_{a}A=\frac{\zeta_{a}}{h}(Ah)}} \tag{11}$$

Since the fluid is incompressible, the volume $Ah$ does not change. Because circulation is conserved around the parcel, the potential vorticity $q$ is also conserved:

$${\color{purple}\boxed{q\equiv\frac{\zeta_{a}}{h}=\frac{1}{h}\left(\frac{\partial v}{\partial x}-\frac{\partial u}{\partial y}+f\right)}} \tag{12}$$

$${\color{purple}\boxed{\frac{dq}{dt}=\frac{\partial q}{\partial t}+u\frac{\partial q}{\partial x}+v\frac{\partial q}{\partial y}=0}} \tag{13}$$

## Incompressible Ocean Potential Vorticity

For an incompressible ocean with spatially variable density, a parcel bounded by density surfaces has a circulation defined by:

$${\color{pink}\boxed{\Gamma=\zeta_{a}A=\zeta_{a}\left(\frac{\rho_{2}-\rho_{1}}{h\rho}\right)\left(\frac{hA\rho}{\rho_{2}-\rho_{1}}\right)=\frac{\zeta_{a}\cdot\nabla\rho}{\rho}\left(\frac{hA\rho}{\rho_{2}-\rho_{1}}\right)}} \tag{14}$$


The mass of the parcel $hA\rho$ is constant in time. With the loop confined to a constant density surface, the circulation is constant, so the potential vorticity is conserved:

$${\color{pink}\boxed{q=\frac{\zeta_{a}\cdot\nabla\rho}{\rho}}} \tag{15}$$

$${\color{pink}\boxed{\frac{dq}{dt}=\frac{\partial q}{\partial t}+v\cdot\nabla q=0}} \tag{16}$$

## Atmospheric Potential Vorticity

In the atmosphere, bounding surfaces are constant potential temperature. The Ertel potential vorticity is:

$${\color{aqua}\boxed{q=\frac{\zeta_{a}\cdot\nabla\theta}{\rho}}} \tag{17}$$

Under the Boussinesq approximation, density is a constant factor and potential temperature is replaced by buoyancy $b$:

$${\color{aqua}\boxed{q=\zeta_{a}\cdot\nabla b}} \tag{18}$$

In pressure coordinates, the absolute vorticity and gradient operator are:

$$\zeta_{a}=\left(-\frac{\partial v}{\partial p},\frac{\partial u}{\partial p},\frac{\partial v}{\partial x}-\frac{\partial u}{\partial y}+f\right) \tag{19}$$

$$\nabla=\left(\frac{\partial}{\partial x},\frac{\partial}{\partial y},\frac{\partial}{\partial p}\right) \tag{20}$$

Using density $-g^{-1}$, the Ertel PV in pressure coordinates is:

$${\color{aqua}\boxed{q=-g\zeta_{a}\cdot\nabla\theta}} \tag{21}$$

Expanding the dot product in geometric variables provides its full form:

$${\color{aqua}\boxed{q=\frac{1}{\rho}\left[-\frac{\partial v}{\partial z}\frac{\partial\theta}{\partial x}+\frac{\partial u}{\partial z}\frac{\partial\theta}{\partial y}+\left(\frac{\partial v}{\partial x}-\frac{\partial u}{\partial y}+f\right)\frac{\partial\theta}{\partial z}\right]}} \tag{22}$$

## Isentropic Coordinates Basics

Flow velocity on the $x-y-\theta$ grid is defined as:

$$v=\left(\frac{dx}{dt},\frac{dy}{dt},\frac{d\theta}{dt}\right)=(u,v,S)=(v_{h},S) \tag{23}$$

Density $\sigma$ relates to geometric density $\rho$ by:

$$\sigma d\theta=\rho dz \tag{24}$$

The mass continuity equation is inferred to be:

$${\color{grey}\boxed{\frac{\partial\sigma}{\partial t}+\nabla_{h}\cdot(\sigma v_{h})+\frac{\partial}{\partial\theta}(\sigma S)=0}} \tag{25}$$

The horizontal momentum equation incorporates the Montgomery potential gradient and Coriolis force:

$${\color{grey}\boxed{\frac{dv_{h}}{dt}+\nabla_{h}M+fk\times v_{h}=0}} \tag{26}$$

Where the Montgomery potential is:

$${\color{grey}\boxed{M=\theta\Pi+\Phi}} \tag{27}$$

The total time derivative in isentropic coordinates is:

$$\frac{d}{dt}=\frac{\partial}{\partial t}+v_{h}\cdot\nabla_{h}+S\frac{\partial}{\partial\theta} \tag{28}$$

Substituting equation (24) into the hydrostatic equation yields:

$${\color{grey}\boxed{\frac{\partial p}{\partial\theta}=-g\sigma}} \tag{29}$$

## Geopotential Diagnostic Equation

Using the alternative atmospheric hydrostatic form $\theta d\Pi=-d\Phi$ and the differential $dM=\Pi d\theta+\theta d\Pi+d\Phi$, we find:

$$\frac{\partial M}{\partial\theta}=\Pi \tag{30}$$

Combining this with equation (27) yields the diagnostic equation for geopotential:

$${\color{yellow}\boxed{\Phi=M-\theta\frac{\partial M}{\partial\theta}}} \tag{31}$$

## Isentropic Potential Vorticity

Combining equations (29) and (30) maps a direct relationship between $M$ and $\sigma$:

$${\color{red}\boxed{\frac{\partial^{2}M}{\partial\theta^{2}}=-\frac{g}{\rho\theta}\sigma}} \tag{32}$$

The surface potential temperature $\theta_{B}$ evolves according to:

$${\color{red}\boxed{\frac{\partial\theta_{B}}{\partial t}+v_{B}\cdot\nabla\theta_{B}=S_{B}}} \tag{33}$$

With these coordinates established, the Ertel potential vorticity takes a simple scalar form based on the vorticity normal to isentropic surfaces:

$${\color{red}\boxed{q=\frac{\zeta_{a}}{\sigma}=\frac{1}{\sigma}\left(\frac{\partial v}{\partial x}-\frac{\partial u}{\partial y}+f\right)}} \tag{34}$$

## Approximations in Isentropic Coordinates

The anelastic approximation replaces $\rho$ with an ambient profile $\rho_{0}(\theta)$ in equation (32):

$${\color{lime}\boxed{\frac{\partial^{2}M}{\partial\theta^{2}}=-\frac{g}{\rho_{0}\theta}\sigma}} \tag{35}$$

Recognizing that $\rho_{0}=\sigma_{0}\Gamma$ and $(g/\theta)\Gamma=N^{2}(\theta)$ (the square of the Brunt-Väisälä frequency), we rewrite this as:

$${\color{lime}\boxed{\frac{\partial^{2}M}{\partial\theta^{2}}=-\frac{N^{2}}{\sigma_{0}\Gamma^{2}}\sigma}} \tag{36}$$

The Boussinesq approximation replaces $\Gamma$ and $N$ with constant reference values $\Gamma_{R}$ and $N_{R}$, assuming constant $\sigma_{0}$:

$${\color{lime}\boxed{\frac{\partial^{2}M}{\partial\theta^{2}}=-\frac{N_{R}^{2}}{\sigma_{0}\Gamma_{R}^{2}}\sigma}} \tag{37}$$

Due to small variations in $\theta$, equation (31) scales down to a simpler diagnostic equation with a constant reference $M_{R}$ and $\theta_{R}$:

$${\color{lime}\boxed{\Phi=M_{R}-\theta_{R}\frac{\partial M}{\partial\theta}}} \tag{38}$$

From equation (30), the corresponding geopotential-Exner simplification is obtained:

$${\color{lime}\boxed{\Phi=\theta_{R}(C_{p}-\Pi)}} \tag{39}$$

### **1.Using conservation of circulation for a horizontal loop, answer:**

The absolute circulation $\Gamma$ of a loop moving with the fluid is conserved. The absolute circulation can be expressed as $\Gamma = (\zeta + f)A$, where $\zeta$ is relative vorticity, $f$ is the Coriolis parameter (planetary vorticity), and $A$ is the area of the loop.  

#### (a) If relative vorticity is initially zero, how does it change if loop area decreases at fixed positive latitude? 

The absolute circulation $\Gamma$ of a loop moving with the fluid is conserved. The absolute circulation can be expressed as $\Gamma = (\zeta + f)A$, where $\zeta$ is relative vorticity, $f$ is the Coriolis parameter (planetary vorticity), and $A$ is the area of the loop.  


#### (b) If circulation is zero at the equator, how does relative vorticity change as a parcel moves to positive latitude without changing area? 

At the equator, $f = 0$. Since circulation is zero ($\Gamma = 0$), the initial relative vorticity is $\zeta = 0$. As the parcel moves to a positive latitude without changing area, $f$ becomes positive. To keep $\Gamma = 0$, the relative vorticity must decrease (become negative) so that $\zeta = -f$.


#### (c) If a loop moves from large positive latitude toward the equator maintaining zero relative vorticity, how does loop area and layer thickness change? 

Initially, $\zeta = 0$ at a large positive latitude, so $\Gamma = f_{large} A_{initial}$. As it moves toward the equator, $f$ decreases. To maintain zero relative vorticity ($\zeta = 0$) and conserve $\Gamma$, the loop area $A$ must increase. Because the volume of a shallow water parcel ($V = Ah$) is constant, an increase in area means the layer thickness $h$ must decrease. 

### **2. Prove that the amount of potential vorticity substance $Q = \int \rho q dV$ between two isentropic surfaces never changes.**

The amount of potential vorticity substance is $Q = \int \rho q dV$.
In isentropic coordinates, the mass element is $\rho dV = \sigma dA d\theta$, and the Ertel potential vorticity is defined as $q = \zeta_a / \sigma$.  Substituting these into the integral gives:$$Q = \int \int \sigma \left(\frac{\zeta_a}{\sigma}\right) dA d\theta = \int \int \zeta_a dA d\theta$$

The inner integral, $\int \zeta_a dA$, is precisely the absolute circulation $\Gamma$ around the bounding loop of the area $A$. According to Kelvin's circulation theorem, on an isentropic surface with no heating or friction, $\Gamma$ is conserved. Since the circulation on every individual isentropic slice is conserved, the total volume integral $Q$ bounded between two constant $\theta$ surfaces never changes.

### **3. Derive governing equations for isopycnal coordinates in the ocean in analogy with atmospheric isentropic equations.**

In analogy with atmospheric isentropic coordinates where $\theta$ replaces $z$ , isopycnal coordinates use density $\rho$ as the vertical coordinate.  Mass Continuity: The isentropic density $\sigma$ is replaced by the isopycnal layer thickness $h = \partial z / \partial \rho$. The equation becomes:$$\frac{\partial h}{\partial t} + \nabla_\rho \cdot (h v_h) = 0$$Momentum: The Montgomery potential is adapted for the ocean using pressure and geopotential depth as $M = p/\rho_0 + gz$. The momentum equation becomes:$$\frac{dv_h}{dt} + \nabla_\rho M + f k \times v_h = 0$$
