# Quasi-Geostrophic Theory

## Shallow Water Case

We begin by defining potential vorticity (PV) in a shallow water system:

$$q = \frac{\zeta_a}{h} \tag{1}$$

Here, the absolute vorticity is defined as:

$$\zeta_a = \frac{\partial v}{\partial x} - \frac{\partial u}{\partial y} + f \tag{2}$$

The geostrophic wind components ($u_g, v_g$) rely on the fluid layer thickness ($h$) and the terrain elevation ($d$):

$$u_g = -\frac{g}{f}\frac{\partial(h+d)}{\partial y}, \quad v_g = \frac{g}{f}\frac{\partial(h+d)}{\partial x} \tag{3}$$

Substituting the geostrophic wind into the absolute vorticity equation gives us the geostrophic absolute vorticity ($\zeta_{ag}$):

$$\zeta_{ag} = h q_g = \frac{\partial}{\partial x}\left(\frac{g}{f}\frac{\partial(h+d)}{\partial x}\right) + \frac{\partial}{\partial y}\left(\frac{g}{f}\frac{\partial(h+d)}{\partial y}\right) + f \tag{4}$$

Assuming small amplitude disturbances on a beta-plane ($h=h_0(1+\eta)$ and $f=f_0+\beta y$), we linearize the potential vorticity:

$$q_g \approx q_A + q_0(L_R^2 \nabla^2 \eta^* - \eta^*) \tag{5}$$

where the base state potential vorticity ($q_A$) is:

$$q_A = q_0\left[1 + \frac{\beta y}{f_0} + \frac{d}{h_0}\right] \tag{6}$$

By defining $q^* = q_g - q_A$ as the part of the potential vorticity associated with motion, we can rewrite equation 5 into a more convenient form representing the shallow water PV inversion:

$${\color{blue}\boxed{L_R^2 \nabla^2 \eta^* - \eta^* = \frac{q^*}{q_0}}} \tag{7}$$

Using $\eta^*$, the geostrophic wind components are written as:

$$u_g = -f_0 L_R^2 \frac{\partial \eta^*}{\partial y}, \quad v_g = f_0 L_R^2 \frac{\partial \eta^*}{\partial x} \tag{8}$$

Next, we invoke the mass continuity equation, using the geostrophic wind in the advection term due to the low Rossby number:

$$\frac{\partial \eta}{\partial t} + v_g \cdot \nabla \eta + \nabla \cdot v_a = 0 \tag{9}$$

The irrotational part of the ageostrophic flow can be represented by a velocity potential $\chi$:

$$v_a = -\nabla \chi \tag{10}$$

Substituting this into the mass continuity equation yields a Poisson equation for the velocity potential:

$$\nabla^2 \chi = \frac{\partial \eta}{\partial t} + v_g \cdot \nabla \eta \tag{11}$$

Finally, we evolve the potential vorticity forward in time using the potential vorticity conservation equation:

$${\color{blue}\boxed{\frac{\partial q_g}{\partial t} + v_g \cdot \nabla q_g = 0}} \tag{12}$$

# Anelastic Isentropic Case

Extending to three dimensions using isentropic coordinates, the potential vorticity becomes:

$$q = \frac{\zeta_a}{\sigma} = \frac{1}{\sigma}\left(\frac{\partial v}{\partial x} - \frac{\partial u}{\partial y} + f\right) \tag{13}$$

The geostrophic velocity components in isentropic coordinates rely on the Montgomery potential ($M$):

$$u_g = -\frac{1}{f}\frac{\partial M}{\partial y}, \quad v_g = \frac{1}{f}\frac{\partial M}{\partial x} \tag{14}$$

The Montgomery potential is related to the isentropic density ($\sigma$) via:

$$\frac{\partial^2 M}{\partial \theta^2} = -\frac{N^2}{\sigma_0 \Gamma^2}\sigma \tag{15}$$

Combining equations 13, 14, and 15 provides the geostrophic approximation to the potential vorticity equation:

$$f + \frac{\partial}{\partial x}\left(\frac{1}{f}\frac{\partial M}{\partial x}\right) + \frac{\partial}{\partial y}\left(\frac{1}{f}\frac{\partial M}{\partial y}\right) + \frac{\sigma_0 \Gamma^2 q_g}{N^2}\frac{\partial^2 M}{\partial \theta^2} = 0 \tag{16}$$

By linearizing this equation (assuming $\sigma = \sigma_0(\theta) + \sigma^*$ and $M = M_0(\theta) + M^*$), we obtain the core 3D PV inversion equation:

$${\color{lime}\boxed{\frac{1}{f_0^2}\nabla_h^2 M^* + \frac{\Gamma^2}{N^2}\frac{\partial^2 M^*}{\partial \theta^2} = \frac{q^*}{q_0}}} \tag{17}$$

where the base state part of the potential vorticity is:

$$q_A = q_0\left(1 + \frac{\beta y}{f_0}\right) \tag{18}$$

To solve these, we need a lower boundary condition derived from the geopotential diagnostic equation:

$$\Phi = M - \theta\frac{\partial M}{\partial \theta} \tag{19}$$

Evaluating this at the lower boundary $\theta_B$ gives:

$$\Phi_B = M_B - \theta_B\left(\frac{\partial M}{\partial \theta}\right)_B \tag{20}$$

Because the lower boundary varies, we use a first-order Taylor series expansion at a fixed reference level $\theta_{B0}$:

$$\Phi_B \approx \Phi_{B0} + \left(\frac{\partial \Phi_0}{\partial \theta}\right)_{B0}\theta_B^* \approx M_{B0} - \theta_{B0}\left(\frac{\partial M}{\partial \theta}\right)_{B0} + \left(\frac{d\Phi_0}{d\theta}\right)_{B0}\theta_B^* \tag{21}$$

The base state satisfies:

$$\Phi_{0B} = M_{0B0} - \theta_{B0}\left(\frac{dM_0}{d\theta}\right)_{B0} \tag{22}$$

Subtracting the base state equation from the Taylor expansion produces the linearized lower boundary condition:

$$\Phi_B^* = M_{B0}^* - \theta_{B0}\left(\frac{\partial M^*}{\partial \theta}\right)_{B0} + \left(\frac{d\Phi_A}{d\theta}\right)_{B0}\theta_B^* \tag{23}$$

The linearized mass continuity equation governing the velocity potential $\chi$ is:

$$\nabla_h^2 \chi = \frac{1}{\sigma_0}\left(\frac{\partial \sigma^*}{\partial t} + v_g \cdot \nabla_h \sigma^*\right) \tag{24}$$

where the density perturbation is derived from the Montgomery potential:

$$\sigma^* = -\frac{\sigma_0 \Gamma^2}{N^2}\frac{\partial^2 M^*}{\partial \theta^2} \tag{25}$$

The potential vorticity is advected only by the geostrophic wind:

$${\color{lime}\boxed{\frac{\partial q_g}{\partial t} + v_g \cdot \nabla_h q_g = 0}} \tag{26}$$

And the surface temperature perturbation obeys:

$$\frac{\partial \theta_B^*}{\partial t} + v_{gB0} \cdot \nabla_h \theta_B^* = 0 \tag{27}$$

## Boussinesq Isentropic Case

Applying the Boussinesq approximation (where $\Gamma \rightarrow \Gamma_R$ and $N \rightarrow N_R$ are constants), the PV inversion equation simplifies to:

$${\color{orange}\boxed{\frac{1}{f_0^2}\nabla_h^2 M^* + \frac{\Gamma_R^2}{N_R^2}\frac{\partial^2 M^*}{\partial \theta^2} = \frac{q^*}{q_0}}} \tag{28}$$

The Montgomery potential and isentropic density relationship also simplifies:

$$\frac{\partial^2 M}{\partial \theta^2} = -\frac{N_R^2 \sigma}{\Gamma_R^2 \sigma_0} \tag{29}$$

And the geopotential diagnostic uses constant reference values $M_R$ and $\theta_R$:

$$\Phi = M_R - \theta_R\frac{\partial M}{\partial \theta} \tag{30}$$

By picking a reference value for the Exner function near the surface ($\Pi_R = C_p$) and setting $\Phi=0$ there, the diagnostic equation becomes:

$$\Phi = \theta_R\left(C_p - \frac{\partial M}{\partial \theta}\right) = \theta_R(C_p - \Pi) \tag{31}$$

Revisiting the surface boundary condition using this formulation yields:

$$\Phi_B = \Phi_{B0} + \left(\frac{\partial \Phi_0}{\partial \theta}\right)_{B0}\theta_B^* = \theta_R\left[C_p - \left(\frac{\partial M}{\partial \theta}\right)_{B0}\right] + \frac{\theta_R N_R^2}{\Gamma_R^2}\theta_B^* \tag{32}$$

For the base state, this is:

$$\Phi_{AB} = \theta_R\left[C_p - \left(\frac{\partial M_0}{\partial \theta}\right)_{B0}\right] \tag{33}$$

Subtracting the base state equation from the full boundary condition gives the final condition on the perturbation quantities:

$$\Phi_B^* = -\theta_R\left(\frac{\partial M^*}{\partial \theta}\right)_{B0} + \frac{\theta_R N_R^2}{\Gamma_R^2}\theta_B^* \tag{34}$$

# Q&A

### 1. Imagine a point vortex in purely two-dimensional, horizontal flow, where the vorticity field is given by $\zeta_z=C\delta(x)\delta(y)$ with C a constant equal to the strength of the vortex. The quantity $\delta()$ is the Dirac delta function. Solve for the horizontal velocity on an infinite domain. Hint: Use cylindrical symmetry and the Kelvin circulation theorem applied to a circular loop centered on the vortex to obtain the velocity field. If you have experience with electromagnetism, think of the problem of the magnetic field surrounding an infinite wire carrying a current.

Using cylindrical symmetry, apply the Kelvin circulation theorem to a circular loop of radius $r$ centered on the vortex:$$\oint v \cdot dl = \iint \zeta_z dA$$Given $\zeta_z = C\delta(x)\delta(y)$, the area integral evaluates to $C$. Evaluating the line integral gives $v_\theta(2\pi r) = C$, which means $v_\theta = \frac{C}{2\pi r}$. In Cartesian coordinates, the horizontal velocity components are:$$u = -\frac{Cy}{2\pi(x^2+y^2)}, \quad v = \frac{Cx}{2\pi(x^2+y^2)}$$

### 2. Consider a two-dimensional flow which is stationary except for the flows associated with two point vortices of equal but opposite strength $\pm C$ separated by a distance d. Describe the speed and direction of motion of the two vortices.

Each vortex moves based solely on the velocity field induced by the other vortex. Because they are separated by a distance $d$, each induces a velocity magnitude of $V = \frac{C}{2\pi d}$ on the other. Because their strengths are equal but opposite, the induced velocities point in the exact same direction. The two vortices will travel together in a straight line (perpendicular to the axis connecting them) at a constant speed of $V = \frac{C}{2\pi d}$.

### 3. Repeat the above problem for the case in which the two vortices have strength of the same sign and magnitude.

Similar to the previous problem, each vortex induces a velocity magnitude of $V = \frac{C}{2\pi d}$ on the other. However, because they have the same sign, the induced velocities point in opposite directions. The two vortices will rotate around their common center of mass (the midpoint between them) in a circular orbit with a tangential speed of $V = \frac{C}{2\pi d}$.

### 4. Given the potential vorticity distribution $q_g=q_0[1+\epsilon\sin(kx)]$ in the shallow water case on an f-plane with no terrain:

#### (a) Invert to obtain the fractional thickness perturbation $\eta$ and the geostrophic velocity $v_g$.

The perturbation potential vorticity is $q^* = q_0\epsilon\sin(kx)$. Using the shallow water inversion equation $L_R^2\nabla^2\eta - \eta = q^*/q_0$, we substitute $q^*/q_0$ and solve:$$\eta(x) = -\frac{\epsilon}{1 + k^2 L_R^2} \sin(kx)$$Using $u_g = 0$ and $v_g = f_0 L_R^2 \frac{\partial \eta}{\partial x}$:$$v_g = -\frac{f_0 L_R^2 \epsilon k}{1 + k^2 L_R^2} \cos(kx)$$


#### (b) From the geostrophic velocity, obtain the relative vorticity $\zeta$.

The relative vorticity is $\zeta = \frac{\partial v_g}{\partial x} - \frac{\partial u_g}{\partial y}$:$$\zeta = \frac{f_0 L_R^2 \epsilon k^2}{1 + k^2 L_R^2} \sin(kx)$$

#### (c) Sketch plots of the maximum amplitude of $\eta/\epsilon$ and $\zeta/(\epsilon f)$ vs $kL_R$ and comment on the relative importance of thickness perturbations to vorticity in small-scale $(kL_R\gg1)$ and large-scale $(kL_R\ll1)$ disturbances

The amplitude ratio of thickness $\eta/\epsilon$ is $\frac{1}{1 + (kL_R)^2}$, and the amplitude ratio of vorticity $\zeta/(\epsilon f_0)$ is $\frac{(kL_R)^2}{1 + (kL_R)^2}$. For small-scale disturbances ($kL_R \gg 1$), relative vorticity dominates while thickness perturbations are negligible. For large-scale disturbances ($kL_R \ll 1$), thickness perturbations dominate while relative vorticity is negligible.

### 5. Consider a distribution of potential vorticity in the shallow water case of the form $q_g=q_0[1+\epsilon\delta(x)]$ where $\epsilon$ is a constant. Assume no terrain and constant f.

#### (a) Invert to obtain $\eta(x)$ and $v_g(x)$. Sketch these variables as functions of $x/L_R$.

The perturbation PV is $q^* = q_0\epsilon\delta(x)$. Inverting via $L_R^2 \frac{\partial^2\eta}{\partial x^2} - \eta = \epsilon\delta(x)$ yields:$$\eta(x) = -\frac{\epsilon}{2 L_R} e^{-|x|/L_R}$$$$v_g(x) = \frac{f_0 \epsilon}{2} \text{sgn}(x) e^{-|x|/L_R}$$

#### (b) Assume now that $\epsilon=Ct$ where C is a constant, i.e., $\epsilon$ increases linearly with time. The above solution now makes $\eta$ time-dependent. From $\eta$, compute the velocity potential $\chi$ and the slaved ageostrophic wind $v_a$.

If $\epsilon = Ct$, then $\frac{\partial \eta}{\partial t} = -\frac{C}{2 L_R} e^{-|x|/L_R}$. From the continuity equation $\nabla^2\chi = \frac{\partial \eta}{\partial t} + v_g \cdot \nabla \eta$, geostrophic advection vanishes because $u_g = 0$. Integrating $\frac{\partial^2\chi}{\partial x^2}$ provides the ageostrophic wind ($u_a = -\frac{\partial \chi}{\partial x}$):$$u_a(x) = -\frac{C}{2} \text{sgn}(x) e^{-|x|/L_R}$$

#### (c) Comment on the role of $v_a$ in maintaining mass continuity. 

As the PV anomaly grows, the layer thickness decreases. Because the geostrophic wind is purely meridional, it is non-divergent and cannot clear mass out of the trough. The ageostrophic wind ($v_a$) diverges mass away from $x=0$ to satisfy mass continuity, allowing the thickness to drop.

### 6. In the isentropic Boussinesq case with constant f:

#### (a) Rescale the coordinate so that $d\theta=(f\Gamma_R/N_R)d\xi$ in the potential vorticity inversion equation.
Substituting $d\theta = (f_0\Gamma_R/N_R)d\xi$ into the 3D inversion equation normalizes it into a standard 3D Poisson equation:$$\nabla_h^2 M^* + \frac{\partial^2 M^*}{\partial \xi^2} = f_0^2 \frac{q^*}{q_0}$$

#### (b) Ignoring boundary conditions, obtain $M'$ as a function of $r=(x^2+y^2+\xi^2)^{1/2}$ assuming that $q_o'/q_0=C\delta(x)\delta(y)\delta(\xi)$. Hint: Think Coulomb's law.
Analogous to Coulomb's Law for a point source, the solution to the 3D Poisson equation is:$$M^*(r) = -\frac{f_0^2 C}{4\pi r} \quad \text{where} \quad r = (x^2+y^2+\xi^2)^{1/2}$$

#### (c) Compute the geostrophic wind $v_g$. Comment on the dependence of $|v_g|$ on r at the level $\xi=0$ compare with that of a vortex in two-dimensional flow.
Using $v_g = \frac{1}{f_0}\nabla_h M^*$, the magnitude at $\xi=0$ is $|v_g| = \frac{f_0 C}{4\pi r^2}$. The 3D Boussinesq vortex decays much faster ($1/r^2$) compared to the 2D point vortex ($1/r$).

#### (d) Compute the geopotential height perturbation. Is the air warmer or colder than the base state above the potential vorticity anomaly? Take C and f to be positive.

The geopotential perturbation $\Phi^*$ is proportional to the vertical derivative of $M^*$. Above the anomaly ($\xi > 0$), $\Phi^*$ is negative. This negative geopotential perturbation pulls the isentropic surfaces downward. Because lower potential temperature surfaces are depressed, higher potential temperature values exist at a fixed altitude. Therefore, the air is warmer than the base state directly above the PV anomaly.
