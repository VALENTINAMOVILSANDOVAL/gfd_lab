# Symmetric Instability and Rossby Waves

## Symmetric Instability

We begin with the hydrostatic Boussinesq equations in geometric coordinates. We assume that the disturbance consists of circulations in the plane normal to the ambient wind (pointing in the $x$-direction), meaning all variations in $x$ are zero ($\partial/\partial x = 0$)

### Governing Equations

For an incompressible Boussinesq fluid, the mass continuity equation is $\nabla \cdot \vec{u} = 0$. Applying $\partial/\partial x = 0$:

$${\color{blue}\boxed{\frac{\partial v}{\partial y} + \frac{\partial w}{\partial z} = 0}} \tag{1}$$

The momentum equations are $\frac{D\vec{u}}{Dt} + f\hat{k} \times \vec{u} = -\nabla \pi + b\hat{k}$, where $\pi$ is kinematic pressure and $b$ is buoyancy. Expanding the material derivative and applying $\partial/\partial x = 0$:

For the $x$-component:

$${\color{blue}\boxed{\frac{\partial u}{\partial t} + v\frac{\partial u}{\partial y} + w\frac{\partial u}{\partial z} - fv = 0}} \tag{2}$$

For the $y$-component:

$${\color{blue}\boxed{\frac{\partial v}{\partial t} + v\frac{\partial v}{\partial y} + w\frac{\partial v}{\partial z} + \frac{\partial \pi}{\partial y} + fu = 0}} \tag{3}$$

For the $z$-component, assuming hydrostatic balance:

$${\color{blue}\boxed{\frac{\partial \pi}{\partial z} - b = 0}} \tag{4}$$

The thermodynamic equation relies on the material conservation of buoyancy ($\frac{Db}{Dt} = 0$):

$${\color{blue}\boxed{\frac{\partial b}{\partial t} + v\frac{\partial b}{\partial y} + w\frac{\partial b}{\partial z} = 0}} \tag{5}$$

### Base State Equations

We define a steady base state where $v=w=0$, and the wind is sheared: $u = Sz$.
Hydrostatic balance in the base state gives:

$${\color{blue}\boxed{\frac{\partial \pi_0}{\partial z} - b_0 = 0}} \tag{6}$$

Geostrophic balance (setting $y$-momentum to zero) gives:

$${\color{blue}\boxed{\frac{\partial \pi_0}{\partial y} + fSz = 0}} \tag{7}$$

Cross-differentiating (6) and (7) with respect to $y$ and $z$ yields the thermal wind condition. Differentiating (7) by $z$ yields $\frac{\partial^2 \pi_0}{\partial z \partial y} = -fS$. Applying this to the $y$-derivative of (6) yields:

$${\color{blue}\boxed{\frac{\partial b_0}{\partial y} = -fS}} \tag{8}$$

Integrating (8) with respect to $y$, and incorporating the constant Brunt-Väisälä frequency condition ($\frac{\partial b_0}{\partial z} = N^2$), we find $b_0 = -fSy + N^2z$. Integrating (7) with respect to $y$ and matching the $z$-dependence using (6) gives the base state kinematic pressure:

$${\color{blue}\boxed{\pi_0 = -fSyz + N^2z^2/2 \quad , \quad b_0 = -fSy + N^2z}} \tag{9}$$

### Linearized Equations

We decompose the variables into base state and perturbation components ($\pi = \pi_0 + \pi'$, $b = b_0 + b'$, $u = Sz + u'$). Substituting these into (1)-(5) and dropping products of perturbation terms yields the linearized system:
Mass continuity remains unchanged:

$${\color{blue}\boxed{\frac{\partial v}{\partial y} + \frac{\partial w}{\partial z} = 0}} \tag{10}$$

The $x$-momentum equation advects the base wind shear $S = \partial u_0 / \partial z$:

$${\color{blue}\boxed{\frac{\partial u'}{\partial t} + Sw - fv = 0}} \tag{11}$$

The $y$-momentum equation utilizes the geostrophic balance substitution ($\frac{\partial \pi_0}{\partial y} + fSz = 0$):

$${\color{blue}\boxed{\frac{\partial v}{\partial t} + \frac{\partial \pi'}{\partial y} + fu' = 0}} \tag{12}$$

The hydrostatic equation simplifies via base state subtraction:

$${\color{blue}\boxed{\frac{\partial \pi'}{\partial z} - b' = 0}} \tag{13}$$

The buoyancy equation utilizes the known base state gradients $\frac{\partial b_0}{\partial y} = -fS$ and $\frac{\partial b_0}{\partial z} = N^2$:

$${\color{blue}\boxed{\frac{\partial b'}{\partial t} - fSv + N^2w = 0}} \tag{14}$$

### Dispersion Relation

Assuming wave-like solutions proportional to $\exp[i(ly+mz-\omega t)]$  transforms equations (10)-(14) into an algebraic eigenvalue problem:

$${\color{blue}\boxed{\begin{pmatrix} 0 & il & im & 0 & 0 \\ -i\omega & -f & S & 0 & 0 \\ f & -i\omega & 0 & il & 0 \\ 0 & 0 & 0 & im & -1 \\ 0 & -fS & N^2 & 0 & -i\omega \end{pmatrix} \begin{pmatrix} u' \\ v \\ w \\ \pi' \\ b' \end{pmatrix} = 0}} \tag{15}$$

Setting the determinant of this matrix to zero gives the secular equation:

$${\color{blue}\boxed{\omega^2 - \frac{l^2N^2}{m^2} - f^2 - \frac{2lfS}{m} = 0}} \tag{16}$$

By adding and subtracting $l^2S^2/m^2$ on the left, we complete the square. Defining the Richardson number $J = N^2/S^2$, we get:

$${\color{blue}\boxed{\omega^2 = \frac{l^2S^2}{m^2}(J-1) + \left(f + \frac{lS}{m}\right)^2}} \tag{17}$$

In the limit where shear $S \to 0$, equation (16) collapses into the classic dispersion relation for hydrostatic inertia-gravity waves:

$${\color{blue}\boxed{\omega^2 = f^2 + \frac{l^2N^2}{m^2}}} \tag{18}$$

### Buoyancy Perturbation

From (14), assuming instability such that $\omega = i\nu$ , we substitute the critical wavenumber $l_C = -mf/S$ (which sets the squared term in (17) to zero) and use continuity $v = -(m/l)w$  to yield:

$${\color{blue}\boxed{b' = \frac{S^2(1-J)}{\nu}w}} \tag{19}$$

## Rossby Waves in the Shallow Water System

Geostrophic and Potential Vorticity Perturbations

We operate in a channel of width $w$ with a varying bottom height $d/h_0 = \mu y$. The fluid thickness relationship is $h+d = h_0(1+\eta^*)$. Taking horizontal derivatives and applying the geostrophic wind approximations ($v_{gx} = -(g/f)\partial(h+d)/\partial y$ and $v_{gy} = (g/f)\partial(h+d)/\partial x$):

$${\color{red}\boxed{v_{gx} = -\frac{gh_0}{f}\frac{\partial \eta^*}{\partial y} \quad , \quad v_{gy} = \frac{gh_0}{f}\frac{\partial \eta^*}{\partial x}}} \tag{20}$$

Potential vorticity (PV) is $q = f/h$. For a fluid at rest ($\eta_A = -\mu y$), the base PV expands via the binomial approximation to:

$${\color{red}\boxed{q_A = q_0(1 - \eta_A) = q_0(1 + \mu y)}} \tag{21}$$

Adding the motion-driven PV anomaly:

$${\color{red}\boxed{q_g = q_A + q^*}} \tag{22}$$

### Dispersion Relations

By substituting (20) and (22) into the linearized PV conservation and inversion equations, and adopting trial solutions $\propto \sin(\pi y/w)\exp[i(kx-\omega t)]$, the resulting matrix determinant yields the dispersion relation:

$${\color{red}\boxed{\omega = -\frac{kL_R^2f\mu}{1 + L_R^2(k^2 + \pi^2/w^2)}}} \tag{23}$$

We adjust the scales, measuring length in units of the Rossby radius $L_R$ and time by $f^{-1}$. Under this non-dimensionalization ($k \to kL_R$, $w \to w/L_R$, $\mu \to \mu L_R$, $\omega \to \omega/f$), the parameters condense:

$${\color{red}\boxed{\omega = -\frac{\mu k}{1 + k^2 + \pi^2/w^2}}} \tag{24}$$

# Surface Rossby Waves

### Dispersion Relation

For continuous stratification over a surface with an ambient temperature gradient, the dynamics are governed by quasi-geostrophic theory. From the resulting algebraic evaluation $[(\omega - kU)m - k\Lambda]M_S = 0$, we isolate $\omega = kU + k\Lambda/m$. Utilizing the known vertical decay rate $m = N_R(k^2+l^2)^{1/2} / (f\Gamma_R)$ , and assuming the ambient isentropic wind shear parameter $\Lambda$ simplifies out as implied by the text's final solution, we get:

$${\color{green}\boxed{\omega = kU + \frac{f\Gamma_R k}{N_R(k^2+l^2)^{1/2}}}} \tag{25}$$

### Trace Speeds and Group Velocities

The trace speeds are the phase velocities along the coordinate axes ($u_t = \omega/k$ and $v_t = \omega/l$):

$${\color{green}\boxed{u_t = U + \frac{f\Gamma_R}{N_R(k^2+l^2)^{1/2}} \quad , \quad v_t = \frac{k}{l}u_t}} \tag{26}$$

Taking the partial derivatives of the dispersion relation (25) with respect to wavenumbers $k$ and $l$ yields the group velocities:

$${\color{green}\boxed{u_g = U + \frac{f\Gamma_R l^2}{N_R(k^2+l^2)^{3/2}} \quad , \quad v_g = -\frac{f\Gamma_R kl}{N_R(k^2+l^2)^{3/2}}}} \tag{27}$$

# Q&A

### **1. Show that the ambient environment has negative potential vorticity (assuming $f>0$) if it is unstable to symmetric instability.**

The Ertel potential vorticity (PV) is $q = \boldsymbol{\omega}_a \cdot \nabla b_0$. For the given base state wind $u = Sz$, the absolute vorticity vector is $\boldsymbol{\omega}_a = \nabla \times (Sz\hat{i}) + f\hat{k} = S\hat{j} + f\hat{k}$. The ambient buoyancy gradient, given $b_0 = -fSy + N^2z$, is $\nabla b_0 = -fS\hat{j} + N^2\hat{k}$. Taking the dot product yields $q = -fS^2 + fN^2 = fS^2(N^2/S^2 - 1)$. Using the Richardson number $J = N^2/S^2$, we get $q = fS^2(J - 1)$. Since the threshold for symmetric instability is $J < 1$, the term $(J-1)$ is negative. Therefore, if $f > 0$, the potential vorticity $q$ must be negative.

### **2. Redo the analysis of the shallow water Rossby wave with zero terrain but non-zero beta effect.**

With zero terrain, the depth $h_0$ is constant. The beta effect introduces a planetary vorticity gradient $f = f_0 + \beta y$. The base state PV is $q_A = \frac{f_0 + \beta y}{h_0} \approx q_0 + \frac{\beta}{h_0} y$. The linearized PV conservation equation, substituting $v_{gy} = \frac{g h_0}{f_0}\frac{\partial \eta^*}{\partial x}$, becomes $\frac{\partial q^*}{\partial t} + \frac{g \beta}{f_0} \frac{\partial \eta^*}{\partial x} = 0$. Following the same algebraic procedure and matrix determinant as equation 8.34, the topographic gradient factor $\mu f L_R^2$ is replaced by the planetary gradient $\beta L_R^2$. The resulting dispersion relation is:

$$\omega = -\frac{k \beta L_R^2}{1 + L_R^2(k^2 + \pi^2/w^2)}$$

### **3. In zonal symmetry ($\partial/\partial x=0$) on an f-plane, show that the quantity $M\equiv u-fy$ is conserved by parcels. (This works for the shallow water equations and all forms of the stably stratified three-dimensional equations, so take your pick.) Hint: Use $v=dy/dt$.** 

The inviscid horizontal $x$-momentum equation is $\frac{Du}{Dt} - fv = -\frac{1}{\rho}\frac{\partial p}{\partial x}$. Under zonal symmetry ($\partial/\partial x = 0$), the pressure gradient term vanishes, leaving $\frac{Du}{Dt} - fv = 0$. Substituting the kinematic definition $v = \frac{Dy}{Dt}$, we obtain $\frac{Du}{Dt} - f\frac{Dy}{Dt} = 0$. Because $f$ is constant on an f-plane, this can be rewritten as the exact material derivative $\frac{D}{Dt}(u - fy) = 0$. Thus, $M \equiv u - fy$ is materially conserved.

### **4. Repeat the above analysis on an equatorial beta-plane, deriving the form of M in this case.**

On an equatorial beta-plane, the Coriolis parameter is $f = \beta y$. The symmetric $x$-momentum equation becomes $\frac{Du}{Dt} - (\beta y)v = 0$. Substituting $v = \frac{Dy}{Dt}$ yields $\frac{Du}{Dt} - \beta y \frac{Dy}{Dt} = 0$. By the chain rule, $y \frac{Dy}{Dt} = \frac{D}{Dt}(\frac{1}{2}y^2)$. Therefore, the equation becomes $\frac{D}{Dt}(u - \frac{1}{2}\beta y^2) = 0$, meaning the conserved quantity is $M \equiv u - \frac{1}{2}\beta y^2$. 

### **5. Consider a shallow water flow in an east-west channel of north-south width w ($0\le y\le$ w) with flat bottom and with a flow moving uniformly in geostrophic balance to the east at speed U. Assume constant f**

#### (a) For steady flow, show that the fractional thickness perturbation takes the form $\eta_A=-Uy/(fL_R^2)\equiv-\mu y$. We assume that $\mu w\ll1$. 

The geostrophic balance for a steady zonal flow $U$ is $-fU = -g \frac{\partial \eta_A}{\partial y} h_0$. Rearranging for the fractional gradient gives $\frac{\partial \eta_A}{\partial y} = \frac{fU}{g h_0}$. Integrating with respect to $y$ gives $\eta_A = \frac{fU}{g h_0}y$. Using the definition of the Rossby radius $L_R^2 = g h_0 / f^2$, we can substitute $g h_0 = f^2 L_R^2$ to get $\eta_A = \frac{fU}{f^2 L_R^2}y = \frac{U}{f L_R^2}y$. If we define a coordinate system where the slope drops to the north, we assign $\eta_A = -\frac{U}{fL_R^2}y \equiv -\mu y$. 

#### (b) Show that the potential vorticity in this steady flow situation is $q_A(y)=q_0(1+\mu y)$ 

The ambient potential vorticity is $q_A = \frac{f}{h_A} = \frac{f}{h_0(1 + \eta_A)}$. Substituting $\eta_A = -\mu y$ gives $q_A = \frac{f}{h_0(1 - \mu y)}$. Using the binomial expansion for $\mu w \ll 1$, $(1 - \mu y)^{-1} \approx 1 + \mu y$. Therefore, $q_A \approx q_0(1 + \mu y)$, where $q_0 = f/h_0$. 


#### (c) Assume a wave of the form $(q^*,\eta^*)\propto \sin(\pi y/w)\exp[i(kx-\omega t)]$ on this basic flow, where $\eta=\eta_A+\eta^*=-\mu y+\eta^*$, and where the potential vorticity $q=q_A+q^*= q_0(1+\mu y)+q^*$, and find the dispersion relation $\omega=\omega(k)$. Hint: Linearize the potential vorticity evolution equation about a state of uniform motion. 

Linearizing the PV conservation equation about the uniform flow $U$ yields the material derivative $(\frac{\partial}{\partial t} + U\frac{\partial}{\partial x})q^* + v_{gy}\frac{\partial q_A}{\partial y} = 0$. We know $\frac{\partial q_A}{\partial y} = q_0 \mu$ and $v_{gy} = \frac{g h_0}{f}\frac{\partial \eta^*}{\partial x}$. Using the relationship $q^* = q_0(L_R^2 \nabla^2 \eta^* - \eta^*)$  and plugging in the assumed wave forms results in $(-i\omega + ikU)q^* + (ik \frac{g h_0}{f}) q_0 \mu \eta^* = 0$. Substituting $q^*$ in terms of $\eta^*$ and solving for $\omega$ yields:

$$\omega = kU - \frac{k \mu f L_R^2}{1 + L_R^2(k^2 + \pi^2/w^2)}$$


#### (d) Compute the geostrophic flow velocity due to the wave and the total flow velocity.

The geostrophic wave velocities are calculated using $v_{gx}^* = -\frac{g h_0}{f} \frac{\partial \eta^*}{\partial y}$ and $v_{gy}^* = \frac{g h_0}{f} \frac{\partial \eta^*}{\partial x}$. Using the wave form $\eta^* = \eta_0^* \sin(\pi y/w) e^{i(kx-\omega t)}$, the wave velocities are:

$$v_{gx}^* = -\frac{g h_0}{f} \eta_0^* \left(\frac{\pi}{w}\right) \cos\left(\frac{\pi y}{w}\right) e^{i(kx-\omega t)}$$$$v_{gy}^* = \frac{g h_0}{f} \eta_0^* (ik) \sin\left(\frac{\pi y}{w}\right) e^{i(kx-\omega t)}$$

The total flow velocity adds the ambient zonal flow $U$:

$$\vec{v}_{total} = (U + v_{gx}^*) \hat{i} + v_{gy}^* \hat{j}$$
