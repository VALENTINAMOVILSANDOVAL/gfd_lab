# Geophysical Fluid Dynamics

When we study fluid dynamics, we are interested in understanding how fluids move and evolve over time. But what exactly is Geophysical Fluid Dynamics (GFD)? GFD is the application of fluid dynamics to naturally occurring fluids on Earth and other planets, such as the atmosphere and the oceans. Since these fluids are part of the Earth system, they are called *geophysical fluids*. By studying their motion, we can better understand important phenomena such as weather patterns, ocean currents, and climate. Two key features that make geophysical fluids unique are the rotation of the Earth, which affects fluid motion, and stratification, which refers to the layering of fluids with different properties. In this series of notebooks, we will introduce the main concepts of GFD, combining physical intuition, mathematical derivations, and simple examples to help build a solid understanding of the subject.

## 5. Shallow Water Equations

<div style="border-left: 5px solid #1f77b4; background-color: #eaf4ff; padding: 12px; border-radius: 5px;">

Until now, we have focused on the behavior of small-scale disturbances in the atmosphere and ocean, assuming relatively simple background flows. We now turn to a flow with a simple vertical structure. Consider a thin layer of water moving over terrain whose height may vary. We neglect friction and assume that the water velocity does not change with height. We also assume that both the terrain slope and the water surface slope are very small. Under these conditions, the vertical pressure distribution can be described using the hydrostatic equation. This leads to the shallow water equations.

### Derivation of the Shallow Water Equations

#### Mass Continuity (Flux Form)

Define the mass per unit area of a fluid column as $\rho h$, where $\rho$ is the constant fluid density and $h$ is the fluid thickness:

Define the vertically integrated horizontal mass flux as $\rho h \mathbf{v}$, where $\mathbf{v}$ is the horizontal velocity:

Equate the net mass flux leaving a unit area, represented by the divergence $\nabla \cdot (\rho h \mathbf{v})$, to the negative time tendency of the mass within that column, $-\frac{\partial (\rho h)}{\partial t}$:

$$\nabla \cdot (\rho h \mathbf{v}) = -\frac{\partial (\rho h)}{\partial t} \tag{5.7}$$

Rearrange the equation to bring all terms to the left side:

$$\frac{\partial (\rho h)}{\partial t} + \nabla \cdot (\rho h \mathbf{v}) = 0 \tag{5.8}$$

Factor out the constant density $\rho$ from both the partial time derivative and the spatial divergence operator:

$$\rho \frac{\partial h}{\partial t} + \rho \nabla \cdot (h \mathbf{v}) = 0 \tag{5.9}$$

Divide the entire equation by $\rho$ to obtain:

$$\boxed{\frac{\partial h}{\partial t} + \nabla \cdot (h \mathbf{v}) = 0} \tag{5.10}$$

#### Mass Continuity (Advective Form)

Begin with the newly derived flux form of the continuity equation:

$$\frac{\partial h}{\partial t} + \nabla \cdot (h \mathbf{v}) = 0 \tag{5.11}$$

Expand the divergence term $\nabla \cdot (h \mathbf{v})$ using the product rule for vector calculus:

$$\frac{\partial h}{\partial t} + \mathbf{v} \cdot \nabla h + h \nabla \cdot \mathbf{v} = 0 \tag{5.12}$$

Identify the definition of the total time derivative (or material derivative) in two dimensions, which is $\frac{d}{dt} = \frac{\partial}{\partial t} + \mathbf{v} \cdot \nabla$:

Substitute the total time derivative into the expanded equation to collapse the first two terms, yielding:

$$\boxed{\frac{dh}{dt} + h \nabla \cdot \mathbf{v} = 0} \tag{5.13}$$

#### Horizontal Momentum Equation

Start with the advective form of the momentum equation derived from Newton's second law, ignoring the vertical Coriolis component:

$$\frac{d\mathbf{v}}{dt} = -\frac{1}{\rho}\nabla p - f\mathbf{k} \times \mathbf{v} \tag{5.14}$$

Express the pressure $p$ as a function of height $z$ by assuming hydrostatic balance, where $d(x,y)$ is the terrain elevation and pressure is zero at the upper fluid surface:

$$p = g\rho(d + h - z) \tag{5.15}$$

Apply the horizontal gradient operator $\nabla$ to the hydrostatic pressure equation, noting that the vertical coordinate $z$ acts as a constant with respect to horizontal derivatives:

$$\nabla p = \nabla [g\rho(d + h - z)] \tag{5.16}$$

Factor out the constants $g$ and $\rho$ to solve for the horizontal pressure gradient:

$$\nabla p = g\rho\nabla(d + h) \tag{5.17}$$

Substitute this pressure gradient expression back into the starting momentum equation:

$$\frac{d\mathbf{v}}{dt} = -\frac{1}{\rho} [g\rho\nabla(d + h)] - f\mathbf{k} \times \mathbf{v} \tag{5.18}$$

Cancel the density $\rho$ in the first term on the right side of the equation:

$$\frac{d\mathbf{v}}{dt} = -g\nabla(d + h) - f\mathbf{k} \times \mathbf{v} \tag{5.19}$$

Rearrange the terms by moving them to the left side to set the equation to zero, yielding:

$$\boxed{\frac{d\mathbf{v}}{dt} + g\nabla(d + h) + f\mathbf{k} \times \mathbf{v} = 0} \tag{5.20}$$

### Fundamental modes

#### The Secular Equation

Begin with the linearized shallow water equations in component form:

$$
\begin{aligned}
\frac{\partial \eta}{\partial t} + \frac{\partial u}{\partial x} + \frac{\partial v}{\partial y} &= 0 \\
\frac{\partial u}{\partial t} + c^2\frac{\partial \eta}{\partial x} - fv &= 0 \\
\frac{\partial v}{\partial t} + c^2\frac{\partial \eta}{\partial y} + fu &= 0
\end{aligned}
\tag{5.21}
$$

Assume a plane wave space-time dependence of the form $\exp[i(kx - \omega t)]$ for all perturbation variables, implying propagation only in the $x$-direction:

Substitute the spatial and temporal partial derivatives with their plane wave equivalents ($\frac{\partial}{\partial t} \to -i\omega$, $\frac{\partial}{\partial x} \to ik$, and $\frac{\partial}{\partial y} \to 0$):

$$
\begin{aligned}
-i\omega\eta + iku + 0 &= 0 \\
-i\omega u + ikc^2\eta - fv &= 0 \\
-i\omega v + 0 + fu &= 0
\end{aligned}
\tag{5.22}
$$

Express this homogeneous system of linear equations in matrix form:

$$
\begin{pmatrix} 
-i\omega & ik & 0 \\ 
ikc^2 & -i\omega & -f \\ 
0 & f & -i\omega 
\end{pmatrix}
\begin{pmatrix} \eta \\ u \\ v \end{pmatrix} = 0
\tag{5.23}
$$

Set the determinant of the coefficient matrix to zero to find the non-trivial solutions (the secular equation):

$$
\det \begin{pmatrix} 
-i\omega & ik & 0 \\ 
ikc^2 & -i\omega & -f \\ 
0 & f & -i\omega 
\end{pmatrix} = 0
\tag{5.24}
$$

Expand the determinant along the first row:

$$
-i\omega [(-i\omega)(-i\omega) - (-f)(f)] - ik [(ikc^2)(-i\omega) - (-f)(0)] = 0 \tag{5.25}
$$

Simplify the terms inside the brackets:

$$
-i\omega (-\omega^2 + f^2) - ik (\omega k c^2) = 0 \tag{5.26}
$$

Distribute the exterior terms into the brackets:

$$
i\omega^3 - i\omega f^2 - i\omega k^2 c^2 = 0 \tag{5.27}
$$

Divide the entire equation by $i$ to remove the imaginary unit:

$$
\omega^3 - \omega f^2 - \omega k^2 c^2 = 0 \tag{5.28}
$$

Factor out $\omega$ and rearrange the terms to match:

$$
\boxed{(\omega^2 - k^2c^2 - f^2)\omega = 0} \tag{5.29}
$$


#### Dispersion Relations

Begin with the newly derived secular equation:

$$
(\omega^2 - k^2c^2 - f^2)\omega = 0 \tag{5.30}
$$

Set the isolated $\omega$ term outside the parentheses to zero to find the steady, geostrophically balanced mode:

$$
\boxed{\omega = 0} \tag{5.31}
$$

Set the expression inside the parentheses to zero to find the Poincaré wave modes:

$$
\omega^2 - k^2c^2 - f^2 = 0 \tag{5.32}
$$

Isolate $\omega^2$ on the left side of the equation:

$$
\omega^2 = k^2c^2 + f^2 \tag{5.33}
$$

Take the square root of both sides to obtain:

$$
\boxed{\omega = \pm(k^2c^2 + f^2)^{1/2}} \tag{5.34}
$$

### Kelvin Wave

Begin with the simplified continuity and longitudinal momentum equations for a Kelvin wave, assuming zero transverse velocity ($v=0$):

$$
\begin{aligned}
-i\omega\eta + iku &= 0 \\
-i\omega u + ikc^2\eta &= 0
\end{aligned}
\tag{5.35}
$$

Isolate $u$ in the first equation by moving the $\eta$ term to the right side and dividing by $ik$:

$$u = \frac{\omega}{k}\eta \tag{5.36}$$

Substitute this expression for $u$ into the second equation:

$$-i\omega \left(\frac{\omega}{k}\eta\right) + ikc^2\eta = 0 \tag{5.37}$$

Divide the entire equation by $-i\eta$ to simplify the terms:

$$\frac{\omega^2}{k} - kc^2 = 0 \tag{5.38}$$

Multiply the equation by $k$ to eliminate the denominator:

$$\omega^2 - k^2c^2 = 0 \tag{5.39}$$

Move the $k^2c^2$ term to the right side and take the square root of both sides, and we get the dispersion relation:

$$\boxed{\omega = \pm kc} \tag{5.18}$$

Substitute the newly found dispersion relation ($\omega = \pm kc$) back into the previously isolated equation for $u$ ($u = \frac{\omega}{k}\eta$):

$$u = \frac{\pm kc}{k}\eta \tag{5.40}$$

Cancel the wavenumber $k$ from the numerator and denominator to yield the polarization relation:

$$\boxed{u = \pm c\eta} \tag{5.19}$$

Now, consider the transverse momentum equation for a Kelvin wave evaluated with $v=0$:

$$-\sigma c^2\eta + fu = 0 \tag{5.41}$$

Substitute the derived polarization relation ($u = \pm c\eta$) into this transverse momentum equation:

$$-\sigma c^2\eta + f(\pm c\eta) = 0 \tag{5.42}$$

Divide the entire equation by $c\eta$ to simplify the terms:

$$-\sigma c \pm f = 0 \tag{5.43}$$

Move the $-\sigma c$ term to the right side of the equation:

$$\pm f = \sigma c \tag{5.44}$$

Divide by the characteristic speed $c$ to solve for the decay rate $\sigma$, yielding:

$$\boxed{\sigma = \pm \frac{f}{c}} \tag{5.20}$$

### Equations on a Sphere

#### Mass Continuity on a Sphere

Define the linear dimensions of a rectangular fluid region on a sphere using the earth's radius $a$, longitude $\lambda$, and latitude $\phi$:

$$
\begin{aligned}
\Delta x_{\lambda} &= a \cos \phi \Delta\lambda \\
\Delta x_{\phi} &= a \Delta\phi
\end{aligned}
\tag{5.45}
$$

Express the total volume of this fluid region by multiplying the area by the fluid depth $h$:

$$\text{Volume} = a^2 h \cos \phi \Delta\lambda \Delta\phi \tag{5.46}$$

Equate the time rate of change of this volume to the net flux of fluid entering from the longitudinal and latitudinal boundaries:

$$a^2 \cos \phi \Delta\lambda \Delta\phi \frac{\partial h}{\partial t} = [u(\lambda)h(\lambda) - u(\lambda + \Delta\lambda)h(\lambda + \Delta\lambda)] a \Delta\phi + [v(\phi)h(\phi)\cos \phi - v(\phi + \Delta\phi)h(\phi + \Delta\phi)\cos(\phi + \Delta\phi)] a \Delta\lambda \tag{5.47}$$

Divide the entire equation by the geometric factor $a^2 \cos \phi \Delta\lambda \Delta\phi$ to isolate the time derivative on the left side:

$$\frac{\partial h}{\partial t} = \frac{u(\lambda)h(\lambda) - u(\lambda + \Delta\lambda)h(\lambda + \Delta\lambda)}{a \cos \phi \Delta\lambda} + \frac{v(\phi)h(\phi)\cos \phi - v(\phi + \Delta\phi)h(\phi + \Delta\phi)\cos(\phi + \Delta\phi)}{a \cos \phi \Delta\phi} \tag{5.48}$$

Take the limit as $\Delta\lambda \to 0$ and $\Delta\phi \to 0$, applying the definition of the negative partial derivative to the finite difference terms:

$$\frac{\partial h}{\partial t} = -\frac{1}{a \cos \phi} \frac{\partial}{\partial \lambda}(hu) - \frac{1}{a \cos \phi} \frac{\partial}{\partial \phi}(hv \cos \phi) \tag{5.49}$$

Rearrange the terms by moving them to the left side to set the equation to zero, yielding the mass continuity equation on a sphere:

$$\boxed{\frac{\partial h}{\partial t} + \frac{1}{a \cos \phi} \left[ \frac{\partial}{\partial \lambda}(hu) + \frac{\partial}{\partial \phi}(hv \cos \phi) \right] = 0} \tag{5.23}$$

#### Geostrophic Wind on a Sphere

Begin with the full shallow water momentum equations on a sphere for the longitudinal ($u$) and latitudinal ($v$) components:

$$
\begin{aligned}
\frac{\partial u}{\partial t} + \frac{u}{a \cos \phi}\frac{\partial u}{\partial \lambda} + \frac{v}{a}\frac{\partial u}{\partial \phi} - \frac{uv \tan \phi}{a} + \frac{g}{a \cos \phi}\frac{\partial h}{\partial \lambda} - fv &= 0 \\
\frac{\partial v}{\partial t} + \frac{u}{a \cos \phi}\frac{\partial v}{\partial \lambda} + \frac{v}{a}\frac{\partial v}{\partial \phi} + \frac{u^2 \tan \phi}{a} + \frac{g}{a}\frac{\partial h}{\partial \phi} + fu &= 0
\end{aligned}
\tag{5.50}
$$

Assume a steady, perfectly balanced geostrophic state by dropping all time derivatives, advective terms, and metric terms (which collectively relate to the total material derivative $d\mathbf{v}_h/dt$):

$$
\begin{aligned}
\frac{g}{a \cos \phi}\frac{\partial h}{\partial \lambda} - fv_g &= 0 \\
\frac{g}{a}\frac{\partial h}{\partial \phi} + fu_g &= 0
\end{aligned}
\tag{5.51}
$$

Isolate the latitudinal geostrophic wind component $v_g$ in the first equation by moving the Coriolis term to the right side and dividing by $f$:

$$v_g = \frac{g}{fa \cos \phi} \frac{\partial h}{\partial \lambda} \tag{5.52}$$

Isolate the longitudinal geostrophic wind component $u_g$ in the second equation by moving the pressure gradient term to the right side and dividing by $f$:

$$u_g = -\frac{g}{fa} \frac{\partial h}{\partial \phi} \tag{5.53}$$

Combine these resulting expressions into the boxed geostrophic wind relations:

$$
\boxed{
\begin{aligned}
u_g &= -\frac{g}{fa}\frac{\partial h}{\partial \phi} \\
v_g &= \frac{g}{fa \cos \phi}\frac{\partial h}{\partial \lambda}
\end{aligned}
}
\tag{5.29}
$$

### Beta-Plane Approximation

A Taylor series expansion is made for the Coriolis parameter $f$ about a central latitude $\phi_0$, defining $y$ as the northward coordinate:

$$f \approx 2\Omega\sin\phi_0 + \left(\frac{2\Omega\cos\phi_0}{a}\right)y \equiv f_0 + \beta y \tag{5.30}$$

For the equatorial beta-plane approximation, the latitude $\phi_0 = 0$, resulting in:

$$\boxed{f \approx 2\Omega\phi = \left(\frac{2\Omega}{a}\right)y = \beta y} \tag{5.31}$$

**1. Consider non-rotating, steady, shallow water flow in the $x$ direction.**

**(a) Show that the mass continuity and momentum equations may be integrated in $x$ to yield $uh = M$ and $u^2/2 + g(h + d) = B$ where $M$ and $B$ are constants.**

**A:** For steady ($\partial/\partial t = 0$), non-rotating ($f=0$), 1D flow:  Continuity: $\frac{d(hu)}{dx} = 0$. Integrating with respect to $x$ yields $hu = M$.  Momentum: $u\frac{du}{dx} + g\frac{d(h+d)}{dx} = 0$, which can be rewritten as $\frac{d}{dx}\left(\frac{1}{2}u^2 + g(h+d)\right) = 0$. Integrating yields Bernoulli's equation: $\frac{1}{2}u^2 + g(h+d) = B$.

**(b) Assume that $h = h_0(1 + \eta)$, $d = \delta h_0$, and $u = U(1 + \mu)$. Assume that $\eta$, $\delta$, and $\mu$ all have magnitudes much less than unity and linearize the above equations in these quantities.**

**A:** Substitute $h=h_0(1+\eta)$, $d=h_0\delta$, and $u=U(1+\mu)$ into the integral equations:  Continuity: $Uh_0(1+\mu)(1+\eta) = M$. Ignoring non-linear cross-terms and subtracting the base state ($Uh_0 = M$) leaves $\mu + \eta = 0$, or $\mu = -\eta$.Momentum: $\frac{1}{2}U^2(1+2\mu) + gh_0(1+\eta+\delta) = B$. Subtracting the base state ($\frac{1}{2}U^2 + gh_0 = B$) yields $U^2\mu + gh_0(\eta + \delta) = 0$.

**(c) Defining $gh_0 = c^2$, solve for $\mu$ and $\eta$ in terms of $\delta$.**

**A:** Substitute $gh_0 = c^2$  and $\mu = -\eta$ into the linearized momentum equation:
  $$-U^2\eta + c^2(\eta + \delta) = 0$$$$\eta(c^2 - U^2) = -c^2\delta \implies \eta = \frac{c^2}{U^2 - c^2}\delta$$$$\mu = \frac{-c^2}{U^2 - c^2}\delta$$

**(d) How does the thickness of the flow vary with $\delta$ for ridges ($\delta > 0$) and valleys ($\delta < 0$) for the two cases $U < c$ and $U > c$?**

**A:** 
- Subcritical ($U < c$): The denominator $U^2 - c^2$ is negative. Therefore, $\eta$ has the opposite sign of $\delta$. The fluid surface dips over ridges ($\delta > 0$) and rises over valleys ($\delta < 0$).
- Supercritical ($U > c$): The denominator is positive. Thus, $\eta$ has the same sign as $\delta$. The fluid surface rises over ridges and dips over valleys.

**3. Cylindrical coordinates:**

**(a) Derive the shallow water equations in cylindrical coordinates.**

**A:** Using $\mathbf{v} = u_r\hat{r} + v_\phi\hat{\phi}$ and the derivatives of unit vectors $\frac{\partial\hat{r}}{\partial\phi} = \hat{\phi}$ and $\frac{\partial\hat{\phi}}{\partial\phi} = -\hat{r}$:  Mass Continuity:$$\frac{\partial h}{\partial t} + \frac{1}{r}\frac{\partial (r h u_r)}{\partial r} + \frac{1}{r}\frac{\partial (h v_\phi)}{\partial \phi} = 0$$Radial Momentum:$$\frac{\partial u_r}{\partial t} + u_r\frac{\partial u_r}{\partial r} + \frac{v_\phi}{r}\frac{\partial u_r}{\partial \phi} - \frac{v_\phi^2}{r} - f v_\phi = -g\frac{\partial (h+d)}{\partial r}$$Azimuthal Momentum:$$\frac{\partial v_\phi}{\partial t} + u_r\frac{\partial v_\phi}{\partial r} + \frac{v_\phi}{r}\frac{\partial v_\phi}{\partial \phi} + \frac{u_r v_\phi}{r} + f u_r = -\frac{g}{r}\frac{\partial (h+d)}{\partial \phi}$$

**(b) Determine the balance in the radial momentum equation in the steady, axisymmetric case with zero radial velocity and explain the physical meaning of each term (gradient balance).**

**A:** For steady ($\partial/\partial t = 0$), axisymmetric ($\partial/\partial \phi = 0$) flow with zero radial velocity ($u_r = 0$), the radial momentum equation reduces to:
  $$\frac{v_\phi^2}{r} + f v_\phi = g\frac{\partial (h+d)}{\partial r}$$Physical Meaning: The outward centrifugal force ($\frac{v_\phi^2}{r}$) plus the Coriolis force ($f v_\phi$) perfectly balances the radial pressure gradient force ($g\frac{\partial(h+d)}{\partial r}$).

**(c) Show that conservation of specific angular momentum $L = rv + r^2f/2$ yields the azimuthally symmetric form of the azimuthal momentum equation.**

**A:** The specific angular momentum is $L = rv_\phi + \frac{1}{2}fr^2$. Taking the material derivative:
  $$\frac{dL}{dt} = \frac{dr}{dt}v_\phi + r\frac{dv_\phi}{dt} + fr\frac{dr}{dt}$$Substitute $u_r = \frac{dr}{dt}$ and expand the material derivative of $v_\phi$:$$\frac{dL}{dt} = r\left(\frac{\partial v_\phi}{\partial t} + u_r\frac{\partial v_\phi}{\partial r} + \frac{v_\phi}{r}\frac{\partial v_\phi}{\partial \phi}\right) + u_r v_\phi + fru_r$$$$\frac{dL}{dt} = r\left[\frac{\partial v_\phi}{\partial t} + u_r\frac{\partial v_\phi}{\partial r} + \frac{v_\phi}{r}\frac{\partial v_\phi}{\partial \phi} + \frac{u_r v_\phi}{r} + fu_r\right]$$Setting $\frac{dL}{dt} = 0$ yields the term in brackets equal to zero, which is exactly the azimuthally symmetric form (where RHS $= 0$) of the azimuthal momentum equation derived in part (a).
