# Shallow Water Equations

## Derivation of the Shallow Water Equations

### Mass Continuity

Consider a vertical column of fluid with unit area. The fluid layer has a thickness $h$ and a constant density $\rho$, making the mass per unit area $\rho h$.

The horizontal velocity is defined as $\mathbf{v} = (u,v)$, and the vertically integrated horizontal mass flux is $\rho h \mathbf{v}$.

The mass per unit time leaving this fluid column is the divergence of the mass flux, $\nabla \cdot (\rho h \mathbf{v})$

By conservation of mass, this must equal the negative rate of change of mass inside the column, $-\frac{\partial(\rho h)}{\partial t}$. Equating these gives the mass continuity equation:

$${\color{blue}\boxed{\frac{\partial h}{\partial t} + \nabla \cdot (h \mathbf{v}) = 0}} \tag {1}$$

Note: The constant density $\rho$ cancels out of the equation.

Expanding the divergence term yields $\frac{\partial h}{\partial t} + \mathbf{v} \cdot \nabla h + h \nabla \cdot \mathbf{v} = 0$. By applying $\frac{d}{dt} = \frac{\partial}{\partial t} + \mathbf{v} \cdot \nabla$, we obtain the advective form:

$${\color{blue}\boxed{\frac{dh}{dt} + h \nabla \cdot \mathbf{v} = 0}} \tag {2}$$

### Momentum Equations

Applying Newton's second law directly to the fluid gives the advective form of the momentum equation. Ignoring the vertical Coriolis force component, we write:

$${\color{blue}\boxed{\frac{d\mathbf{v}}{dt} = -\frac{1}{\rho}\nabla p - f\mathbf{k} \times \mathbf{v}}} \tag{3}$$

Assuming hydrostatic balance and zero pressure at the upper fluid surface, the pressure as a function of height $z$ depends on the terrain elevation $d(x,y)$ and the fluid depth $h$:

$$p = g\rho(d + h - z) \tag{4}$$

Taking the horizontal gradient of the pressure yields $\nabla p = g\rho\nabla(d + h)$.

Substituting this pressure gradient into equation 3 provides the full shallow water momentum equation:

$${\color{blue}\boxed{\frac{d\mathbf{v}}{dt} + g\nabla(d + h) + f\mathbf{k} \times \mathbf{v} = 0}} \tag{5}$$

## Linearization

We linearize the equations about a state of rest. The layer thickness $h$ is represented as a constant mean thickness $h_0$ and a fractional thickness perturbation $\eta$:

$$h = h_0(1 + \eta) \tag{6}$$

We define the square of a characteristic speed as:

$$c^2 = gh_0 \tag{7}$$

Substituting equation 6 into the mass continuity equation gives:

$$\frac{\partial \eta}{\partial t} + \frac{\partial u}{\partial x} + \frac{\partial v}{\partial y} = 0 \tag{8}$$

The linearized momentum equations in component form become:

$${\color{orange}\boxed{\frac{\partial u}{\partial t} + c^2\frac{\partial \eta}{\partial x} - fv = 0}} \tag{9}$$

$${\color{orange}\boxed{\frac{\partial v}{\partial t} + c^2\frac{\partial \eta}{\partial y} + fu = 0}} \tag{10}$$

## Dispersion Relation

Assuming a plane wave space-time dependence of the form $\exp[i(kx - \omega t)]$ results in:

$$\begin{pmatrix} -i\omega & ik & 0 \\ ikc^2 & -i\omega & -f \\ 0 & f & -i\omega \end{pmatrix} \begin{pmatrix} \eta \\ u \\ v \end{pmatrix} = 0 \tag{11}$$

Setting the determinant of the coefficients to zero yields the secular equation:

$$(\omega^2 - k^2c^2 - f^2)\omega = 0 \tag{12}$$

The solutions to this equation are:

$${\color{yellow}\boxed{\omega = \pm(k^2c^2 + f^2)^{1/2}}} \tag{13}$$

$${\color{yellow}\boxed{\omega = 0}} \tag{14}$$

## Kelvin Wave Derivation

Assuming $v=0$ everywhere and a trial structure $\exp[i(kx - \omega t) - \sigma y]$, we substitute this into equations 8, 9, and 10:

$$-i\omega\eta + iku = 0 \tag{15}$$

$$-i\omega u + ikc^2\eta = 0 \tag{16}$$

$$-\sigma c^2\eta + fu = 0 \tag{17}$$

Equations 15 and 16 can be solved by elimination to yield the dispersion relation:

$${\color{green}\boxed{\omega = \pm kc}} \tag{18}$$

And the corresponding polarization relation:

$${\color{green}\boxed{u = \pm c\eta}} \tag{19}$$

Substitution of the polarization relation into equation 17 results in a value for the decay rate $\sigma$:

$${\color{green}\boxed{\sigma = \pm \frac{f}{c}}} \tag{20}$$

## Equations on a Sphere

### Mass Continuity

For a rectangular region of fluid on a sphere of radius $a$, the linear dimensions are:

$$\Delta x_\lambda = a \cos\phi \Delta\lambda, \quad \Delta x_\phi = a\Delta\phi \tag{21}$$

The rate at which fluid volume enters this region from the sides is:

$$a^2\cos\phi\Delta\lambda\Delta\phi\frac{\partial h}{\partial t} = [u(\lambda)h(\lambda) - u(\lambda')h(\lambda')]a\Delta\phi + [v(\phi)h(\phi)\cos(\phi) - v(\phi')h(\phi')\cos(\phi')]a\Delta\lambda \tag{22}$$


Dividing by $a^2\cos\phi\Delta\lambda\Delta\phi$ and taking the limit of small $\Delta\lambda$ and $\Delta\phi$ provides the mass continuity equation on a sphere:

$${\color{pink}\boxed{\frac{\partial h}{\partial t} + \frac{1}{a\cos\phi}\left[\frac{\partial}{\partial\lambda}(hu) + \frac{\partial}{\partial\phi}(hv\cos\phi)\right] = 0}} \tag{23}$$

### Geostrophic Wind

The total time derivative of horizontal velocity $\mathbf{v}$ on a sphere must account for changing unit vectors:

$$\frac{d\mathbf{v}}{dt} = \frac{\partial\mathbf{v}}{\partial t} + \frac{d\lambda}{dt}\frac{\partial\mathbf{v}}{\partial\lambda} + \frac{d\phi}{dt}\frac{\partial\mathbf{v}}{\partial\phi} \tag{24}$$

Using $\frac{d\lambda}{dt} = \frac{u}{a\cos\phi}$ and $\frac{d\phi}{dt} = \frac{v}{a}$ , the momentum equations separate into longitudinal and latitudinal components:

$$\frac{\partial u}{\partial t} + \frac{u}{a\cos\phi}\frac{\partial u}{\partial\lambda} + \frac{v}{a}\frac{\partial u}{\partial\phi} - \frac{uv\tan\phi}{a} + \frac{g}{a\cos\phi}\frac{\partial h}{\partial\lambda} - fv = 0 \tag{25}$$

$$\frac{\partial v}{\partial t} + \frac{u}{a\cos\phi}\frac{\partial v}{\partial\lambda} + \frac{v}{a}\frac{\partial v}{\partial\phi} + \frac{u^2\tan\phi}{a} + \frac{g}{a}\frac{\partial h}{\partial\phi} + fu = 0 \tag{26}$$

The geostrophic wind is obtained by dropping all terms related to the total derivative $d\mathbf{v}_h/dt$:

$${\color{pink}\boxed{u_g = -\frac{g}{fa}\frac{\partial h}{\partial\phi} \quad \text{and} \quad v_g = \frac{g}{fa\cos\phi}\frac{\partial h}{\partial\lambda}}} \tag{27}$$

## Beta-Plane Approximation

A Taylor series expansion is made for the Coriolis parameter $f$ about a central latitude $\phi_0$, defining $y$ as the northward coordinate:

$$f \approx 2\Omega\sin\phi_0 + \left(\frac{2\Omega\cos\phi_0}{a}\right)y \equiv f_0 + \beta y \tag{28}$$

For the equatorial beta-plane approximation, the latitude $\phi_0 = 0$, resulting in:

$${\color{purple}\boxed{f \approx 2\Omega\phi = \left(\frac{2\Omega}{a}\right)y = \beta y}} \tag{29}$$

### **1. Consider non-rotating, steady, shallow water flow in the $x$ direction.**

#### (a) Show that the mass continuity and momentum equations may be integrated in $x$ to yield $uh = M$ and $u^2/2 + g(h + d) = B$ where $M$ and $B$ are constants.

For steady ($\partial/\partial t = 0$), non-rotating ($f=0$), 1D flow:  Continuity: $\frac{d(hu)}{dx} = 0$. Integrating with respect to $x$ yields $hu = M$.  Momentum: $u\frac{du}{dx} + g\frac{d(h+d)}{dx} = 0$, which can be rewritten as $\frac{d}{dx}\left(\frac{1}{2}u^2 + g(h+d)\right) = 0$. Integrating yields Bernoulli's equation: $\frac{1}{2}u^2 + g(h+d) = B$.

#### (b) Assume that $h = h_0(1 + \eta)$, $d = \delta h_0$, and $u = U(1 + \mu)$. Assume that $\eta$, $\delta$, and $\mu$ all have magnitudes much less than unity and linearize the above equations in these quantities.

Substitute $h=h_0(1+\eta)$, $d=h_0\delta$, and $u=U(1+\mu)$ into the integral equations:  Continuity: $Uh_0(1+\mu)(1+\eta) = M$. Ignoring non-linear cross-terms and subtracting the base state ($Uh_0 = M$) leaves $\mu + \eta = 0$, or $\mu = -\eta$.Momentum: $\frac{1}{2}U^2(1+2\mu) + gh_0(1+\eta+\delta) = B$. Subtracting the base state ($\frac{1}{2}U^2 + gh_0 = B$) yields $U^2\mu + gh_0(\eta + \delta) = 0$.

#### (c) Defining $gh_0 = c^2$, solve for $\mu$ and $\eta$ in terms of $\delta$.

Substitute $gh_0 = c^2$  and $\mu = -\eta$ into the linearized momentum equation:
  $$-U^2\eta + c^2(\eta + \delta) = 0$$$$\eta(c^2 - U^2) = -c^2\delta \implies \eta = \frac{c^2}{U^2 - c^2}\delta$$$$\mu = \frac{-c^2}{U^2 - c^2}\delta$$

#### (d) How does the thickness of the flow vary with $\delta$ for ridges ($\delta > 0$) and valleys ($\delta < 0$) for the two cases $U < c$ and $U > c$? 

- Subcritical ($U < c$): The denominator $U^2 - c^2$ is negative. Therefore, $\eta$ has the opposite sign of $\delta$. The fluid surface dips over ridges ($\delta > 0$) and rises over valleys ($\delta < 0$).
- Supercritical ($U > c$): The denominator is positive. Thus, $\eta$ has the same sign as $\delta$. The fluid surface rises over ridges and dips over valleys.

### **3. Cylindrical coordinates:**

#### (a) Derive the shallow water equations in cylindrical coordinates.

Using $\mathbf{v} = u_r\hat{r} + v_\phi\hat{\phi}$ and the derivatives of unit vectors $\frac{\partial\hat{r}}{\partial\phi} = \hat{\phi}$ and $\frac{\partial\hat{\phi}}{\partial\phi} = -\hat{r}$:  Mass Continuity:$$\frac{\partial h}{\partial t} + \frac{1}{r}\frac{\partial (r h u_r)}{\partial r} + \frac{1}{r}\frac{\partial (h v_\phi)}{\partial \phi} = 0$$Radial Momentum:$$\frac{\partial u_r}{\partial t} + u_r\frac{\partial u_r}{\partial r} + \frac{v_\phi}{r}\frac{\partial u_r}{\partial \phi} - \frac{v_\phi^2}{r} - f v_\phi = -g\frac{\partial (h+d)}{\partial r}$$Azimuthal Momentum:$$\frac{\partial v_\phi}{\partial t} + u_r\frac{\partial v_\phi}{\partial r} + \frac{v_\phi}{r}\frac{\partial v_\phi}{\partial \phi} + \frac{u_r v_\phi}{r} + f u_r = -\frac{g}{r}\frac{\partial (h+d)}{\partial \phi}$$

#### (b) Determine the balance in the radial momentum equation in the steady, axisymmetric case with zero radial velocity and explain the physical meaning of each term (gradient balance).

For steady ($\partial/\partial t = 0$), axisymmetric ($\partial/\partial \phi = 0$) flow with zero radial velocity ($u_r = 0$), the radial momentum equation reduces to:
  $$\frac{v_\phi^2}{r} + f v_\phi = g\frac{\partial (h+d)}{\partial r}$$Physical Meaning: The outward centrifugal force ($\frac{v_\phi^2}{r}$) plus the Coriolis force ($f v_\phi$) perfectly balances the radial pressure gradient force ($g\frac{\partial(h+d)}{\partial r}$).

#### (c) Show that conservation of specific angular momentum $L = rv + r^2f/2$ yields the azimuthally symmetric form of the azimuthal momentum equation.

The specific angular momentum is $L = rv_\phi + \frac{1}{2}fr^2$. Taking the material derivative:
  $$\frac{dL}{dt} = \frac{dr}{dt}v_\phi + r\frac{dv_\phi}{dt} + fr\frac{dr}{dt}$$Substitute $u_r = \frac{dr}{dt}$ and expand the material derivative of $v_\phi$:$$\frac{dL}{dt} = r\left(\frac{\partial v_\phi}{\partial t} + u_r\frac{\partial v_\phi}{\partial r} + \frac{v_\phi}{r}\frac{\partial v_\phi}{\partial \phi}\right) + u_r v_\phi + fru_r$$$$\frac{dL}{dt} = r\left[\frac{\partial v_\phi}{\partial t} + u_r\frac{\partial v_\phi}{\partial r} + \frac{v_\phi}{r}\frac{\partial v_\phi}{\partial \phi} + \frac{u_r v_\phi}{r} + fu_r\right]$$Setting $\frac{dL}{dt} = 0$ yields the term in brackets equal to zero, which is exactly the azimuthally symmetric form (where RHS $= 0$) of the azimuthal momentum equation derived in part (a).
