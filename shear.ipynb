# Gravity Waves in Shear

## Non-rotating shear flow

Assume a 2D, non-rotating flow ($f=0$) where variables are split into a $z$-dependent base state and a small perturbation:

- Velocity:

$$\mathbf{v}=(U(z)+u^{\prime})\mathbf{i}+w\mathbf{k}$$

- Buoyancy: The base state relates to the Brunt-Väisälä frequency via $db_0/dz=N^{2}$:

$$b=b_{0}(z)+b^{\prime}=\int N^{2}dz+b^{\prime}$$

- Pressure: $\pi=\pi_0(z)+\pi^{\prime}$, satisfying hydrostatic balance $\frac{d\pi_0}{dz}=b_0$

We want to neglect products of perturbation terms (e.g., $u^{\prime}\frac{\partial u^{\prime}}{\partial x}\approx 0$). Then, we apply the advective derivative $\mathbf{v}\cdot\nabla=(U+u^{\prime})\frac{\partial}{\partial x}+w\frac{\partial}{\partial z}$.

- Velocity Advection:
$$\mathbf{v}\cdot\nabla\mathbf{v}\approx\left(U\frac{\partial u^{\prime}}{\partial x}+\frac{dU}{dz}w, U\frac{\partial w}{\partial x}\right)$$

- Buoyancy Advection: Substitute $db_0/dz=N^{2}$:

$$\mathbf{v}\cdot\nabla b\approx U\frac{\partial b^{\prime}}{\partial x}+N^{2}w$$

We replace the linearized terms into the general Boussinesq equations, noting that base states are steady ($\partial_t=0$) and horizontally uniform ($\partial_x=0$).

- Horizontal Momentum ($\frac{\partial u}{\partial t}+(\mathbf{v}\cdot\nabla\mathbf{v})_x=-\frac{\partial \pi}{\partial x}$):

$${\color{yellow}\boxed{\frac{\partial u^{\prime}}{\partial t}+U\frac{\partial u^{\prime}}{\partial x}+\frac{dU}{dz}w+\frac{\partial\pi^{\prime}}{\partial x}=0}} \tag{1}$$

- Vertical Momentum ($\frac{\partial w}{\partial t}+(\mathbf{v}\cdot\nabla\mathbf{v})_z=-\frac{\partial \pi}{\partial z}+b$):
Base state terms cancel due to hydrostatic balance ($\frac{d\pi_0}{dz}=b_0$).

$${\color{yellow}\boxed{\frac{\partial w}{\partial t}+U\frac{\partial w}{\partial x}+\frac{\partial\pi^{\prime}}{\partial z}-b^{\prime}=0}} \tag{2}$$

- Continuity ($\frac{\partial u}{\partial x}+\frac{\partial w}{\partial z}=0$):

Since $\frac{\partial U}{\partial x}=0$:

$${\color{yellow}\boxed{\frac{\partial u^{\prime}}{\partial x}+\frac{\partial w}{\partial z}=0}} \tag{3}$$

- Buoyancy ($\frac{\partial b}{\partial t}+\mathbf{v}\cdot\nabla b=0$):

Substitute the linearized buoyancy advection term:

$${\color{yellow}\boxed{\frac{\partial b^{\prime}}{\partial t}+U\frac{\partial b^{\prime}}{\partial x}+N^{2}w=0}} \tag{4}$$

## Derivation of the Taylor-Goldstein equation

We begin with the linearized governing equations for a non-rotating, sheared environment, defined in the previous section.

Assume that all dependent variables exhibit $x$ and $t$ dependence in the form of a plane wave $\exp[i(kx-\omega t)]$. This allows us to convert the partial derivatives into algebraic multipliers:  

- $\frac{\partial}{\partial t} \rightarrow -i\omega$

- $\frac{\partial}{\partial x} \rightarrow ik$

The advection operator $\frac{\partial}{\partial t} + U\frac{\partial}{\partial x}$ becomes $-i\omega + iUk$. By substituting the phase speed $c = \omega/k$, we can factor this operator into $ik(U-c)$.

Applying these substitutions to our four governing equations yields:

$$ik(U-c)u^{\prime} + \frac{dU}{dz}w + ik\pi^{\prime} = 0 \tag{5}$$

$$ik(U-c)w + \frac{\partial\pi^{\prime}}{\partial z} - b^{\prime} = 0 \tag{6}$$

$$iku^{\prime} + \frac{\partial w}{\partial z} = 0 \tag{7}$$

$$ik(U-c)b^{\prime} + N^{2}w = 0 \tag{8}$$

From the Continuity equation (7), we can isolate the horizontal velocity perturbation $u^{\prime}$:

$$u^{\prime} = \frac{i}{k}\frac{\partial w}{\partial z} \tag{9}$$

From the Buoyancy equation (8), we can isolate the buoyancy perturbation $b^{\prime}$:

$$b^{\prime} = -\frac{N^{2}}{ik(U-c)}w = \frac{iN^{2}}{k(U-c)}w \tag{10}$$

Substitute (9) into (5):

$$ik(U-c)\left(\frac{i}{k}\frac{\partial w}{\partial z}\right) + \frac{dU}{dz}w + ik\pi^{\prime} = 0 \tag{11}$$

Simplify the first term (since $i \cdot i = -1$):

$$-(U-c)\frac{\partial w}{\partial z} + \frac{dU}{dz}w + ik\pi^{\prime} = 0 \tag{12}$$

Solve for $ik\pi^{\prime}$:

$$ik\pi^{\prime} = (U-c)\frac{\partial w}{\partial z} - \frac{dU}{dz}w \tag{13}$$

To employ (6), we need the vertical derivative of the kinematic pressure. Take the derivative of (13) with respect to $z$:

$$ik\frac{\partial\pi^{\prime}}{\partial z} = \frac{\partial}{\partial z}\left[ (U-c)\frac{\partial w}{\partial z} - \frac{dU}{dz}w \right] \tag{14}$$

Apply the product rule to both terms on the right side:

$$ik\frac{\partial\pi^{\prime}}{\partial z} = \left(\frac{dU}{dz}\frac{\partial w}{\partial z} + (U-c)\frac{\partial^{2}w}{\partial z^{2}}\right) - \left(\frac{d^{2}U}{dz^{2}}w + \frac{dU}{dz}\frac{\partial w}{\partial z}\right) \tag{15}$$

The $\frac{dU}{dz}\frac{\partial w}{\partial z}$ terms cancel out, leaving:

$$ik\frac{\partial\pi^{\prime}}{\partial z} = (U-c)\frac{\partial^{2}w}{\partial z^{2}} - \frac{d^{2}U}{dz^{2}}w \tag{16}$$

Now, multiply (6) by $ik$ to clear its denominators:

$$-k^{2}(U-c)w + ik\frac{\partial\pi^{\prime}}{\partial z} - ikb^{\prime} = 0 \tag{17}$$

Substitute(16) and $ikb^{\prime}$ (noting from 10 that $ikb^{\prime} = -\frac{N^{2}}{U-c}w$):

$$-k^{2}(U-c)w + \left[ (U-c)\frac{\partial^{2}w}{\partial z^{2}} - \frac{d^{2}U}{dz^{2}}w \right] - \left( -\frac{N^{2}}{U-c}w \right) = 0 \tag{18}$$

Divide the entire equation (18) by $(U-c)$:

$$-k^{2}w + \frac{\partial^{2}w}{\partial z^{2}} - \frac{1}{U-c}\frac{d^{2}U}{dz^{2}}w + \frac{N^{2}}{(U-c)^{2}}w = 0 \tag{19}$$

Rearrange the terms and factor out $w$:

$$\frac{\partial^{2}w}{\partial z^{2}} + \left[ \frac{1}{c-U}\frac{d^{2}U}{dz^{2}} + \frac{N^{2}}{(c-U)^{2}} - k^{2} \right] w = 0 \tag{20}$$

This simplifies exactly to the single, second-order Taylor-Goldstein differential equation:

$${\color{red}\boxed{\frac{\partial^{2}w}{\partial z^{2}} + m^{2}w = 0}} \tag{21}$$

where $m^{2}(z)$ represents the square of the spatially varying vertical wavenumber:

$${\color{red}\boxed{m^{2}(z) = \frac{1}{c-U}\frac{d^{2}U}{dz^{2}} + \frac{N^{2}}{(c-U)^{2}} - k^{2}}} \tag{22}$$

### **2. Prove that steady, non-damped gravity waves transport no buoyancy vertically.**

From the linearized buoyancy equation (4.7), apply the plane wave assumption to express the buoyancy perturbation $b^{\prime}$ in terms of $w$:$$ik(U-c)b^{\prime} + N^{2}w = 0 \implies b^{\prime} = \frac{iN^{2}}{k(U-c)}w$$The vertical flux of buoyancy is defined analogously to momentum flux as $F_{b} = \frac{\rho_{R}}{4}(b^{\prime}w^{*} + b^{\prime*}w)$. Substituting our expression for $b^{\prime}$ yields:$$F_{b} = \frac{\rho_{R}}{4} \left( \frac{iN^{2}}{k(U-c)}ww^{*} - \frac{iN^{2}}{k(U-c)}w^{*}w \right) = 0$$For steady, non-damped waves ($c$ is strictly real), the terms inside the parentheses perfectly cancel, meaning steady gravity waves transport identically zero buoyancy.

### **3. Show by direct substitution in equation (4.16) that the WKB solutions obey the non-interaction theorem.**

The WKB approximation provides approximate analytical solutions for wave equations in environments where the properties of the medium (like the vertical wavenumber $m$) vary slowly over a single wavelength.

The vertical flux of horizontal momentum is $F_{p} = \frac{\rho_{R}}{4}(u^{\prime}w^{*} + u^{\prime*}w)$.
The WKB solution is $w = \frac{C}{m^{1/2}}\exp(\pm i\int mdz)$. Therefore, $|w|^{2} = ww^{*} = \frac{|C|^{2}}{m}$.
Using the continuity equation $u^{\prime} = \frac{i}{k}\frac{\partial w}{\partial z}$, we take the $z$-derivative of the WKB solution:

$$u^{\prime} = \frac{i}{k} \left[ -\frac{1}{2}m^{-3/2}\frac{dm}{dz} \pm im^{1/2} \right] C\exp(\pm i\int mdz) = \left( \mp \frac{m}{k} - i\frac{m^{\prime}}{2km} \right) w$$Substitute $u^{\prime}$ and its conjugate into the momentum flux equation:$$u^{\prime}w^{*} + u^{\prime*}w = \left( \mp \frac{m}{k} - i\frac{m^{\prime}}{2km} \right) ww^{*} + \left( \mp \frac{m}{k} + i\frac{m^{\prime}}{2km} \right) ww^{*} = \mp \frac{2m}{k}|w|^{2}$$

Substitute $|w|^{2} = \frac{|C|^{2}}{m}$:$$F_{p} = \frac{\rho_{R}}{4} \left( \mp \frac{2m}{k} \frac{|C|^{2}}{m} \right) = \mp \frac{\rho_{R}|C|^{2}}{2k}$$Because $F_{p}$ evaluates to a constant that is entirely independent of $z$, its vertical derivative is $\frac{\partial F_{p}}{\partial z} = 0$, satisfying the non-interaction theorem.

### **4. Use the results of your solution for upward-propagating gravity waves from the last chapter to determine the upward flux of horizontal momentum in these waves. Where does this momentum come from?**

For a constant-$N$, constant-$U$ environment, an upward-propagating gravity wave takes the form $w = C\exp(imz)$.Using continuity ($u^{\prime} = \frac{i}{k}\frac{\partial w}{\partial z}$), we find $u^{\prime} = -\frac{m}{k}w$.Substituting this into the momentum flux equation $F_{p} = \frac{\rho_{R}}{4}(u^{\prime}w^{*} + u^{\prime*}w)$:$$F_{p} = \frac{\rho_{R}}{4}\left(-\frac{m}{k}ww^{*} - \frac{m}{k}w^{*}w\right) = -\frac{\rho_{R}m}{2k}|w|^{2}$$Where does it come from? This momentum is extracted from the mean flow or boundaries at the specific altitude where the waves are generated (e.g., by surface topography or convection) and is transported until it is deposited at the altitude where the waves break, dissipate, or are absorbed.

### **5. Consider a parallel, stratified shear flow with $N$ constant and $U = Sz$ with the shear $S$ constant. Note that the Richardson number in this case is $Ri = N^2/S^2$.**

#### (a) Obtain solutions to the Taylor-Goldstein equation (4.8) for the limit in which $k^2$ is small enough to be ignored. Don't worry about boundary conditions. Hint: Try $w = (c - Sz)^X$ where $X$ is to be determined.

With $U=Sz$, $U^{\prime\prime}=0$. Ignoring $k^{2}$, the Taylor-Goldstein equation (4.8) becomes:$$\frac{\partial^{2}w}{\partial z^{2}} + \frac{N^{2}}{(c-Sz)^{2}}w = 0$$Substitute the trial solution $w = (c-Sz)^{X}$. The second derivative is $w^{\prime\prime} = S^{2}X(X-1)(c-Sz)^{X-2}$.$$S^{2}X(X-1)(c-Sz)^{X-2} + N^{2}(c-Sz)^{X-2} = 0$$
Dividing by $S^{2}(c-Sz)^{X-2}$ leaves $X(X-1) + \frac{N^{2}}{S^{2}} = 0$. Since $J = N^{2}/S^{2}$, we solve the quadratic $X^{2} - X + J = 0$:

$$X = \frac{1 \pm \sqrt{1 - 4J}}{2} = \frac{1}{2} \pm i\sqrt{J - \frac{1}{4}}$$The exact solution is $w = (c-Sz)^{\frac{1}{2} \pm i\sqrt{J - 1/4}}$

#### (b) Determine how the character of the solutions change as the Richardson number changes from $Ri > 1/4$ to $Ri < 1/4$.

- $J > 1/4$: The roots are complex. The solution becomes $w = \sqrt{c-Sz} \exp\left[\pm i\sqrt{J - 1/4} \ln(c-Sz)\right]$. These are purely oscillatory traveling waves whose vertical wavelength shrinks infinitely as they approach the critical level ($c=Sz$).
- $J < 1/4$: The roots are real. The solution becomes purely exponential, meaning the waves do not propagate vertically but either decay or grow in amplitude, indicating potential shear instability.

#### (c) Compare your solutions to WKB solutions to this problem. When are they approximately the same? 

For WKB, the vertical wavenumber is $m = \frac{N}{c-Sz}$. The phase integral is $\int m dz = -\frac{N}{S}\ln(c-Sz) = -\sqrt{J}\ln(c-Sz)$. The amplitude factor is $C m^{-1/2} = C^{\prime}(c-Sz)^{1/2}$.The WKB solution is:$$w_{WKB} = C^{\prime}(c-Sz)^{1/2 \pm i\sqrt{J}}$$Comparing the exponents, the exact solution relies on $\sqrt{J - 1/4}$, whereas the WKB approximation relies solely on $\sqrt{J}$. The WKB solution is approximately the same as the exact solution when $J \gg 1/4$ (when stratification strongly dominates over shear).
