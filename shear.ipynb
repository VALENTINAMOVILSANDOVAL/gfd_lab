# Geophysical Fluid Dynamics

When we study fluid dynamics, we are interested in understanding how fluids move and evolve over time. But what exactly is Geophysical Fluid Dynamics (GFD)? GFD is the application of fluid dynamics to naturally occurring fluids on Earth and other planets, such as the atmosphere and the oceans. Since these fluids are part of the Earth system, they are called *geophysical fluids*. By studying their motion, we can better understand important phenomena such as weather patterns, ocean currents, and climate. Two key features that make geophysical fluids unique are the rotation of the Earth, which affects fluid motion, and stratification, which refers to the layering of fluids with different properties. In this series of notebooks, we will introduce the main concepts of GFD, combining physical intuition, mathematical derivations, and simple examples to help build a solid understanding of the subject.

## 4. Gravity Waves in Shear

---
### Non-rotating shear flow

We now study the special case of gravity waves in a non-rotating, sheared environment, utilizing the linearized Boussinesq equations with $f=0$. We assume an ambient horizontal wind in the $x$-direction, $U$, which is a function of height alone. The total wind is defined as:
$$u = U(z) + u^{\prime} \tag{4.0a}$$
where $u^{\prime}$ is a small perturbation about this base state. Since rotation is neglected, the $y$-component of the wind is decoupled from the dynamics of a non-rotating wave exhibiting slab symmetry in the $x-z$ plane, and is thus ignored.

We assume the buoyancy takes the form of a base state plus perturbation:
$$b=b_{0}(z)+b^{\prime}=\int N^{2}dz+b^{\prime} \tag{4.1}$$
where the Brunt-Väisälä frequency $N(z)$ is not necessarily constant. The kinematic pressure follows a similar assumption, where the base state $\pi_{0}(z)$ is in hydrostatic balance with $b_{0}$:
$$\pi=\pi_{0}(z)+\pi^{\prime} \tag{4.1a}$$

The $z$-dependence of the ambient wind $U$ requires a careful linearization of the velocity and buoyancy advection terms:
$$v\cdot\nabla v \approx \left(U\frac{\partial u^{\prime}}{\partial x}+\frac{dU}{dz}w, U\frac{\partial w}{\partial x}\right) \tag{4.2}$$
$$v\cdot\nabla b \approx U\frac{\partial b^{\prime}}{\partial x}+N^{2}w \tag{4.3}$$

#### Linearized Governing Equations

We replace the linearized terms into the general Boussinesq equations, noting that base states are steady ($\partial_t=0$) and horizontally uniform ($\partial_x=0$).

- **Horizontal Momentum** ($\frac{\partial u}{\partial t}+(\mathbf{v}\cdot\nabla\mathbf{v})_x=-\frac{\partial \pi}{\partial x}$):

$$\boxed{\frac{\partial u^{\prime}}{\partial t}+U\frac{\partial u^{\prime}}{\partial x}+\frac{dU}{dz}w+\frac{\partial\pi^{\prime}}{\partial x}=0} \tag{4.4}$$

- **Vertical Momentum** ($\frac{\partial w}{\partial t}+(\mathbf{v}\cdot\nabla\mathbf{v})_z=-\frac{\partial \pi}{\partial z}+b$):
Base state terms cancel due to hydrostatic balance ($\frac{d\pi_0}{dz}=b_0$).

$$\boxed{\frac{\partial w}{\partial t}+U\frac{\partial w}{\partial x}+\frac{\partial\pi^{\prime}}{\partial z}-b^{\prime}=0} \tag{4.5}$$

- **Continuity** ($\frac{\partial u}{\partial x}+\frac{\partial w}{\partial z}=0$):
Since $\frac{\partial U}{\partial x}=0$:

$$\boxed{\frac{\partial u^{\prime}}{\partial x}+\frac{\partial w}{\partial z}=0} \tag{4.6}$$

- **Buoyancy** ($\frac{\partial b}{\partial t}+\mathbf{v}\cdot\nabla b=0$):
Substitute the linearized buoyancy advection term:

$$\boxed{\frac{\partial b^{\prime}}{\partial t}+U\frac{\partial b^{\prime}}{\partial x}+N^{2}w=0} \tag{4.7}$$

---

### Taylor-Goldstein Equation

To isolate the vertical structure of these waves, we assume that all dependent variables exhibit $x$ and $t$ dependence in the form of a plane wave:
$$\exp[i(kx-\omega t)] \tag{4.7a}$$
This transforms the horizontal and temporal derivatives into algebraic multipliers:
$$\frac{\partial}{\partial x} \rightarrow ik, \quad \frac{\partial}{\partial t} \rightarrow -i\omega \tag{4.7b}$$

Applying these transformations to the continuity equation (4.6) and rearranging allows us to express the horizontal velocity perturbation in terms of the vertical velocity gradient:
$$iku^{\prime} + \frac{\partial w}{\partial z} = 0 \implies u^{\prime} = \frac{i}{k}\frac{\partial w}{\partial z} \tag{4.7c}$$

Applying the plane wave transformations to the thermodynamic equation (4.7) yields an algebraic relation for buoyancy:
$$-i\omega b^{\prime} + ikUb^{\prime} + N^2w = 0 \tag{4.7d}$$
We factor out $ik$ and define the horizontal trace speed $c = \omega/k$:
$$ik(U-c)b^{\prime} + N^2w = 0 \implies b^{\prime} = \frac{iN^2}{k(U-c)}w \tag{4.7e}$$

Next, we substitute the transformations and equation (4.7c) into the $x$-momentum equation (4.4) to isolate the pressure perturbation $\pi^{\prime}$:
$$-i\omega\left(\frac{i}{k}\frac{\partial w}{\partial z}\right) + ikU\left(\frac{i}{k}\frac{\partial w}{\partial z}\right) + \frac{dU}{dz}w + ik\pi^{\prime} = 0 \tag{4.7f}$$
Grouping terms and multiplying by $k/i$ yields:
$$(U-c)\frac{\partial w}{\partial z} - \frac{dU}{dz}w = ik\pi^{\prime} \tag{4.7g}$$

We differentiate this expression with respect to $z$ to find $\frac{\partial\pi^{\prime}}{\partial z}$:
$$ik\frac{\partial\pi^{\prime}}{\partial z} = \frac{dU}{dz}\frac{\partial w}{\partial z} + (U-c)\frac{\partial^2 w}{\partial z^2} - \frac{d^2 U}{dz^2}w - \frac{dU}{dz}\frac{\partial w}{\partial z} \tag{4.7h}$$
$$ik\frac{\partial\pi^{\prime}}{\partial z} = (U-c)\frac{\partial^2 w}{\partial z^2} - \frac{d^2 U}{dz^2}w \tag{4.7i}$$

Now, we apply the plane wave transformations to the $z$-momentum equation (4.5) and multiply the entire equation by $ik$:
$$ik\left[ik(U-c)w + \frac{\partial\pi^{\prime}}{\partial z} - b^{\prime}\right] = 0 \tag{4.7j}$$
Substitute $\frac{\partial\pi^{\prime}}{\partial z}$ from equation (4.7i) and $b^{\prime}$ from equation (4.7e):
$$-k^2(U-c)w + (U-c)\frac{\partial^2 w}{\partial z^2} - \frac{d^2 U}{dz^2}w - ik\left[\frac{iN^2}{k(U-c)}w\right] = 0 \tag{4.7k}$$

Dividing the entire expression by $(U-c)$ and rearranging the terms leads directly to a single, second-order differential equation in $w$:
$$\boxed{ \frac{\partial^{2}w}{\partial z^{2}}+m^{2}w=0 } \tag{4.8}$$
where we identify the square of the spatially varying vertical wavenumber $m$ as:
$$\boxed{ m^{2}(z)=\frac{1}{c-U}\frac{d^{2}U}{dz^{2}}+\frac{N^{2}}{(c-U)^{2}}-k^{2} } \tag{4.9}$$
This result is known as the Taylor-Goldstein equation, and the $z$-dependence of $m$ means it is analytically soluble in exact form in only a few special cases.

---

### The WKB Approximation

Now that we have the Taylor-Goldstein equation, we face a mathematical wall: because the vertical wavenumber $m$ depends on height $z$, the equation is rarely soluble in an exact, analytic form. 

To get around this, we use the **WKB Approximation** (named after Wentzel, Kramers, and Brillouin). The physical intuition is simple: if the background environment (wind and stability) changes very *slowly* compared to the vertical wavelength of the gravity wave, the wave essentially "feels" like it is traveling through a constant environment at any given moment. 

Mathematically, this requires the condition:
$$mZ \gg 1$$
where $Z$ is the vertical distance over which the environment changes significantly. By applying this assumption and dropping negligible terms, we can find an approximate solution for the vertical velocity structure:
$$\boxed{w(x,z,t) \approx \frac{C}{m^{1/2}} \exp\left[ \pm i \int m dz + i(kx - \omega t) \right]}$$

**When does it fail?** The WKB approximation breaks down at "turning points" where $m^2$ changes sign (often causing waves to reflect rather than propagate). However, it remarkably remains an adequate approximation near critical levels where the wind speed matches the horizontal trace speed ($U = c$).

---

### The Non-Interaction Theorem (Part 1)

Once we understand how these waves propagate through shear, we must ask: *how do they affect the background wind they are traveling through?* In 1961, Eliassen and Palm established a fundamental dynamic property of gravity waves known as the **Non-Interaction Theorem**. 

The theorem states that if a gravity wave is steady (not growing or decaying in time) and non-damped (no friction or heat loss), it travels through the background shear flow **without extracting or depositing any momentum into it**. 

Think of the wave as a sealed delivery truck carrying momentum. It takes momentum from the level where it is generated (like mountains at the surface), drives through the atmosphere without interacting with the background wind, and only unloads its momentum where it finally breaks or dissipates higher up. 

Mathematically, this means the vertical flux of horizontal momentum ($F_p$) remains perfectly constant with height:
$$\boxed{ \frac{\partial F_p}{\partial z} = 0 }$$

Through a similar mathematical proof, it can also be shown that these steady, non-damped gravity waves transport exactly zero buoyancy vertically. They are pure transporters of momentum.

---

### Shear Instability

If the wind shear in the atmosphere becomes incredibly strong, the previously stable atmospheric profile can break down into turbulence and instability. You can think of this as a battle between two competing forces: 
1. **The atmosphere's natural stability** (which requires potential energy to overcome as it tries to push a displaced parcel of air back to its starting point).
2. **The background wind shear** (which provides excess kinetic energy that tries to forcefully roll or overturn the air).

This competition is measured by a famous dimensionless ratio called the **Richardson number** ($Ri$ or $J$). 

A simplified energy argument might suggest that if the Richardson number drops below **1**, the shear has enough energy to overcome the stability and cause the flow to spontaneously overturn. However, the rigorous **Miles-Howard Theorem** proves a much stricter, mathematical rule: for a shear flow to become unstable, the Richardson number must drop strictly below **1/4** somewhere in the vertical profile:
$$\boxed{J = \frac{N^2}{(dU/dz)^2} < \frac{1}{4}}$$

If the Richardson number is $\ge$ **1/4** everywhere, the flow is mathematically guaranteed to remain stable. 

**Where does it break?** When this instability does occur, it is highly likely to be triggered near a **critical level**—a specific altitude where the horizontal phase speed of the gravity wave perfectly matches the speed of the background wind ($c_r = U$).

**2. Prove that steady, non-damped gravity waves transport no buoyancy vertically.**

**A:** From the linearized buoyancy equation (4.7), apply the plane wave assumption to express the buoyancy perturbation $b^{\prime}$ in terms of $w$:$$ik(U-c)b^{\prime} + N^{2}w = 0 \implies b^{\prime} = \frac{iN^{2}}{k(U-c)}w$$The vertical flux of buoyancy is defined analogously to momentum flux as $F_{b} = \frac{\rho_{R}}{4}(b^{\prime}w^{*} + b^{\prime*}w)$. Substituting our expression for $b^{\prime}$ yields:$$F_{b} = \frac{\rho_{R}}{4} \left( \frac{iN^{2}}{k(U-c)}ww^{*} - \frac{iN^{2}}{k(U-c)}w^{*}w \right) = 0$$For steady, non-damped waves ($c$ is strictly real), the terms inside the parentheses perfectly cancel, meaning steady gravity waves transport identically zero buoyancy.

**3. Show by direct substitution in equation (4.16) that the WKB solutions obey the non-interaction theorem.**

**A:** The WKB approximation provides approximate analytical solutions for wave equations in environments where the properties of the medium (like the vertical wavenumber $m$) vary slowly over a single wavelength.

The vertical flux of horizontal momentum is $F_{p} = \frac{\rho_{R}}{4}(u^{\prime}w^{*} + u^{\prime*}w)$.
The WKB solution is $w = \frac{C}{m^{1/2}}\exp(\pm i\int mdz)$. Therefore, $|w|^{2} = ww^{*} = \frac{|C|^{2}}{m}$.
Using the continuity equation $u^{\prime} = \frac{i}{k}\frac{\partial w}{\partial z}$, we take the $z$-derivative of the WKB solution:

$$u^{\prime} = \frac{i}{k} \left[ -\frac{1}{2}m^{-3/2}\frac{dm}{dz} \pm im^{1/2} \right] C\exp(\pm i\int mdz) = \left( \mp \frac{m}{k} - i\frac{m^{\prime}}{2km} \right) w$$Substitute $u^{\prime}$ and its conjugate into the momentum flux equation:$$u^{\prime}w^{*} + u^{\prime*}w = \left( \mp \frac{m}{k} - i\frac{m^{\prime}}{2km} \right) ww^{*} + \left( \mp \frac{m}{k} + i\frac{m^{\prime}}{2km} \right) ww^{*} = \mp \frac{2m}{k}|w|^{2}$$

Substitute $|w|^{2} = \frac{|C|^{2}}{m}$:$$F_{p} = \frac{\rho_{R}}{4} \left( \mp \frac{2m}{k} \frac{|C|^{2}}{m} \right) = \mp \frac{\rho_{R}|C|^{2}}{2k}$$Because $F_{p}$ evaluates to a constant that is entirely independent of $z$, its vertical derivative is $\frac{\partial F_{p}}{\partial z} = 0$, satisfying the non-interaction theorem.

**4. Use the results of your solution for upward-propagating gravity waves from the last chapter to determine the upward flux of horizontal momentum in these waves. Where does this momentum come from?**

**A:** For a constant-$N$, constant-$U$ environment, an upward-propagating gravity wave takes the form $w = C\exp(imz)$.Using continuity ($u^{\prime} = \frac{i}{k}\frac{\partial w}{\partial z}$), we find $u^{\prime} = -\frac{m}{k}w$.Substituting this into the momentum flux equation $F_{p} = \frac{\rho_{R}}{4}(u^{\prime}w^{*} + u^{\prime*}w)$:$$F_{p} = \frac{\rho_{R}}{4}\left(-\frac{m}{k}ww^{*} - \frac{m}{k}w^{*}w\right) = -\frac{\rho_{R}m}{2k}|w|^{2}$$Where does it come from? **5. Consider a parallel, stratified shear flow with $N$ constant and $U = Sz$ with the shear $S$ constant. Note that the Richardson number in this case is $Ri = N^2/S^2$.**

**(a) Obtain solutions to the Taylor-Goldstein equation (4.8) for the limit in which $k^2$ is small enough to be ignored. Don't worry about boundary conditions. Hint: Try $w = (c - Sz)^X$ where $X$ is to be determined.**

**A:** With $U=Sz$, $U^{\prime\prime}=0$. Ignoring $k^{2}$, the Taylor-Goldstein equation (4.8) becomes:$$\frac{\partial^{2}w}{\partial z^{2}} + \frac{N^{2}}{(c-Sz)^{2}}w = 0$$Substitute the trial solution $w = (c-Sz)^{X}$. The second derivative is $w^{\prime\prime} = S^{2}X(X-1)(c-Sz)^{X-2}$.$$S^{2}X(X-1)(c-Sz)^{X-2} + N^{2}(c-Sz)^{X-2} = 0$$
Dividing by $S^{2}(c-Sz)^{X-2}$ leaves $X(X-1) + \frac{N^{2}}{S^{2}} = 0$. Since $J = N^{2}/S^{2}$, we solve the quadratic $X^{2} - X + J = 0$:

$$X = \frac{1 \pm \sqrt{1 - 4J}}{2} = \frac{1}{2} \pm i\sqrt{J - \frac{1}{4}}$$The exact solution is $w = (c-Sz)^{\frac{1}{2} \pm i\sqrt{J - 1/4}}$

**(b) Determine how the character of the solutions change as the Richardson number changes from $Ri > 1/4$ to $Ri < 1/4$.**

**A:** 

- $J > 1/4$: The roots are complex. The solution becomes $w = \sqrt{c-Sz} \exp\left[\pm i\sqrt{J - 1/4} \ln(c-Sz)\right]$. These are purely oscillatory traveling waves whose vertical wavelength shrinks infinitely as they approach the critical level ($c=Sz$).
- $J < 1/4$: The roots are real. The solution becomes purely exponential, meaning the waves do not propagate vertically but either decay or grow in amplitude, indicating potential shear instability.

**(c) Compare your solutions to WKB solutions to this problem. When are they approximately the same?** 

**A:** For WKB, the vertical wavenumber is $m = \frac{N}{c-Sz}$. The phase integral is $\int m dz = -\frac{N}{S}\ln(c-Sz) = -\sqrt{J}\ln(c-Sz)$. The amplitude factor is $C m^{-1/2} = C^{\prime}(c-Sz)^{1/2}$.The WKB solution is:$$w_{WKB} = C^{\prime}(c-Sz)^{1/2 \pm i\sqrt{J}}$$Comparing the exponents, the exact solution relies on $\sqrt{J - 1/4}$, whereas the WKB approximation relies solely on $\sqrt{J}$. The WKB solution is approximately the same as the exact solution when $J \gg 1/4$ (when stratification strongly dominates over shear).
