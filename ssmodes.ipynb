# Geophysical Fluid Dynamics

When we study fluid dynamics, we are interested in understanding how fluids move and evolve over time. But what exactly is Geophysical Fluid Dynamics (GFD)? GFD is the application of fluid dynamics to naturally occurring fluids on Earth and other planets, such as the atmosphere and the oceans. Since these fluids are part of the Earth system, they are called *geophysical fluids*. By studying their motion, we can better understand important phenomena such as weather patterns, ocean currents, and climate. Two key features that make geophysical fluids unique are the rotation of the Earth, which affects fluid motion, and stratification, which refers to the layering of fluids with different properties. In this series of notebooks, we will introduce the main concepts of GFD, combining physical intuition, mathematical derivations, and simple examples to help build a solid understanding of the subject.

## 3. Sound, Inertia-Gravity Waves, and Lamb Waves

## Resting, non-rotating atmosphere

### The Linearized Continuity Equation

The full, non-linear mass continuity equation is given by equation:

$$\frac{\partial\rho}{\partial t} + v \cdot \nabla\rho + \rho\nabla \cdot v = 0 \tag{1}$$

We expand the density using Taylor series: $\rho = \rho_0(z) + \rho'$. We assume that $\rho(z)$ is at rest, meaning the velocity vector $v = (u, v, w)$ is composed entirely of small perturbation componets. Expanding the equations yields:

$$\frac{\partial(\rho_0 + \rho')}{\partial t} + u\frac{\partial(\rho_0 + \rho')}{\partial x} + v\frac{\partial(\rho_0 + \rho')}{\partial y} + w\frac{\partial(\rho_0 + \rho')}{\partial z} + (\rho_0 + \rho')(\frac{\partial u}{\partial x} + \frac{\partial v}{\partial y} + \frac{\partial w}{\partial z}) = 0 \tag{2}$$

Since $\rho(z)$ is steady (i.e., $\partial\rho_0/\partial t = 0$), and only depends on height $z$ ($\partial\rho_0/\partial x = 0$ and $\partial\rho_0/\partial y = 0$), and because we drop products of primed quantities (like $w\frac{\partial\rho'}{\partial z}$ or $\rho'\frac{\partial u}{\partial x}$) because they are negligibly small, the linearized equation reduces to:

$$\frac{\partial\rho'}{\partial t} + w\frac{d\rho_0}{dz} + \rho_0(\frac{\partial u}{\partial x} + \frac{\partial v}{\partial y} + \frac{\partial w}{\partial z}) = 0 \tag{3}$$

Assuming wave motion only in the $x$-direction due to isotropy in the $x-y$ plane, we can drop the $y$-derivatives, leaving:

$$\frac{\partial\rho'}{\partial t} + w\frac{d\rho_0}{dz} + \rho_0(\frac{\partial u}{\partial x} + \frac{\partial w}{\partial z}) = 0 \tag{4}$$

We divide the entire equation by $\rho_0$:

$$\frac{1}{\rho_0}\frac{\partial\rho'}{\partial t} + \frac{w}{\rho_0}\frac{d\rho_0}{dz} + \frac{\partial u}{\partial x} + \frac{\partial w}{\partial z} = 0 \tag{5}$$


Considering $\rho_0 = \frac{p_R \exp(-z/z_S)}{g z_S}$, and taking the derivative with respect to $z$ yields:

$$\frac{d\rho_0}{dz} = -\frac{1}{z_S}\left[ \frac{p_R \exp(-z/z_S)}{g z_S} \right] = -\frac{\rho_0}{z_S} \tag{6}$$

Therefore, $\frac{1}{\rho_0}\frac{d\rho_0}{dz} = -\frac{1}{z_S}$. Substituting this into our continuity equation:

$$\frac{1}{\rho_0}\frac{\partial\rho'}{\partial t} - \frac{w}{z_S} + \frac{\partial u}{\partial x} + \frac{\partial w}{\partial z} = 0 \tag{7}$$

Let $N = \rho'/\rho_0$. Since $\rho_0$ does not depend on time, $\frac{1}{\rho_0}\frac{\partial\rho'}{\partial t} = \frac{\partial N}{\partial t}$.

Next, assume a plane wave form proportional to $\exp[i(kx - \omega t)]$ for all perturbation variables. This transforms spatial derivatives in $x$ to multiplication by $ik$ and time derivatives to multiplication by $-i\omega$.

$${\color{orange}\boxed{-i\omega N - \frac{w}{z_S} + iku + \frac{\partial w}{\partial z} = 0 \tag{8}}}$$

## The Linearized Horizontal Momentum Equation

The full momentum equation is given by:

$$\frac{\partial v}{\partial t} + v \cdot \nabla v + \frac{1}{\rho}\nabla p + gk = 0 \tag{9}$$

Isolate the horizontal ($x$-direction) component of this equation. Gravity ($gk$) acts only in the vertical ($z$-direction), so it drops out:

$$\frac{\partial u}{\partial t} + u\frac{\partial u}{\partial x} + v\frac{\partial u}{\partial y} + w\frac{\partial u}{\partial z} + \frac{1}{\rho}\frac{\partial p}{\partial x} = 0 \tag{10}$$

We linearize around the resting background state. Since the velocities $(u, v, w)$ are small perturbations, the non-linear advection terms (like $u\frac{\partial u}{\partial x}$) are dropped:

$$\frac{\partial u}{\partial t} + \frac{1}{\rho_0 + \rho'}\frac{\partial (p_0 + p')}{\partial x} = 0 \tag{11}$$

Because the background pressure $p_0$ only depends on height $z$, $\frac{\partial p_0}{\partial x} = 0$. Approximating $\frac{1}{\rho_0 + \rho'} \approx \frac{1}{\rho_0}$ to keep only first-order perturbation terms, the equation simplifies to: 

$$\frac{\partial u}{\partial t} + \frac{1}{\rho_0}\frac{\partial p'}{\partial x} = 0 \tag{12}$$

Let the fractional pressure perturbation be $P = p'/p_0$, which means $p' = P p_0$. Substituting this into the pressure gradient term:

$$\frac{1}{\rho_0}\frac{\partial (P p_0)}{\partial x} = \frac{p_0}{\rho_0}\frac{\partial P}{\partial x} \tag{13}$$

From the ideal gas law applied to the base state, $p_0 = \rho_0 R T_0$.

Furthermore, the scale height is defined as $z_S = R T_0 / g$, which means $R T_0 = g z_S$. Therefore, the ratio $\frac{p_0}{\rho_0} = R T_0 = g z_S$. Substituting this back gives:

$$\frac{\partial u}{\partial t} + g z_S \frac{\partial P}{\partial x} = 0 \tag{14}$$

Apply the plane wave form proportional to $\exp[i(kx - \omega t)]$ to the variables. The time derivative becomes $-i\omega$ and the $x$-derivative becomes $ik$:

$$-i\omega u + g z_S (ik P) = 0 \tag{15}$$

$${\color{orange}\boxed{-i\omega u + ik g z_S P = 0}} \tag{16}$$

## The Linearized Vertical Momentum Equation

The fu momentum equation is given by: 

$$\frac{\partial v}{\partial t} + v \cdot \nabla v + \frac{1}{\rho}\nabla p + gk = 0 \tag{17}$$

Isolate the vertical ($z$-direction) component of this equation. Gravity ($gk$) acts exclusively in this direction:

$$\frac{\partial w}{\partial t} + u\frac{\partial w}{\partial x} + v\frac{\partial w}{\partial y} + w\frac{\partial w}{\partial z} + \frac{1}{\rho}\frac{\partial p}{\partial z} + g = 0 \tag{18}$$

Linearize around the resting background state by dropping the non-linear advective terms (e.g., $w\frac{\partial w}{\partial z}$), and substituting the base state and perturbation variables $\rho = \rho_0 + \rho'$ and $p = p_0 + p'$:

$$\frac{\partial w}{\partial t} + \frac{1}{\rho_0 + \rho'}\frac{\partial (p_0 + p')}{\partial z} + g = 0 \tag{19}$$

The pressure gradient and gravity terms simplify to $\frac{1}{\rho_0}\frac{\partial p'}{\partial z} + \frac{g\rho'}{\rho_0}$ after invoking hydrostatic balance to eliminate $\nabla p_0$. Substituting this into our vertical equation yields:

$$\frac{\partial w}{\partial t} + \frac{1}{\rho_0}\frac{\partial p'}{\partial z} + \frac{g\rho'}{\rho_0} = 0 \tag{20}$$

Let the fractional pressure perturbation be $P = p'/p_0$, which means $p' = P p_0$. Taking the vertical derivative:

$$\frac{1}{\rho_0}\frac{\partial (p_0 P)}{\partial z} = \frac{p_0}{\rho_0}\frac{\partial P}{\partial z} + \frac{P}{\rho_0}\frac{\partial p_0}{\partial z} \tag{21}$$

From the hydrostatic equation (3.5), $\frac{d \ln p_0}{dz} = -\frac{1}{z_S}$, which means $\frac{\partial p_0}{\partial z} = -\frac{p_0}{z_S}$. Substitute this into the pressure term:

$$\frac{p_0}{\rho_0}\frac{\partial P}{\partial z} - \frac{P}{\rho_0}\frac{p_0}{z_S} = \frac{p_0}{\rho_0}(\frac{\partial P}{\partial z} - \frac{P}{z_S}) \tag{22}$$

From the ideal gas law applied to the base state (3.4), $p_0 = \rho_0 R T_0$. Furthermore, the scale height is $z_S = R T_0 / g$, which means $\frac{p_0}{\rho_0} = R T_0 = g z_S$. Substituting this back gives:

$$g z_S (\frac{\partial P}{\partial z} - \frac{P}{z_S}) = g z_S \frac{\partial P}{\partial z} - g P \tag{23}$$

Substitute this simplified pressure term back into the momentum equation (20), and use the fractional density perturbation $N = \rho'/\rho_0$:

$$\frac{\partial w}{\partial t} + g z_S \frac{\partial P}{\partial z} - g P + g N = 0 \tag{24}$$

Apply the plane wave form proportional to $\exp[i(kx - \omega t)]$ to the variables. The time derivative on $w$ becomes $-i\omega$:

$$-i\omega w + g z_S \frac{\partial P}{\partial z} + g(N - P) = 0 \tag{25}$$


$${\color{orange}\boxed{-i\omega w + g z_S \frac{\partial P}{\partial z} + g(N - P) = 0}} \tag{26}$$

## The Linearized Energy Equation

The full thermodynamic energy equation is given by:

$$\frac{\partial\theta}{\partial t} + v \cdot \nabla\theta = 0 \tag{27}$$

Linearize the equation by introducing $\theta = \theta_0 + \theta'$, dropping non-linear advection, and retaining only the vertical motion acting on the background potential temperature gradient:

$$\frac{\partial \theta'}{\partial t} + w\frac{\partial \theta_0}{\partial z} = 0 \tag{28}$$

Divide the entire equation by the background potential temperature $\theta_0$

$$\frac{1}{\theta_0}\frac{\partial \theta'}{\partial t} + \frac{w}{\theta_0}\frac{\partial \theta_0}{\partial z} = 0 \tag{29}$$

From equation (3.8), the base state potential temperature profile is $\theta_0 = T_0 \exp(\kappa z/z_S)$. Taking the derivative with respect to $z$ yields:

$$\frac{\partial \theta_0}{\partial z} = \theta_0 \frac{\kappa}{z_S} \implies \frac{1}{\theta_0}\frac{\partial \theta_0}{\partial z} = \frac{\kappa}{z_S} \tag{30}$$


Substitute this profile derivative back into our thermodynamic equation:

$$\frac{1}{\theta_0}\frac{\partial \theta'}{\partial t} + \frac{\kappa w}{z_S} = 0 \tag{31}$$

We consider: $\frac{\theta'}{\theta_0} = \frac{P}{\gamma} - N$.  Because $\theta_0$ only depends on $z$ and not $t$, $\frac{1}{\theta_0}\frac{\partial \theta'}{\partial t} = \frac{\partial}{\partial t}(\frac{\theta'}{\theta_0})$. Apply the plane wave form $\exp[i(kx - \omega t)]$, where the time derivative becomes $-i\omega$:

$$-i\omega (\frac{\theta'}{\theta_0}) + \frac{\kappa w}{z_S} = 0 \tag{32}$$

Substitute the $(P, N)$ relation into this equation:

$$-i\omega [P/\gamma - N] + \frac{\kappa w}{z_S} = 0 \tag{33}$$

$${\color{orange}\boxed{-i\omega [P/\gamma - N] + \frac{\kappa w}{z_S} = 0}} \tag{34}$$

## Sound, Inetia-Graviy Waves, and Lamb Waves

We start with the four linearized governing algebraic and differential equations:


$$-i\omega N - \frac{w}{z_S} + iku + \frac{\partial w}{\partial z} = 0 \tag{8}$$
$$-i\omega u + ikgz_S P = 0 \tag{16}$$
$$-i\omega w + gz_S\frac{\partial P}{\partial z} + g(N - P) = 0 \tag{26}$$$$-i\omega[P/\gamma - N] + \frac{\kappa w}{z_S} = 0 \tag{34}$$

From equation (16), isolate the horizontal velocity $u$:$$u = \frac{kgz_S}{\omega}P \tag{35}$$

$$u = \frac{kgz_S}{\omega}P \tag{35}$$

From equation (34), isolate the term $-i\omega N$:

$$-i\omega N = -i\omega \frac{P}{\gamma} + \frac{\kappa w}{z_S} \tag{36}$$

For use in equation (26), we also need an expression for $N$. Dividing equation (34) by $-i\omega$ gives:

$$N = \frac{P}{\gamma} - \frac{\kappa w}{i\omega z_S} = \frac{P}{\gamma} + \frac{i\kappa w}{\omega z_S} \tag{37}$$

We substitute the isolated expression for $u$ (35) and the expression for $-i\omega N$ (36) into equation (8):

$$(-i\omega \frac{P}{\gamma} + \frac{\kappa w}{z_S}) - \frac{w}{z_S} + ik(\frac{kgz_S}{\omega}P) + \frac{\partial w}{\partial z} = 0 \tag{38}$$

Group the terms containing the vertical velocity $w$ and its spatial derivative:

$$\frac{\partial w}{\partial z} + (\frac{\kappa}{z_S} - \frac{1}{z_S})w = (\frac{\partial}{\partial z} + \frac{\kappa - 1}{z_S})w \tag{39}$$

Recalling the thermodynamic relationship $1 - \kappa = 1/\gamma$, which implies $\kappa - 1 = -1/\gamma$. Substituting this into the $w$ term yields:

$$(\frac{\partial}{\partial z} - \frac{1}{\gamma z_S})w \tag{40}$$

Next, we group the terms containing the pressure perturbation $P$:

$$-i\omega \frac{P}{\gamma} + \frac{ik^2gz_S}{\omega}P = i\omega(\frac{k^2gz_S}{\omega^2} - \frac{1}{\gamma})P \tag{41}$$

Combining the grouped $w$ and $P$ terms yields:

$${\color{orange}\boxed{(\frac{\partial}{\partial z} - \frac{1}{\gamma z_S})w + i\omega(\frac{k^2 g z_S}{\omega^2} - \frac{1}{\gamma})P = 0}} \tag{42}$$

Now, we substitute the expression derived for $N$ (37) into the $g(N - P)$ term of equation (26):

$$-i\omega w + gz_S\frac{\partial P}{\partial z} + g[(\frac{P}{\gamma} + \frac{i\kappa w}{\omega z_S}) - P] = 0 \tag{43}$$

Group the $P$ terms inside the bracket and use $1/\gamma - 1 = -\kappa$:

$$N - P = P(\frac{1}{\gamma} - 1) + \frac{i\kappa w}{\omega z_S} = -\kappa P + \frac{i\kappa w}{\omega z_S} \tag{44}$$

Now place this back into equation (43) and divide the entire equation by $gz_S$:

$$-\frac{i\omega}{gz_S}w + \frac{\partial P}{\partial z} - \frac{\kappa}{z_S}P + \frac{i\kappa}{\omega z_S^2}w = 0 \tag{45}$$

Group the terms containing $P$ and its spatial derivative, and factor out $-\frac{i\omega}{gz_S}$ from the $w$ terms:

$$(\frac{\partial}{\partial z} - \frac{\kappa}{z_S})P - \frac{i\omega}{gz_S}(1 - \frac{\kappa g}{\omega^2 z_S})w = 0 \tag{46}$$

$${\color{orange}\boxed{(\frac{\partial}{\partial z} - \frac{\kappa}{z_S})P - \frac{i\omega}{gz_S}(1 - \frac{\kappa g}{\omega^2 z_S})w = 0}} \tag{47}$$

Rearrange the previously derived equation (47) to isolate $w$:

$$w = \frac{g z_S}{i\omega(1 - \frac{\kappa g}{\omega^2 z_S})} (\frac{\partial}{\partial z} - \frac{\kappa}{z_S})P \tag{48}$$

Substitute this expression for $w$ into the first term of equation (42). Because the background state is isothermal, the parameters ($z_S$, $g$, $\gamma$, and $\kappa$) are constant with respect to $z$, allowing the constant coefficient to be pulled out of the spatial derivative:

$$\frac{g z_S}{i\omega(1 - \frac{\kappa g}{\omega^2 z_S})} (\frac{\partial}{\partial z} - \frac{1}{\gamma z_S})(\frac{\partial}{\partial z} - \frac{\kappa}{z_S})P + i\omega(\frac{k^2 g z_S}{\omega^2} - \frac{1}{\gamma})P = 0 \tag{49}$$

Expand the product of the linear differential operators acting on $P$, applying the identity $\kappa + 1/\gamma = 1$:

$$(\frac{\partial}{\partial z} - \frac{1}{\gamma z_S})(\frac{\partial}{\partial z} - \frac{\kappa}{z_S})P = [\frac{\partial^2}{\partial z^2} - \frac{1}{z_S}\frac{\partial}{\partial z} + \frac{\kappa}{\gamma z_S^2}]P \tag{50}$$

Multiply the entire substituted equation (49) by the reciprocal of the leading constant, $\frac{i\omega(1 - \frac{\kappa g}{\omega^2 z_S})}{g z_S}$, and note that $(i\omega)^2 = -\omega^2$:

$$[\frac{\partial^2}{\partial z^2} - \frac{1}{z_S}\frac{\partial}{\partial z} + \frac{\kappa}{\gamma z_S^2}]P - (\frac{\omega^2}{g z_S} - \frac{\kappa}{z_S^2})(\frac{k^2 g z_S}{\omega^2} - \frac{1}{\gamma})P = 0 \tag{51}$$

Expand the algebraic product term by term:

$$- [k^2 - \frac{\omega^2}{\gamma g z_S} - \frac{\kappa g k^2}{\omega^2 z_S} + \frac{\kappa}{\gamma z_S^2}]P = [\frac{\omega^2}{\gamma g z_S} + (\frac{\kappa g}{\omega^2 z_S} - 1)k^2 - \frac{\kappa}{\gamma z_S^2}]P \tag{52}$$

Add this expanded algebraic portion back to the differential portion (50). The constant $\frac{\kappa}{\gamma z_S^2}$ terms cancel out, yielding:

$${\color{orange}\boxed{[\frac{\partial^2}{\partial z^2} - \frac{1}{z_S}\frac{\partial}{\partial z} + (\frac{\kappa g}{z_S\omega^2} - 1)k^2 + \frac{\omega^2}{\gamma g z_S}]P = 0}} \tag{53}$$

### The Dispersion Relation for Sound and Gravity Waves

Assume a solution of the form $P \propto \exp[(im + \mu)z]$. Setting the imaginary part of the coefficient of $P$ to zero yields $\mu = 1/(2z_S)$, giving:

$$P \propto \exp[imz + \frac{z}{2z_S}] \tag{54}$$

Find the first and second derivatives of $P$ with respect to $z$:

$$\frac{\partial P}{\partial z} = (im + \frac{1}{2z_S})P \tag{55}$$

$$\frac{\partial^2 P}{\partial z^2} = (im + \frac{1}{2z_S})^2 P = (-m^2 + \frac{im}{z_S} + \frac{1}{4z_S^2})P \tag{56}$$

We substitute the derivatives into $[\frac{\partial^2}{\partial z^2} - \frac{1}{z_S}\frac{\partial}{\partial z}]P$:

$$(-m^2 + \frac{im}{z_S} + \frac{1}{4z_S^2})P - \frac{1}{z_S}(im + \frac{1}{2z_S})P = -(m^2 + \frac{1}{4z_S^2})P \tag{57}$$

Substitute this simplified result back into the full second-order differential equation for $P$ (from 53), divide by $P$, and group the terms using the definition $l^2 = k^2 + m^2 + 1/(4z_S^2)$:

$$-l^2 + \frac{\kappa g k^2}{z_S \omega^2} + \frac{\omega^2}{\gamma g z_S} = 0 \tag{58}$$

Multiply the entire equation by $\gamma g z_S \omega^2$ to clear the denominators, and rearrange to form a quadratic in terms of $\omega^2$:

$$\omega^4 - \gamma g z_S l^2 \omega^2 + \gamma \kappa g^2 k^2 = 0 \tag{59}$$

Apply the standard quadratic formula, factor out $(\gamma g z_S l^2)^2$ from the square root, and simplify to arrive at the final dispersion relation:

$${\color{orange}\boxed{\omega^2 = \frac{\gamma g z_S l^2}{2} \left[ 1 \pm \left( 1 - \frac{4 \kappa k^2}{\gamma z_S^2 l^4} \right)^{1/2} \right]}} \tag{60}$$

### The Dispersion Relation for Lamb Waves

Set $w = 0$ in equation (42). The first term vanishes entirely, leaving:

$$i\omega(\frac{k^2 g z_S}{\omega^2} - \frac{1}{\gamma})P = 0 \tag{61}$$

Assuming $P \neq 0$ and $\omega \neq 0$, divide the equation by $i\omega P$ and isolate $\omega^2$:

$$\frac{k^2 g z_S}{\omega^2} - \frac{1}{\gamma} = 0 \implies \omega^2 = \gamma g z_S k^2 \tag{62}$$

From the hydrostatic balance and ideal gas law, the scale height is $z_S = R T_0 / g$. Substituting $g z_S = R T_0$ yields:

$$\omega^2 = \gamma R T_0 k^2 \tag{63}$$

Recognizing that the phase speed of pure sound waves is defined as $c^2 = \gamma R T_0$, this gives the final relation:

$${\color{orange}\boxed{\omega^2 = c^2 k^2}} \tag{64}$$

#### The Vertical Structure of Lamb Waves

Set $w = 0$ in equation (47). The entire second term is multiplied by zero and vanishes:

$$(\frac{\partial}{\partial z} - \frac{\kappa}{z_S})P - \frac{i\omega}{g z_S}(1 - \frac{\kappa g}{\omega^2 z_S})(0) = 0 \tag{65}$$

The elimination of the second term leaves only the first term, immediately establishing the vertical structure:

$${\color{orange}\boxed{(\frac{\partial}{\partial z} - \frac{\kappa}{z_S})P = 0}} \tag{66}$$

## Boussinesq inertia-gravity waves

We start with the five algebraic equations obtained by applying the plane wave assumption $exp[i(kx+mz-\omega t)]$ to the linearized Boussinesq equations:

$$iku + imw = 0 \tag{67}$$$$-i\omega u + ik\pi' - fv = 0 \tag{68}$$
$$-i\omega v + fu = 0 \tag{69}$$
$$-i\omega w + im\pi' - b' = 0 \tag{70}$$
$$-i\omega b' + N^2 w = 0 \tag{71}$$

From equation (69), isolate the transverse velocity $v$:

$$v = \frac{f}{i\omega}u = -\frac{if}{\omega}u \tag{72}$$

Substitute this expression for $v$ into equation (68):

$$-i\omega u + ik\pi' - f(-\frac{if}{\omega}u) = 0 \tag{73}$$

Multiply the terms out and factor $u$, knowing that $i^2 = -1$:

$$-i\omega u - \frac{if^2}{\omega}u + ik\pi' = 0 \implies -i\left(\frac{\omega^2 - f^2}{\omega}\right)u + ik\pi' = 0 \tag{74}$$

Solving for the pressure perturbation $\pi'$:

$$\pi' = \frac{\omega^2 - f^2}{\omega k}u \tag{75}$$

From equation (67), isolate the horizontal velocity $u$:

$$u = -\frac{m}{k}w \tag{76}$$

We substitute equation (76) into equation (75) to express $\pi'$ entirely in terms of $w$:

$$\pi' = -\frac{m(\omega^2 - f^2)}{\omega k^2}w \tag{77}$$

Next, from equation (71), isolate the buoyancy perturbation $b'$:

$$b' = \frac{N^2}{i\omega}w = -\frac{iN^2}{\omega}w \tag{78}$$

Now substitute both derived expressions for $\pi'$ (77) and $b'$ (78) into the vertical momentum equation (70):

$$-i\omega w + im\left(-\frac{m(\omega^2 - f^2)}{\omega k^2}w\right) - \left(-\frac{iN^2}{\omega}w\right) = 0 \tag{79}$$

Assuming $w \neq 0$, divide the entire equation by $-iw$:

$$\omega + \frac{m^2(\omega^2 - f^2)}{\omega k^2} - \frac{N^2}{\omega} = 0 \tag{80}$$

Multiply the entire equation by $\omega k^2$ to clear the denominators:

$$\omega^2 k^2 + m^2(\omega^2 - f^2) - N^2k^2 = 0 \tag{81}$$

Expand and group the terms multiplied by $\omega^2$:

$$\omega^2(k^2 + m^2) - (N^2k^2 + f^2m^2) = 0 \tag{82}$$

Multiply by $\omega$, yields:

$$[\omega^2(k^2 + m^2) - (N^2k^2 + f^2m^2)]\omega = 0 \tag{83}$$

For non-trivial wave solutions, assume $\omega \neq 0$, isolate $\omega^2$, and take the square root to find the dispersion relation:

$${\color{green}\boxed{\omega = \pm \left(\frac{N^2k^2 + f^2m^2}{k^2+m^2}\right)^{1/2}}} \tag{84}$$

### The "Trivial" Mode 

For the zero-frequency solution of the secular equation, set $\omega = 0$.

Starting with equation (69), setting $\omega = 0$ yields:

$$fu = 0 \implies u = 0 \tag{85}$$

We replace $u=0$ into equation (67):

$$imw = 0 \implies w = 0 \tag{86}$$

Substitute $w=0$ into equation (70) and isolate $b'$ to establish hydrostatic balance (3.38):

$$im\pi' - b' = 0 \implies b' = im\pi' \tag{87}$$

$${\color{green}\boxed{b' = im\pi'}} \tag{88}$$

Substitute $\omega=0$ into equation (68) and rearrange to find the geostrophic balance of the $y$ component (3.39):

$$ik\pi' - fv = 0 \implies fv = ik\pi' \tag{89}$$

$${\color{green}\boxed{v = \frac{ik}{f}\pi'}} \tag{90}$$

### Inertia-Gravity Wave

For the case where $\omega^2 > 0$, we rely on equations derived during the dispersion relation steps to formulate the polarization relations in terms of $w$.From equation (67), rearrange to find $u$:

$${\color{green}\boxed{u = -\frac{m}{k}w}} \tag{91}$$

Recall equation (72) and substitute the established relation for $u$ (3.41):

$$v = -\frac{if}{\omega}u = -\frac{if}{\omega}\left(-\frac{m}{k}w\right) \implies {\color{green}\boxed{v = -\frac{if}{\omega}u = \frac{ifm}{\omega k}w}} \tag{92}$$

Recall equation (78) directly to establish $b'$:

$${\color{green}\boxed{b' = -\frac{iN^2}{\omega}w}} \tag{93}$$

Recall equation (77) directly to establish $\pi'$:

$${\color{green}\boxed{\pi' = -\frac{m(\omega^2 - f^2)}{k^2\omega}w}} \tag{94}$$

### Phase and Trace Speeds

The phase speed of a wave in three-dimensional space is the frequency divided by the total wavenumber magnitude:
$$v_p = \frac{\omega}{(k^2+m^2)^{1/2}} \tag{95}$$

We substitute the positive root of the dispersion relation (84) into the numerator to find the Boussinesq phase speed:

$${\color{green}\boxed{v_p = \frac{(k^2N^2+m^2f^2)^{1/2}}{k^2+m^2}}} \tag{96}$$

The trace speeds are simply the frequency divided by the respective Cartesian components of the wave vector:

$${\color{green}\boxed{u_t = \frac{\omega}{k}}} \tag{97}$$

$${\color{green}\boxed{w_t = \frac{\omega}{m}}} \tag{98}$$


To establish the relationship between them, calculate the sum of the inverse squares of the trace speeds:
$$\frac{1}{u_t^2} + \frac{1}{w_t^2} = \frac{k^2}{\omega^2} + \frac{m^2}{\omega^2} = \frac{k^2+m^2}{\omega^2} \tag{99}$$

Notice that this result is the inverse square of the phase speed definition from (95), establishing equation:

$${\color{green}\boxed{\frac{1}{v_p^2} = \frac{1}{u_t^2} + \frac{1}{w_t^2}}} \tag{100}$$

### Group Velocity

The group velocity is defined as the gradient of the frequency with respect to the wave vector components:

$${\color{green}\boxed{v_g = \left(\frac{\partial\omega}{\partial k}, \frac{\partial\omega}{\partial l}, \frac{\partial\omega}{\partial m}\right)}} \tag{101}$$

Assuming $l = 0$, start by squaring the dispersion relation (84) to perform implicit differentiation:

$$\omega^2 = \frac{N^2k^2 + f^2m^2}{k^2+m^2} \tag{102}$$

Differentiate implicitly with respect to $k$ using the quotient rule:

$$2\omega \frac{\partial\omega}{\partial k} = \frac{(k^2+m^2)(2kN^2) - (N^2k^2 + f^2m^2)(2k)}{(k^2+m^2)^2} \tag{103}$$

Expand and simplify the numerator:

$$2\omega \frac{\partial\omega}{\partial k} = \frac{2k[N^2k^2 + N^2m^2 - N^2k^2 - f^2m^2]}{(k^2+m^2)^2} = \frac{2km^2(N^2-f^2)}{(k^2+m^2)^2} \tag{104}$$

Divide by $2\omega$:

$$\frac{\partial\omega}{\partial k} = \frac{km^2(N^2-f^2)}{\omega(k^2+m^2)^2} \tag{105}$$

From the trace speed definition (97), we know $\frac{k}{\omega} = \frac{1}{u_t}$. Substitute this to find the horizontal group velocity $u_g$:

$${\color{green}\boxed{u_g = \frac{m^2(N^2-f^2)}{u_t(k^2+m^2)^2}}} \tag{106}$$

Now, differentiate equation (102) implicitly with respect to $m$ using the quotient rule:

$$2\omega \frac{\partial\omega}{\partial m} = \frac{(k^2+m^2)(2mf^2) - (N^2k^2 + f^2m^2)(2m)}{(k^2+m^2)^2} \tag{107}$$

Expand and simplify the numerator:

$$2\omega \frac{\partial\omega}{\partial m} = \frac{2m[f^2k^2 + f^2m^2 - N^2k^2 - f^2m^2]}{(k^2+m^2)^2} = \frac{2m(-k^2(N^2-f^2))}{(k^2+m^2)^2} \tag{108}$$

Divide by $2\omega$:

$$\frac{\partial\omega}{\partial m} = -\frac{mk^2(N^2-f^2)}{\omega(k^2+m^2)^2} \tag{109}$$

From the trace speed definition (98), we know $\frac{m}{\omega} = \frac{1}{w_t}$. 

Finally, we substitute this to find the vertical group velocity $w_g$:

$${\color{green}\boxed{w_g = -\frac{k^2(N^2-f^2)}{w_t(k^2+m^2)^2}}} \tag{110}$$
