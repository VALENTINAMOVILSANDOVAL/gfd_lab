# Geophysical Fluid Dynamics

When we study fluid dynamics, we are interested in understanding how fluids move and evolve over time. But what exactly is Geophysical Fluid Dynamics (GFD)? GFD is the application of fluid dynamics to naturally occurring fluids on Earth and other planets, such as the atmosphere and the oceans. Since these fluids are part of the Earth system, they are called *geophysical fluids*. By studying their motion, we can better understand important phenomena such as weather patterns, ocean currents, and climate. Two key features that make geophysical fluids unique are the rotation of the Earth, which affects fluid motion, and stratification, which refers to the layering of fluids with different properties. In this series of notebooks, we will introduce the main concepts of GFD, combining physical intuition, mathematical derivations, and simple examples to help build a solid understanding of the subject.

## 3. Sound, Inertia-Gravity Waves, and Lamb Waves

We want to examine the fundamental modes of the GFD equations for small perturbations in amplitude and scale, where a linearized treatment is appropriate. We first develop a framework for the modes of an isothermal, non-rotating atmosphere at rest. We then extend the analysis to a rotating atmosphere at rest with constant Brunt–Väisälä frequency under the Boussinesq approximation.

<hr style="border:none; border-top:1px solid #888;">

### Resting, non-rotating atmosphere

We write the full governing equations for a resting, non-rotating atmosphere:
$$\frac{\partial \rho}{\partial t} + v \cdot \nabla \rho + \rho \nabla \cdot v = 0 \tag{3.1}$$
$$\frac{\partial v}{\partial t} + v \cdot \nabla v + \frac{1}{\rho}\nabla p + gk = 0 \tag{3.2}$$
$$\frac{\partial \theta}{\partial t} + v \cdot \nabla \theta = 0 \tag{3.3}$$

We define the isothermal, hydrostatic base state:
$$\frac{d \ln p_0}{dz} = -\frac{1}{z_S} \tag{3.4}$$

We establish the base state profile for density:
$$\rho_0 = \frac{p_0}{g z_S} \tag{3.5}$$

and the base state profile for potential temperature:
$$\theta_0 = T_0 \exp(\kappa z / z_S) \tag{3.6}$$

We introduce small perturbations about a state of rest:
$$p = p_0(z) + p', \quad \rho = \rho_0(z) + \rho', \quad \theta = \theta_0(z) + \theta', \quad v = (u, 0, w) \tag{3.7}$$

We assume a plane wave form for all perturbation variables:
$$\exp[i(kx - \omega t)] \tag{3.8}$$

and define how this plane wave form transforms spatial derivatives in $x$ to multiplication by $ik$ and time derivatives to multiplication by $-i\omega$:
$$\frac{\partial}{\partial x} \rightarrow ik, \quad \frac{\partial}{\partial t} \rightarrow -i\omega \tag{3.9}$$

We define the non-dimensional pressure perturbation variable:
$$P = \frac{p'}{p_0} \tag{3.10}$$

and the non-dimensional density perturbation variable:
$$N = \frac{\rho'}{\rho_0} \tag{3.11}$$

---

#### Linearized Governing Equations

**Continuity Equation**

We expand the continuity equation using the defined perturbations:
$$\frac{\partial \rho'}{\partial t} + w \frac{\partial \rho_0}{\partial z} + \rho_0 \left( \frac{\partial u}{\partial x} + \frac{\partial w}{\partial z} \right) = 0$$

Divide the expression by the base state density $\rho_0$:
$$\frac{1}{\rho_0} \frac{\partial \rho'}{\partial t} + \frac{w}{\rho_0} \frac{\partial \rho_0}{\partial z} + \frac{\partial u}{\partial x} + \frac{\partial w}{\partial z} = 0$$

Substitute the hydrostatic scale height derivative $-\frac{1}{z_S}$ for $\frac{1}{\rho_0} \frac{\partial \rho_0}{\partial z}$:
$$\frac{1}{\rho_0} \frac{\partial \rho'}{\partial t} - \frac{w}{z_S} + \frac{\partial u}{\partial x} + \frac{\partial w}{\partial z} = 0$$

Finally, we apply the derivative transformations for $x$ and $t$, and substitute the definition of $N$:
$$\boxed{ -i\omega N - \frac{w}{z_S} + iku + \frac{\partial w}{\partial z} = 0 } \tag{3.12}$$

**Horizontal Momentum Equation**

We linearize the $x$-component of the momentum equation:
$$\frac{\partial u}{\partial t} + \frac{1}{\rho_0}\frac{\partial p'}{\partial x} = 0$$

We apply the derivative transformations for $x$ and $t$:
$$-i\omega u + \frac{ik p'}{\rho_0} = 0$$

Substitute the expression $\frac{p_0}{g z_S}$ for $\rho_0$:
$$-i\omega u + \frac{ik g z_S p'}{p_0} = 0$$

Finally, we replace the non-dimensional pressure $P$ to obtain the final form:
$$\boxed{ -i\omega u + ikgz_S P = 0 } \tag{3.13}$$

**Vertical Momentum Equation**

We linearize the $z$-component of the momentum equation using the approximated pressure gradient and gravity terms:
$$\frac{\partial w}{\partial t} + \frac{1}{\rho_0}\frac{\partial p'}{\partial z} + g \frac{\rho'}{\rho_0} = 0$$

We apply the derivative transformation for $t$ and substitute $N$:
$$-i\omega w + \frac{1}{\rho_0}\frac{\partial p'}{\partial z} + gN = 0$$

Expand the vertical derivative of $p'$ using the product rule on $p' = p_0 P$:
$$\frac{\partial p'}{\partial z} = P \frac{\partial p_0}{\partial z} + p_0 \frac{\partial P}{\partial z}$$

Substitute the hydrostatic relation $-\frac{p_0}{z_S}$ for the derivative of the base pressure $\frac{\partial p_0}{\partial z}$:
$$\frac{\partial p'}{\partial z} = -\frac{p_0}{z_S} P + p_0 \frac{\partial P}{\partial z}$$

Substitute this expanded derivative and $\rho_0 = \frac{p_0}{g z_S}$ back into the vertical momentum equation:
$$-i\omega w + \frac{g z_S}{p_0} \left( -\frac{p_0}{z_S} P + p_0 \frac{\partial P}{\partial z} \right) + gN = 0$$

We distribute the terms to simplify the expression:
$$-i\omega w - gP + gz_S \frac{\partial P}{\partial z} + gN = 0$$

Finally, we group the common variables to yield the final equation:
$$\boxed{ -i\omega w + gz_S \frac{\partial P}{\partial z} + g(N - P) = 0 } \tag{3.14}$$

**Thermodynamic Energy Equation**

We linearize the thermodynamic equation:
$$\frac{\partial \theta'}{\partial t} + w \frac{\partial \theta_0}{\partial z} = 0$$

Divide the expression by $\theta_0$:
$$\frac{1}{\theta_0}\frac{\partial \theta'}{\partial t} + \frac{w}{\theta_0} \frac{\partial \theta_0}{\partial z} = 0$$

Evaluate the fractional vertical derivative of the base potential temperature:
$$\frac{1}{\theta_0} \frac{\partial \theta_0}{\partial z} = \frac{\kappa}{z_S}$$

Rewrite the fractional potential temperature perturbation using the non-dimensional pressure and density perturbations:
$$\frac{\theta'}{\theta_0} = \frac{P}{\gamma} - N$$

Substitute these expressions back into the divided equation:
$$\frac{\partial}{\partial t} \left( \frac{P}{\gamma} - N \right) + \frac{\kappa w}{z_S} = 0$$

We apply the derivative transformation for $t$ to finalize the equation:
$$\boxed{ -i\omega \left[ \frac{P}{\gamma} - N \right] + \frac{\kappa w}{z_S} = 0 } \tag{3.15}$$

---

#### The Dispersion Relation

We state the system of two first-order differential equations for $w$ and $P$ obtained before eliminating variables:
$$\left(\frac{\partial}{\partial z} - \frac{1}{\gamma z_S}\right)w + i\omega\left(\frac{k^2 g z_S}{\omega^2} - \frac{1}{\gamma}\right)P = 0 \tag{3.16}$$
$$\left(\frac{\partial}{\partial z} - \frac{\kappa}{z_S}\right)P - \frac{i\omega}{g z_S}\left(1 - \frac{\kappa g}{\omega^2 z_S}\right)w = 0 \tag{3.17}$$

Combining these by eliminating $w$ to yield the single second-order differential equation for $P$:
$$\left[ \frac{\partial^2}{\partial z^2} - \frac{1}{z_S} \frac{\partial}{\partial z} + \left(\frac{\kappa g}{z_S \omega^2} - 1\right)k^2 + \frac{\omega^2}{\gamma g z_S} \right] P = 0 \tag{3.18}$$

We assume a vertically growing exponential solution for $P$:
$$P \propto \exp\left[ i m z + \frac{z}{2z_S} \right] \tag{3.19}$$

We compute the first vertical derivative operator for this assumed solution:
$$\frac{\partial}{\partial z} \rightarrow \left(im + \frac{1}{2z_S}\right)$$

We compute the second vertical derivative operator by squaring the first:
$$\frac{\partial^2}{\partial z^2} \rightarrow \left(im + \frac{1}{2z_S}\right)^2 = -m^2 + \frac{im}{z_S} + \frac{1}{4z_S^2}$$

Substitute these differential operators into the second-order equation:
$$\left[ -m^2 + \frac{im}{z_S} + \frac{1}{4z_S^2} - \frac{1}{z_S}\left(im + \frac{1}{2z_S}\right) + \left(\frac{\kappa g}{z_S \omega^2} - 1\right)k^2 + \frac{\omega^2}{\gamma g z_S} \right] P = 0$$

Distribute the $-\frac{1}{z_S}$ term:
$$\left[ -m^2 + \frac{im}{z_S} + \frac{1}{4z_S^2} - \frac{im}{z_S} - \frac{1}{2z_S^2} + \left(\frac{\kappa g}{z_S \omega^2} - 1\right)k^2 + \frac{\omega^2}{\gamma g z_S} \right] P = 0$$

We cancel the imaginary terms and combine the constant fractions:
$$\left[ -m^2 - \frac{1}{4z_S^2} + \left(\frac{\kappa g}{z_S \omega^2} - 1\right)k^2 + \frac{\omega^2}{\gamma g z_S} \right] P = 0$$

Distribute the $k^2$ term:
$$\left[ -m^2 - \frac{1}{4z_S^2} + \frac{\kappa g k^2}{z_S \omega^2} - k^2 + \frac{\omega^2}{\gamma g z_S} \right] P = 0$$

Grouping the wavenumber and scale height terms together:
$$\left[ -\left(k^2 + m^2 + \frac{1}{4z_S^2}\right) + \frac{\kappa g k^2}{z_S \omega^2} + \frac{\omega^2}{\gamma g z_S} \right] P = 0$$

Substitute the definition of the modified total wavenumber squared $l^2 = k^2 + m^2 + \frac{1}{4z_S^2}$:
$$-l^2 + \frac{\kappa g k^2}{z_S \omega^2} + \frac{\omega^2}{\gamma g z_S} = 0$$

Multiply the entire equation by $\gamma g z_S \omega^2$ to clear the denominators:
$$-\gamma g z_S l^2 \omega^2 + \gamma \kappa g^2 k^2 + \omega^4 = 0$$

We rearrange the terms into descending powers of $\omega$ to obtain the standard quadratic form:
$$\boxed{ \omega^4 - \gamma g z_S l^2 \omega^2 + \gamma \kappa g^2 k^2 = 0 } \tag{3.20}$$

We apply the quadratic formula to solve for $\omega^2$:
$$\omega^2 = \frac{\gamma g z_S l^2 \pm \sqrt{(\gamma g z_S l^2)^2 - 4(1)(\gamma \kappa g^2 k^2)}}{2}$$

Factor out the term $(\gamma g z_S l^2)^2$ from inside the square root:
$$\omega^2 = \frac{\gamma g z_S l^2 \pm \sqrt{(\gamma g z_S l^2)^2 \left[ 1 - \frac{4 \gamma \kappa g^2 k^2}{(\gamma g z_S l^2)^2} \right]}}{2}$$

Extracting the squared term from the radical:
$$\omega^2 = \frac{\gamma g z_S l^2 \pm \gamma g z_S l^2 \sqrt{1 - \frac{4 \gamma \kappa g^2 k^2}{\gamma^2 g^2 z_S^2 l^4}}}{2}$$

We factor out $\frac{\gamma g z_S l^2}{2}$ from the numerator:
$$\omega^2 = \frac{\gamma g z_S l^2}{2} \left[ 1 \pm \sqrt{1 - \frac{4 \gamma \kappa g^2 k^2}{\gamma^2 g^2 z_S^2 l^4}} \right]$$

We simplify the fraction inside the square root by canceling common terms to reach the final solution:
$$\boxed{ \omega^2 = \frac{\gamma g z_S l^2}{2} \left[ 1 \pm \left(1 - \frac{4 \kappa k^2}{\gamma z_S^2 l^4}\right)^{1/2} \right] } \tag{3.21}$$

The branch with the positive root represents sound waves, or at low frequencies, infrasound.
The negative root represents gravity waves, a type of internal wave.

#### Lamb Waves

We note that eliminating $w$ to obtain equation (3.18) inadvertently eliminates a solution where the vertical velocity is identically zero:
$$w = 0 \tag{3.22}$$

Substitute $w = 0$ into the previously established first-order equation (3.16) to yield the dispersion relation for these modes, called Lamb waves:
$$i\omega\left(\frac{k^2 g z_S}{\omega^2} - \frac{1}{\gamma}\right)P = 0 \tag{3.23}$$

We divide the expression by $i\omega P$ assuming non-zero frequency and pressure perturbation:
$$\frac{k^2 g z_S}{\omega^2} - \frac{1}{\gamma} = 0$$

Rearranging the terms to solve for $\omega^2$:
$$\omega^2 = \gamma g z_S k^2$$

We use the ideal gas relations $g z_S = R T_0$ and $\gamma R T_0 = c^2$ to obtain the final dispersion relation for horizontally propagating sound waves:
$$\boxed{ \omega^2 = \gamma g z_S k^2 = \gamma R T_0 k^2 = c^2 k^2 } \tag{3.24}$$

Finally, we replace $w = 0$ into the previously established first-order equation (3.17) to yield an equation for the vertical structure of Lamb waves:
$$\boxed{ \left(\frac{\partial}{\partial z} - \frac{\kappa}{z_S}\right)P = 0 } \tag{3.25}$$

### Resting, Rotating Atmosphere

#### Boussinesq Inertia-Gravity Waves

We extend the analysis of gravity waves to a rotating atmosphere, applying the Boussinesq approximation to keep the equations tractable. The linearized Boussinesq equations about a state of rest are given by:
$$\frac{\partial u}{\partial x} + \frac{\partial v}{\partial y} + \frac{\partial w}{\partial z} = 0 \tag{3.26}$$
$$\frac{\partial u}{\partial t} + \frac{\partial \pi'}{\partial x} - fv = 0 \tag{3.27}$$
$$\frac{\partial v}{\partial t} + \frac{\partial \pi'}{\partial y} + fu = 0 \tag{3.28}$$
$$\frac{\partial w}{\partial t} + \frac{\partial \pi'}{\partial z} - b' = 0 \tag{3.29}$$
$$\frac{\partial b'}{\partial t} + N^2 w = 0 \tag{3.30}$$

Assuming a plane wave moving in the $x-z$ plane, the perturbation variables take the form:
$$\exp[i(kx+mz-\omega t)] \tag{3.30a}$$

This assumption allows us to replace spatial and temporal derivatives with algebraic multipliers:
$$\frac{\partial}{\partial x} \rightarrow ik, \quad \frac{\partial}{\partial y} \rightarrow 0, \quad \frac{\partial}{\partial z} \rightarrow im, \quad \frac{\partial}{\partial t} \rightarrow -i\omega \tag{3.30b}$$

Applying these transformations reduces the partial differential equations (3.26)-(3.30) to a system of algebraic equations:
$$iku + imw = 0 \tag{3.31}$$
$$-i\omega u + ik\pi' - fv = 0 \tag{3.32}$$
$$-i\omega v + fu = 0 \tag{3.33}$$
$$-i\omega w + im\pi' - b' = 0 \tag{3.34}$$
$$-i\omega b' + N^2 w = 0 \tag{3.35}$$

This forms a set of five linear, homogeneous equations. A consistent solution requires the determinant of the coefficients of the dependent variables to be exactly zero, which directly yields the cubic secular equation for $\omega$:
$$[\omega^2(k^2 + m^2) - (N^2 k^2 + f^2 m^2)]\omega = 0 \tag{3.36}$$

Assuming $\omega \neq 0$, we isolate the frequency to obtain the final dispersion relation for inertia-gravity waves:
$$\boxed{ \omega = \pm \left( \frac{N^2 k^2 + f^2 m^2}{k^2 + m^2} \right)^{1/2} } \tag{3.37}$$

---

#### "Trivial" Mode

Returning to the secular equation (3.36), there exists a steady-state solution where the frequency is zero:
$$\omega = 0 \tag{3.37a}$$

Applying this to the system of algebraic equations (3.31)-(3.35), we find that the corresponding velocity components in the $x-z$ plane must vanish:
$$u = w = 0 \tag{3.37b}$$

Substituting these zeroed velocities into equations (3.34) and (3.32) simplifies the system to yield the hydrostatic and geostrophic balance relations, respectively:
$$\boxed{ b' = im\pi' } \tag{3.38}$$
$$\boxed{ v = \frac{ik}{f}\pi' } \tag{3.39}$$

---

#### Inertia-Gravity Mode Polarization Relations

For active waves where $\omega^2 > 0$, we can establish polarization relations linking the variables to the vertical velocity $w$.

Directly from the continuity relation (3.31), we isolate the horizontal velocity $u$:
$$\boxed{ u = -\frac{m}{k}w } \tag{3.40}$$

From the $y$-momentum equation (3.33), we isolate $v$ and substitute equation (3.40) to link it to $w$:
$$\boxed{ v = \frac{ifm}{\omega k}w } \tag{3.41}$$

From the thermodynamic equation (3.35), isolating $b'$ yields:
$$\boxed{ b' = -\frac{iN^2}{\omega}w } \tag{3.42}$$

Finally, manipulating the $x$-momentum equation (3.32) to isolate $\pi'$, and substituting the established relations for $u$ (3.40) and $v$ (3.41), provides the comprehensive pressure perturbation relation:
$$\boxed{ \pi' = -\frac{m(\omega^2 - f^2)}{k^2\omega}w } \tag{3.43}$$

---

#### Speeds and Velocities

The phase speed of these waves in three-dimensional space is derived directly from the dispersion relation (3.37):
$$v_p = \frac{\omega}{(k^2 + m^2)^{1/2}} = \frac{(k^2 N^2 + m^2 f^2)^{1/2}}{k^2 + m^2} \tag{3.44}$$

The phase lines move along the Cartesian axes at their respective trace speeds:
$$u_t = \frac{\omega}{k}, \quad w_t = \frac{\omega}{m} \tag{3.45}$$

The energy propagation of the wave packet is described by the group velocity components, defined in Cartesian coordinates as:
$$v_g = \left(\frac{\partial\omega}{\partial k}, \frac{\partial\omega}{\partial l}, \frac{\partial\omega}{\partial m}\right) \tag{3.47}$$

Differentiating the dispersion relation (3.37) with respect to $k$ and $m$, and substituting the trace speeds from (3.45) to simplify the notation, we obtain the Cartesian components of the group velocity for Boussinesq inertia-gravity waves:
$$\boxed{ u_g = \frac{m^2(N^2 - f^2)}{u_t(k^2 + m^2)^2} } \tag{3.48a}$$
$$\boxed{ w_g = -\frac{k^2(N^2 - f^2)}{w_t(k^2 + m^2)^2} } \tag{3.48b}$$

**4. Standing gravity waves over sinusoidal terrain: Find the steady, non-rotating, flow over terrain with terrain elevation of the form $h = h_0 \sin(kx)$. Do this by carrying out the following steps:**

Setting $w(x,0) = W \cos(kx) = U h_0 k \cos(kx)$, we find the amplitude $W = U h_0 k$.  Substituting $W$ back into the variable equations yields the final solutions for the flow over sinusoidal terrain:
  $$w(x,z) = U h_0 k \cos(kx+mz)$$$$u'(x,z) = -U h_0 m \cos(kx+mz)$$$$\pi'(x,z) = U^2 h_0 m \cos(kx+mz)$$$$b'(x,z) = -N^2 h_0 \sin(kx+mz)$$
