# Geophysical Fluid Dynamics

When we study fluid dynamics, we are interested in understanding how fluids move and evolve over time. But what exactly is Geophysical Fluid Dynamics (GFD)? GFD is the application of fluid dynamics to naturally occurring fluids on Earth and other planets, such as the atmosphere and the oceans. Since these fluids are part of the Earth system, they are called *geophysical fluids*. By studying their motion, we can better understand important phenomena such as weather patterns, ocean currents, and climate. Two key features that make geophysical fluids unique are the rotation of the Earth, which affects fluid motion, and stratification, which refers to the layering of fluids with different properties. In this series of notebooks, we will introduce the main concepts of GFD, combining physical intuition, mathematical derivations, and simple examples to help build a solid understanding of the subject.

## 2. Tools and Tricks of Fluid Dynamics

## The Reynolds Number

The Reynolds number quantifies the ratio of inertial (acceleration) forces to viscous (frictional) forces in the momentum equation (see governing equations, Eq. 25).

To derive its mathematical form, we start by estimating the magnitude of the acceleration term. Assuming a velocity scale $V$, length scale $L$, and time scale $T$, the acceleration $V/T$ can be rewritten by substituting $T=L/V$. This yields an acceleration magnitude of $V^{2}/L$. Next, we estimate the magnitude of the frictional term, which is proportional to $\nu \nabla^{2}v$. Assuming $\nabla^{2}\approx1/L^{2}$, the frictional term scales as $\nu V/L^{2}$.

Taking the ratio of the acceleration term to the friction term yields:

$$Re=\frac{V^{2}/L}{\nu V/L^{2}}\tag{1}$$

Simplifying the fraction gives the final formula for the Reynolds number:

$${\color{blue}\boxed{Re=\frac{VL}{\nu}\tag{2}}}$$

## The Energy Equations

We start with the advective energy equation:

$$\frac{d}{dt}(v^{2}/2+e+\Phi)+\frac{1}{\rho}\nabla\cdot[pv-\mu\nabla(v^{2}/2)]=T\frac{ds_{X}}{dt}\tag{3}$$

We multiply the equation by density $\rho$. To move $\rho$ inside the derivative, we use the mass continuity equation to convert from advective to flux form (combining the Eq. 7 and 13 in governing equations): $\rho \frac{d\chi}{dt} = \frac{\partial (\rho\chi)}{\partial t} + \nabla \cdot (\rho\chi v)$. Applying this to the energy terms gives:

$$\frac{\partial}{\partial t}[\rho(v^{2}/2+e+\Phi)]+\nabla\cdot[\rho(v^{2}/2+e+\Phi)v]+\nabla\cdot[pv-\mu\nabla(v^{2}/2)]=\rho T\frac{ds_{X}}{dt}\tag{4}$$

Now, we combine the divergence terms: $\nabla \cdot [\rho(v^2/2 + e + \Phi)v + pv]$. Then, we factor out $\rho v$ from the first two terms inside the bracket to get $\rho(v^2/2 + e + p/\rho + \Phi)v$. Here, we apply the definition of specific enthalpy, $h=e+p/\rho$, yielding the flux form:

$${\color{orange}\boxed{\frac{\partial}{\partial t}[\rho(v^{2}/2+e+\Phi)]+\nabla\cdot[\rho(v^{2}/2+h+\Phi)v-\mu\nabla(v^{2}/2)]=\rho T\frac{ds_{X}}{dt}\tag{5}}}$$

To find the energy budget in a control volume, we integrate (5) over a volume $V$. We do this by applying the divergence theorem to the flux term, converting the volume integral of the divergence into a surface integral ($\oint$) over the volume boundary:
 
$${\color{orange}\boxed{\frac{d}{dt}\int\rho(v^{2}/2+e+\Phi)dV+\oint[\rho(v^{2}/2+h+\Phi)v-\mu\nabla(v^{2}/2)]\cdot ndV=\int\rho T\frac{ds_{X}}{dt}dV\tag{6}}}$$

## Entropy and Bernoulli Equations

The material derivative of specific entropy $s$ is $ds/dt$. If there are no heat sources or heat conduction, entropy is conserved for a parcel, meaning $ds/dt = 0$. Expanding the material derivative yields:

$${\color{purple}\boxed{\frac{\partial s}{\partial t}+v\cdot\nabla s=0 \tag{7}}}$$

egin with the advective momentum equation (see governing equations Eq. 25) dotted with $v$, expanded using the enthalpy definition $dh=Tds+dp/\rho$, resulting in:

$$\frac{\partial v^{2}/2}{\partial t}+v\cdot\nabla(v^{2}/2)+v\cdot\nabla h-Tv\cdot\nabla s+\frac{\partial\Phi}{\partial t}+v\cdot\nabla\Phi=0\tag{8}$$

Assume a steady state, dropping the time derivatives ($\partial/\partial t = 0$). Rearranging and factoring out $v\cdot\nabla$ yields:

$$v\cdot\nabla(v^{2}/2+h+\Phi)=Tv\cdot\nabla s\tag{9}$$

In a steady state, the entropy equation (7) requires that $\partial s/\partial t = 0$, leaving $v\cdot\nabla s=0$. Thus, the right side is zero, meaning the quantity in the parentheses is constant along parcel trajectories. This quantity is the Bernoulli constant:

$${\color{purple}\boxed{B=v^{2}/2+h+\Phi\tag{10}}}$$

For an ideal gas, specific enthalpy is defined as $h=C_{p}T$. Substitute the ideal gas relation $C_{p}T=\theta\Pi$ (utilizing the potential temperature and Exner function definitions found in goveqns 49 and goveqns 44) into the Bernoulli constant in place of $h$:

$${\color{purple}\boxed{B=v^{2}/2+\theta\Pi+\Phi\tag{11}}}$$

## Potential Flow in the Anelastic Case

By definition, purely irrotational velocity fields can be represented by a velocity potential $\chi$ such that $v=-\nabla\chi$.

In the anelastic case, expanding the anelastic mass continuity equation (see Eq. 47 in governing equations: $\vec{\nabla} \cdot (\rho_0 \vec{u}) = 0$) gives the three-dimensional divergence of velocity:

$$\nabla\cdot v=-\frac{d \ln \rho_{0}}{dz}w\tag{12}$$

Substitute $v=-\nabla\chi$ into the left side of the divergence equation to get $\nabla\cdot(-\nabla\chi) = -\nabla^{2}\chi$. Substitute the vertical velocity component $w = v_{z} = -\frac{\partial\chi}{\partial z}$ into the right side:

$$-\nabla^{2}\chi = -\frac{d \ln \rho_{0}}{dz}\left(-\frac{\partial\chi}{\partial z}\right)\tag{13}$$

Moving all terms to one side yields the potential flow equation:

$${\color{yellow}\boxed{\nabla^{2}\chi+\frac{d \ln \rho_{0}}{dz}\frac{\partial\chi}{\partial z}=0\tag{14}}}$$

## Two-Dimensional Slab Symmetry Streamfunction

Assuming independence of the $y$ dimension ($\partial/\partial y = 0$), the anelastic mass continuity equation (See Eq. 47 in governing equations) reduces to:

$$\frac{\partial\rho_{0}v_{x}}{\partial x}+\frac{\partial\rho_{0}v_{z}}{\partial z}=0\tag{15}$$

To satisfy this continuity equation automatically, a streamfunction $\psi$ is defined. By setting the partial derivatives of $\psi$ to balance the mass flux, we define the components:

$${\color{green}\boxed{v_{x}=-\frac{1}{\rho_{0}}\frac{\partial\psi}{\partial z}\tag{16}}}$$

$${\color{green}\boxed{v_{z}=\frac{1}{\rho_{0}}\frac{\partial\psi}{\partial x}\tag{17}}}$$

Note that by replacing equations (16) and (17) back into (15) confirms they consistently sum to zero.

## Pressure Coordinate Equations

These map the governing equations using pressure $p$ as the vertical coordinate rather than geometric height.

### **Mass Continuity**

In pressure coordinates, the "density" is constant ($\rho_{p}=-1/g$). Because this density doesn't change, the fluid behaves incompressibly in this coordinate system, making the 3D divergence zero. Utilizing the pressure vertical velocity $\omega=dp/dt$, this yields:


$${\color{red}\boxed{\frac{\partial u}{\partial x}+\frac{\partial v}{\partial y}+\frac{\partial\omega}{\partial p}=0\tag{18}}}$$

### **Horizontal Momentum**

Along constant pressure surfaces, there is no pressure gradient force. Instead, a gravitational force exists, acting as minus the slope of the pressure surface times gravity, or $-\nabla_{h}\Phi$. Balancing the material derivative of the horizontal velocities with this force and the Coriolis terms ($-fv$ and $fu$) while neglecting source and viscous terms yields:

$${\color{red}\boxed{\frac{du}{dt}+\frac{\partial\Phi}{\partial x}-fv=0\tag{19}}}$$

$${\color{red}\boxed{\frac{dv}{dt}+\frac{\partial\Phi}{\partial y}+fu=0\tag{20}}}$$

### **Geopotential/Hydrostatic Equation**

The hydrostatic equation is $d\Phi/dp=-RT/p$. Using the ideal gas relationship (see Eq. 48 in governing equations) $\kappa = R/C_{p}$, we can write $R = \kappa C_{p}$. Substituting $C_{p}T=\theta\Pi$ means $RT = \kappa\theta\Pi$. Replacing $RT$ in the hydrostatic equation gives:

$${\color{red}\boxed{\frac{\partial\Phi}{\partial p}+\frac{\kappa\theta\Pi}{p}=0\tag{21}}}$$

### **Thermodynamic Energy**

Assuming all source terms and heat additions are neglected, entropy is conserved. Consequently, potential temperature is constant along parcel trajectories, leaving:

$${\color{red}\boxed{\frac{d\theta}{dt}=0\tag{22}}}$$

### **1.Develop a total energy equation for the Boussinesq equations, ignoring viscosity and the buoyancy source term. Hint: The internal energy does not enter!**

We start with the Boussinesq momentum equation, omitting the viscous term $\nu \nabla^2 \vec{u}$:

$$\frac{d\vec{u}}{dt} + \vec{\nabla}\pi - b\hat{k} - f\hat{k} \times \vec{u} = 0$$

To find kinetic energy, we take the dot product of the entire equation with the velocity vector $\vec{u}$. Note that the Coriolis term vanishes because $\vec{u}\cdot(f\hat{k}\times\vec{u}) = 0$. Additionally, $\vec{u}\cdot\frac{d\vec{u}}{dt} = \frac{d(u^2/2)}{dt}$ and $\vec{u}\cdot\hat{k} = w$ (the vertical velocity):

$$\frac{d}{dt}\left(\frac{u^2}{2}\right) + \vec{u}\cdot\vec{\nabla}\pi - bw = 0$$

For the pressure gradient term, we apply the product rule and the Boussinesq mass continuity equation ($\vec{\nabla}\cdot\vec{u}=0$):

$$\vec{u}\cdot\vec{\nabla}\pi = \vec{\nabla}\cdot(\pi\vec{u}) - \pi(\vec{\nabla}\cdot\vec{u}) = \vec{\nabla}\cdot(\pi\vec{u})$$

For the buoyancy term, recall that $w = \frac{dz}{dt}$ and that buoyancy is conserved following the motion ($\frac{db}{dt}=0$) when there is no source. Applying the product rule gives:

$$-bw = -b\frac{dz}{dt} = -\frac{d(bz)}{dt} + z\frac{db}{dt} = -\frac{d(bz)}{dt}$$

We substitute these simplified terms back into the dotted momentum equation:

$$\frac{d}{dt}\left(\frac{u^2}{2} - bz\right) + \vec{\nabla}\cdot(\pi\vec{u}) = 0$$

This represents the conservation of total energy. Internal energy does not enter the equation because the fluid is assumed incompressible, meaning thermodynamic energy exchanges are entirely captured by the potential energy anomaly term ($-bz$).

## **2.The quantity $C_pT + \Phi$ is called the dry static energy.**

### (a) Assuming that the entropy is constant, obtain a relationship between $dp$ and $dT$ in a parcel of air.

From the definition of specific enthalpy ($h = e + p/\rho$) and the specific internal energy equation ($de = Tds - pd\alpha$), the differential of enthalpy is $dh = Tds + \frac{dp}{\rho}$.

For an ideal gas, $dh = C_p dT$. Equating these expressions yields $C_p dT = Tds + \frac{dp}{\rho}$.

Assuming the parcel's entropy is constant ($ds = 0$), the $Tds$ term drops out, leaving:

$$C_p dT = \frac{dp}{\rho}$$

### (b) Use the hydrostatic equation to eliminate $dp$ in favor of $dz$.

The hydrostatic equation dictates how pressure changes with height: $\frac{dp}{dz} = -g\rho$, which rearranges to $dp = -g\rho dz$.

Substitute this into the result from (a):

$$C_p dT = \frac{-g\rho dz}{\rho} = -g dz$$

### (c) Use in addition the ideal gas law to infer that $d(C_pT + gz) = d(C_pT + \Phi) = 0$ for a parcel, assuming that the parcel entropy is conserved.

Rearranging the result from (b) gives $C_p dT + g dz = 0$.

Geopotential is defined as $\Phi = gz$, so its differential is $d\Phi = g dz$.

Substituting $d\Phi$ into the equation gives $C_p dT + d\Phi = 0$, which integrates perfectly to prove that dry static energy is conserved for the parcel:

$$d(C_p T + \Phi) = 0$$

### (d) Can you reconcile this result with the Bernoulli equation? Explain.

The Bernoulli constant for an ideal gas in a steady state flow is $B = v^2/2 + C_p T + \Phi = \text{constant}$.

The derivation in steps (a)-(c) assumes a parcel in hydrostatic equilibrium, where vertical accelerations and kinetic energy ($v^2/2$) are negligible. If we set $v \approx 0$, the Bernoulli equation reduces identically to $C_p T + \Phi = \text{constant}$. Thus, dry static energy conservation is simply a specific case of the Bernoulli principle applied to static or slow-moving flows.

## **3. Derive the vorticity equation as in section 2.5, but for the Boussinesq equations.**

Start with the Boussinesq momentum equation, written with the material derivative expanded and viscosity ignored:

$$\frac{\partial \vec{u}}{\partial t} + (\vec{u}\cdot\vec{\nabla})\vec{u} + \vec{\nabla}\pi - b\hat{k} - f\hat{k}\times\vec{u} = 0$$

We apply the vector identity $(\vec{u}\cdot\vec{\nabla})\vec{u} = \vec{\nabla}(u^2/2) - \vec{u}\times\zeta$, where $\zeta = \vec{\nabla}\times\vec{u}$. Let absolute vorticity be $\zeta_a = \zeta + f\hat{k}$. Substitute and rearrange:

$$\frac{\partial \vec{u}}{\partial t} + \vec{\nabla}\left(\frac{u^2}{2} + \pi\right) + \zeta_a\times\vec{u} - b\hat{k} = 0$$

Take the curl ($\vec{\nabla}\times$) of the entire equation. Because the curl of any gradient is zero, $\vec{\nabla}\times\vec{\nabla}(u^2/2 + \pi)$ drops out entirely:

$$\frac{\partial \zeta_a}{\partial t} + \vec{\nabla}\times(\zeta_a\times\vec{u}) - \vec{\nabla}\times(b\hat{k}) = 0$$

Expand the middle term using the vector identity $\vec{\nabla}\times(\zeta_a\times\vec{u}) = \vec{\nabla}\cdot(\vec{u}\zeta_a - \zeta_a\vec{u})$. The final term is the baroclinic generation term. Rearranging gives the final vorticity equation:

$$\frac{\partial \zeta_a}{\partial t} + \vec{\nabla}\cdot(\vec{u}\zeta_a - \zeta_a\vec{u}) = \vec{\nabla}\times(b\hat{k})$$
