# Geophysical Fluid Dynamics

When we study fluid dynamics, we are interested in understanding how fluids move and evolve over time. But what exactly is Geophysical Fluid Dynamics (GFD)? GFD is the application of fluid dynamics to naturally occurring fluids on Earth and other planets, such as the atmosphere and the oceans. Since these fluids are part of the Earth system, they are called *geophysical fluids*. By studying their motion, we can better understand important phenomena such as weather patterns, ocean currents, and climate. Two key features that make geophysical fluids unique are the rotation of the Earth, which affects fluid motion, and stratification, which refers to the layering of fluids with different properties. In this series of notebooks, we will introduce the main concepts of GFD, combining physical intuition, mathematical derivations, and simple examples to help build a solid understanding of the subject.

## 2. Tools and Tricks of Fluid Dynamics

Here we introduce a collection of mathematical tools designed to help you "make friends" with the governing equations of geophysical fluid dynamics. Furthermore, these tools provide a clearer physical perspective, helping you better interpret the fundamental concepts of atmospheric and oceanic motion explored throughout these notebooks.
<hr style="border:none; border-top:1px solid #888;">

### Exterior boundary conditions
<div style="border-left: 5px solid #1f77b4; background-color: #eaf4ff; padding: 12px; border-radius: 5px;">
    
When using the momentum equation (<a href="https://github.com/VALENTINAMOVILSANDOVAL/gfd_lab/blob/main/goveqns.ipynb" target="_blank">see the governing equations notebook</a>), we must specify appropriate boundary conditions (e.g., at the Earth's surface). A key question is whether viscous effects can be neglected or must be retained in the momentum equation. To assess this, we compare inertial (acceleration) forces with viscous (frictional) forces. Their ratio defines the **Reynolds number**, which indicates the relative importance of viscosity and helps determine the appropriate dynamical regime and boundary conditions.

</div>
<br>
To derive its mathematical form, we start by estimating the magnitude of the acceleration term. Assuming a velocity scale $V$, length scale $L$, and time scale $T$, the acceleration $V/T$ can be rewritten by substituting $T=L/V$. This yields an acceleration magnitude of $V^{2}/L$. Next, we estimate the magnitude of the frictional term, which is proportional to $\nu \nabla^{2}v$. Assuming $\nabla^{2}\approx1/L^{2}$, the frictional term scales as $\nu V/L^{2}$.

Taking the ratio of the acceleration term to the friction term yields:

$$Re=\frac{V^{2}/L}{\nu V/L^{2}}\tag{2.1}$$

Simplifying the fraction gives the final formula for the Reynolds number:

$$\boxed{Re=\frac{VL}{\nu}\tag{2.2}}$$

**Note:** For flows with a large Reynolds number ($Re \gg 1$), viscous effects are often neglected in the fluid interior, leading to **free-slip boundary conditions**, where the fluid cannot flow through the boundary but may slide tangentially along it. When viscosity is included, **no-slip boundary conditions** apply, requiring the fluid velocity at the boundary to match the velocity of the surface.

### Interior boundary conditions

<div style="border-left: 5px solid #1f77b4; background-color: #eaf4ff; padding: 12px; border-radius: 5px;">

When a flow is divided into different fluid regions (e.g., the atmosphere and the ocean), the interface between them must satisfy additional boundary conditions. The two fluids must move consistently across the interface, and the forces they exert on each other (traction) must be continuous to satisfy Newton's third law. Likewise, the normal heat flux must be continuous to prevent artificial sources or sinks of heat at the interface. If viscosity and heat conduction are neglected, pressure becomes the only force transmitted across the interface, and temperature is not directly affected by the adjacent fluid.

</div>

### Energy equation

We begin with the momentum equation for a geophysical fluid:

$$
\frac{dv}{dt}+\frac{1}{\rho}\nabla p+gk+2\Omega\times v=\nu\nabla^{2}v
$$

To obtain an equation for kinetic energy, we take the dot product of the momentum equation with the velocity vector $v$, noting that the Coriolis force does no work and the gravity term relates to the material derivative of the geopotential $\Phi$:

$$
\frac{d(v^{2}/2)}{dt}=-\frac{1}{\rho}\nabla\cdot(pv)+p\frac{d\alpha}{dt}-\frac{d\Phi}{dt}+\nu v\cdot\nabla^{2}v \tag{2.3}
$$

The pressure contribution to the work is expanded by invoking the product rule, using the mass continuity equation $d\rho/dt=-\rho\nabla\cdot v$, and defining the specific volume $\alpha=1/\rho$:

$$
-\frac{1}{\rho}v\cdot\nabla p=-\frac{1}{\rho}\nabla\cdot(pv)+\frac{p}{\rho}\nabla\cdot v=-\frac{1}{\rho}\nabla\cdot(pv)-\frac{p}{\rho^{2}}\frac{d\rho}{dt}=-\frac{1}{\rho}\nabla\cdot(pv)+p\frac{d\alpha}{dt} \tag{2.4}
$$

Using the differential form of the specific internal energy equation $de=Tds-pd\alpha$, the pressure term may finally be written:

$$
-\frac{1}{\rho}v\cdot\nabla p=-\frac{1}{\rho}\nabla\cdot(pv)-\frac{de}{dt}+T\frac{ds}{dt} \tag{2.5}
$$

The viscous term is expanded using tensor notation to demonstrate the standard vector identity:

$$
v\cdot\nabla^{2}v=v_{i}\frac{\partial^{2}v_{i}}{\partial x_{j}\partial x_{j}}=\frac{\partial}{\partial x_{j}}\left(v_{i}\frac{\partial v_{i}}{\partial x_{j}}\right)-\frac{\partial v_{i}}{\partial x_{j}}\frac{\partial v_{i}}{\partial x_{j}}=\nabla^{2}(v^{2}/2)-|\nabla v|^{2} \tag{2.6}
$$

Substituting the expanded pressure and viscous terms back into the kinetic energy equation, and expressing the kinematic viscosity as $\nu=\mu/\rho$, allows us to group the material derivatives and divergence terms to obtain the first advective form of the energy equation:

$$
\frac{d}{dt}(v^{2}/2+e+\Phi)+\frac{1}{\rho}\nabla\cdot[pv-\mu\nabla(v^{2}/2)]=T\frac{ds}{dt}-\nu|\nabla v|^{2} \tag{2.7}
$$

Because dissipated mechanical energy acts as an internal heat source that balances entropy increase, the terms on the right side are combined into a single term representing heating strictly from external sources, producing the final advective form:

$$
\boxed{\frac{d}{dt}(v^{2}/2+e+\Phi)+\frac{1}{\rho}\nabla\cdot[pv-\mu\nabla(v^{2}/2)]=T\frac{ds_X}{dt}} \tag{2.8}
$$

To convert the advective form into the flux form, we multiply the equation by the density $\rho$ and apply the mass continuity equation, $\frac{\partial\rho}{\partial t} + \nabla\cdot(\rho v) = 0$, to transform the material derivative:

$$
\frac{\partial}{\partial t}[\rho(v^{2}/2+e+\Phi)]+\nabla\cdot[\rho v(v^{2}/2+e+\Phi)]+\nabla\cdot[pv-\mu\nabla(v^{2}/2)]=\rho T\frac{ds_X}{dt} \tag{2.9}
$$

Grouping the divergence terms together and substituting the specific enthalpy $h=e+p/\rho$ yields the flux form of the energy equation:

$$
\boxed{\frac{\partial}{\partial t}[\rho(v^{2}/2+e+\Phi)]+\nabla\cdot[\rho(v^{2}/2+h+\Phi)v-\mu\nabla(v^{2}/2)]=\rho T\frac{ds_X}{dt}} \tag{2.10}
$$

Finally, integrating over a control volume yields the energy budget equation. This states that the rate of change of total energy (kinetic, internal, and potential) within the volume equals the net energy flux across its boundaries plus the energy added by external heating sources:

$$
\boxed{\frac{d}{dt}\int\rho(v^{2}/2+e+\Phi)dV+\oint[\rho(v^{2}/2+h+\Phi)v-\mu\nabla(v^{2}/2)]\cdot n dA=\int\rho T\frac{ds_X}{dt}dV} \tag{2.11}
$$

### Bernoulli Equation

We extend the analysis to an arbitrary Newtonian fluid subject to conservative forces, explicitly neglecting viscosity and external heat sources. As in the energy equation derivation, we take the dot product of the momentum equation with the velocity vector $v$. Using the definition of specific enthalpy $dh=Tds+dp/\rho$, expanding the material derivatives of kinetic energy and geopotential yields the following equation:

$$
\frac{\partial (v^{2}/2)}{\partial t}+v\cdot\nabla(v^{2}/2)+v\cdot\nabla h-Tv\cdot\nabla s+\frac{\partial\Phi}{\partial t}+v\cdot\nabla\Phi=0
$$

Combining terms and dropping the time derivatives to assume a steady state results in:

$$
v\cdot\nabla(v^{2}/2+h+\Phi)=Tv\cdot\nabla s
$$

The entropy equation with no heat source or heat conduction is:

$$
\frac{\partial s}{\partial t}+v\cdot\nabla s=0
$$

Applying the steady-state assumption to this entropy equation dictates that the right side of the combined gradient equation is zero. The remaining quantity inside the parentheses on the left side is identified as the Bernoulli constant:

$$
\boxed{B=v^{2}/2+h+\Phi}
$$

For an ideal gas, the specific enthalpy is $h=C_{p}T$. Using the relationship $C_{p}T=\theta\Pi$, we substitute the specific enthalpy with the potential temperature and the Exner function. This leads to the most useful form of the Bernoulli constant for the atmosphere:

$$
\boxed{B=v^{2}/2+\theta\Pi+\Phi}
$$

### Three-Dimensional Flow

To understand complex three-dimensional fluid motion, it is helpful to break the velocity field down into simpler components. Any vector field can be decomposed into two parts: an **irrotational** part (which has no spin or rotation) and a **solenoidal** part (which has no divergence, meaning it neither expands nor compresses).

We express this decomposition of the velocity field $v$ as:

$$
v=v_{I}+v_{S} \tag{2.12}
$$

Because the solenoidal component $v_{S}$ is non-divergent by definition, it can be mathematically represented as the curl (a measure of rotation) of a vector potential, denoted as $\Psi$:

$$
v_{S}=\nabla\times\Psi \tag{2.13}
$$

A fundamental concept in fluid dynamics is **vorticity** ($\zeta$), which describes the local spinning motion of the fluid. It is defined as the curl of the velocity field. Since the irrotational part $v_{I}$ has zero spin, the total vorticity is simply the curl of the solenoidal part:

$$
\zeta=\nabla\times v_{S} \tag{2.14}
$$

Substituting our definition of $v_{S}$ from equation 2.13 into equation 2.14 gives:

$$
\zeta=\nabla\times(\nabla\times\Psi) \tag{2.15}
$$

To simplify this, we apply a standard vector identity for the curl of a curl:

$$
\nabla\times(\nabla\times\Psi) = \nabla(\nabla\cdot\Psi) - \nabla^{2}\Psi \tag{2.16}
$$

We are free to choose a mathematical constraint to make our equations easier to solve without changing the physical reality. Here, we apply the "Coulomb gauge condition," which sets the divergence of the vector potential to zero:

$$
\nabla\cdot\Psi=0 \tag{2.17}
$$

Applying this condition to our vector identity eliminates the first term on the right side, yielding a direct and elegant relationship between vorticity and the vector potential:

$$
\boxed{\zeta=-\nabla^{2}\Psi} \tag{2.18}
$$

<div style="border-left: 5px solid #fbc02d; background-color: #fff9c4; padding: 12px; border-radius: 5px;">

Before proceeding to the vorticity equation, we must define the term **baroclinic**. In a baroclinic fluid, surfaces of constant pressure cross surfaces of constant density (or temperature). This misalignment creates an imbalance that generates rotational motion (vorticity). The mathematical representation of this effect is the "baroclinic term," formed by the cross product of the temperature gradient and the pressure gradient. 

</div>

<br>

To see how vorticity evolves over time, we rewrite the momentum equation. We ignore viscosity, represent the pressure gradient with the Exner function ($\Pi$), represent gravity with the geopotential ($\Phi$), and define the absolute vorticity as $\zeta_{a} = \zeta + 2\Omega$ (which includes the Earth's rotation):

$$
\frac{\partial v}{\partial t}+\nabla(v^{2}/2+\Phi)+\zeta_{a}\times v+\theta\nabla\Pi=0 \tag{2.19}
$$

Taking the curl of this entire momentum equation gives us the **governing equation for vorticity**. The curl of the gradient terms vanishes. The time derivative of velocity becomes the time derivative of absolute vorticity. Finally, applying the product rule to the pressure and temperature terms yields the baroclinic term ($\nabla\theta\times\nabla\Pi$). The resulting vorticity equation is:

$$
\boxed{\frac{\partial\zeta_{a}}{\partial t}+\nabla\cdot(v\zeta_{a}-\zeta_{a}v)+\nabla\theta\times\nabla\Pi=0} \tag{2.20}
$$

Alternatively, if a flow has absolutely no spin, it is purely irrotational ($\zeta = 0$). In this special case, the entire velocity field can be described simply by the gradient of a single scalar quantity called a velocity potential, $\chi$:

$$
v=-\nabla\chi \tag{2.21}
$$

For an **anelastic** flow—a flow model that filters out sound waves but allows for changes in density with height—the three-dimensional divergence of the velocity is proportional to the vertical velocity $w$ and the vertical gradient of the ambient density profile $\rho_{0}(z)$:

$$
\nabla\cdot v=-\frac{d \ln \rho_{0}}{dz}w \tag{2.22}
$$

By substituting our velocity potential definition ($v=-\nabla\chi$) into this anelastic divergence equation, the left side becomes $-\nabla^{2}\chi$ and the vertical velocity $w$ on the right side becomes $-\frac{\partial\chi}{\partial z}$. Equating these and moving them to the same side yields the **governing equation for anelastic potential flow**:

$$
\boxed{\nabla^{2}\chi+\frac{d \ln \rho_{0}}{dz}\frac{\partial\chi}{\partial z}=0} \tag{2.23}
$$

### Two-dimensional slab symmetry

We consider a case of reduced dimensionality by assuming slab symmetry, where the flow is independent of the horizontal $y$ dimension ($\partial/\partial y=0$). In the anelastic case, the mass continuity equation reduces to:

$$
\frac{\partial\rho_{0}v_{x}}{\partial x}+\frac{\partial\rho_{0}v_{z}}{\partial z}=0 \tag{2.24}
$$

This formulation allows us to define a streamfunction $\psi$ such that the velocity components are expressed as:

$$
v_{x}=-\frac{1}{\rho_{0}}\frac{\partial\psi}{\partial z}, \quad v_{z}=\frac{1}{\rho_{0}}\frac{\partial\psi}{\partial x} \tag{2.25}
$$

Direct substitution of these components into equation 2.24 verifies that this definition is perfectly consistent with mass continuity. We then define the vorticity for this two-dimensional flow in the $x-z$ plane.

$$
\boxed{\zeta=\frac{\partial v_{x}}{\partial z}-\frac{\partial v_{z}}{\partial x}} \tag{2.26}
$$

Substituting the streamfunction definitions of the velocity components into this vorticity equation yields a Poisson-like elliptic equation for the streamfunction in terms of the vorticity.

$$
\boxed{\frac{\partial}{\partial x}\left(\frac{1}{\rho_{0}}\frac{\partial\psi}{\partial x}\right)+\frac{\partial}{\partial z}\left(\frac{1}{\rho_{0}}\frac{\partial\psi}{\partial z}\right)=-\zeta} \tag{2.27}
$$

### Pressure Coordinates

<div style="border-left: 5px solid #1f77b4; background-color: #eaf4ff; padding: 12px; border-radius: 5px;">

In meteorology, it is frequent to use pressure as the vertical coordinate instead of geometric height. This coordinate transformation is generally performed only in the hydrostatic limit. The fundamental difference is that instead of treating pressure as something that changes at a given fixed height, the geopotential $\Phi=gz$ becomes a dependent variable while pressure becomes an independent variable. 

Working in this system means that surfaces of constant pressure become the "horizontal" surfaces. Because these constant pressure surfaces undulate and are not perfectly flat in geometric space, this coordinate system is not strictly Cartesian. Consequently, fluid moving strictly along a constant pressure surface experiences no pressure gradient force. Instead, a gravitational force drives horizontal motion, represented by: $-\nabla_{h}\Phi$. 
</div>

<br>

We begin with the hydrostatic equation, which relates changes in geopotential to changes in pressure:

$$
\frac{d\Phi}{dp}=-\frac{RT}{p} \tag{2.28}
$$

To determine mass continuity, we define the density in pressure coordinates as the amount of mass divided by a volume element defined in pressure space. In hydrostatic equilibrium, the weight of the material equals the volume, which means dividing by gravity yields a constant density of $-1/g$:

$$
\rho_{p}=-\frac{1}{g} \tag{2.29}
$$

The vertical velocity in this system is defined simply as the material derivative of pressure. Because pressure naturally decreases as you move upward in the atmosphere, upward physical motion translates to a negative vertical velocity in pressure coordinates:

$$
\omega=\frac{dp}{dt} \tag{2.30}
$$

The material derivative is then expanded to include the partial time derivative, the horizontal advection terms, and our newly defined vertical velocity $\omega$:

$$
\frac{d}{dt}=\frac{\partial}{\partial t}+u\frac{\partial}{\partial x}+v\frac{\partial}{\partial y}+\omega\frac{\partial}{\partial p} \tag{2.31}
$$

Because the density in pressure coordinates is a constant, **the mass continuity equation** simplifies dramatically, taking on a form identical to an incompressible fluid:

$$
\boxed{\frac{\partial u}{\partial x}+\frac{\partial v}{\partial y}+\frac{\partial\omega}{\partial p}=0} \tag{2.32}
$$

**The horizontal momentum equations** are modified by replacing the standard pressure gradient force with the gravitational force acting along the sloping pressure surfaces, yielding **the zonal ($x$) momentum equation**:

$$
\boxed{\frac{du}{dt}+\frac{\partial\Phi}{\partial x}-fv=0} \tag{2.33}
$$

Applying the exact same substitution for **the meridional direction yields the $y$-momentum equation**:

$$
\boxed{\frac{dv}{dt}+\frac{\partial\Phi}{\partial y}+fu=0} \tag{2.34}
$$

**The hydrostatic equation** is reformulated by eliminating temperature in favor of the potential temperature $\theta$ and the pressure-dependent Exner function $\Pi$:

$$
\boxed{\frac{\partial\Phi}{\partial p}+\frac{\kappa\theta\Pi}{p}=0} \tag{2.35}
$$

Finally, having neglected all source and viscous terms, the thermodynamic equation simply states that **potential temperature** is conserved following the fluid motion:

$$
\boxed{\frac{d\theta}{dt}=0} \tag{2.36}
$$

**1.Develop a total energy equation for the Boussinesq equations, ignoring viscosity and the buoyancy source term. Hint: The internal energy does not enter!**

**A:** We start with the Boussinesq momentum equation, omitting the viscous term $\nu \nabla^2 \vec{u}$:

$$\frac{d\vec{u}}{dt} + \vec{\nabla}\pi - b\hat{k} - f\hat{k} \times \vec{u} = 0$$

To find kinetic energy, we take the dot product of the entire equation with the velocity vector $\vec{u}$. Note that the Coriolis term vanishes because $\vec{u}\cdot(f\hat{k}\times\vec{u}) = 0$. Additionally, $\vec{u}\cdot\frac{d\vec{u}}{dt} = \frac{d(u^2/2)}{dt}$ and $\vec{u}\cdot\hat{k} = w$ (the vertical velocity):

$$\frac{d}{dt}\left(\frac{u^2}{2}\right) + \vec{u}\cdot\vec{\nabla}\pi - bw = 0$$

For the pressure gradient term, we apply the product rule and the Boussinesq mass continuity equation ($\vec{\nabla}\cdot\vec{u}=0$):

$$\vec{u}\cdot\vec{\nabla}\pi = \vec{\nabla}\cdot(\pi\vec{u}) - \pi(\vec{\nabla}\cdot\vec{u}) = \vec{\nabla}\cdot(\pi\vec{u})$$

For the buoyancy term, recall that $w = \frac{dz}{dt}$ and that buoyancy is conserved following the motion ($\frac{db}{dt}=0$) when there is no source. Applying the product rule gives:

$$-bw = -b\frac{dz}{dt} = -\frac{d(bz)}{dt} + z\frac{db}{dt} = -\frac{d(bz)}{dt}$$

We substitute these simplified terms back into the dotted momentum equation:

$$\frac{d}{dt}\left(\frac{u^2}{2} - bz\right) + \vec{\nabla}\cdot(\pi\vec{u}) = 0$$

This represents the conservation of total energy. Internal energy does not enter the equation because the fluid is assumed incompressible, meaning thermodynamic energy exchanges are entirely captured by the potential energy anomaly term ($-bz$).

**2.The quantity $C_pT + \Phi$ is called the dry static energy.**

**(a) Assuming that the entropy is constant, obtain a relationship between $dp$ and $dT$ in a parcel of air.**

**A:** From the definition of specific enthalpy ($h = e + p/\rho$) and the specific internal energy equation ($de = Tds - pd\alpha$), the differential of enthalpy is $dh = Tds + \frac{dp}{\rho}$.

For an ideal gas, $dh = C_p dT$. Equating these expressions yields $C_p dT = Tds + \frac{dp}{\rho}$.

Assuming the parcel's entropy is constant ($ds = 0$), the $Tds$ term drops out, leaving:

$$C_p dT = \frac{dp}{\rho}$$

**(b) Use the hydrostatic equation to eliminate $dp$ in favor of $dz$.**

**A:** The hydrostatic equation dictates how pressure changes with height: $\frac{dp}{dz} = -g\rho$, which rearranges to $dp = -g\rho dz$.

Substitute this into the result from (a):

$$C_p dT = \frac{-g\rho dz}{\rho} = -g dz$$

**(c) Use in addition the ideal gas law to infer that $d(C_pT + gz) = d(C_pT + \Phi) = 0$ for a parcel, assuming that the parcel entropy is conserved.**

**A:** Rearranging the result from (b) gives $C_p dT + g dz = 0$.

Geopotential is defined as $\Phi = gz$, so its differential is $d\Phi = g dz$.

Substituting $d\Phi$ into the equation gives $C_p dT + d\Phi = 0$, which integrates perfectly to prove that dry static energy is conserved for the parcel:

$$d(C_p T + \Phi) = 0$$

**(d) Can you reconcile this result with the Bernoulli equation? Explain.**

**A:** The Bernoulli constant for an ideal gas in a steady state flow is $B = v^2/2 + C_p T + \Phi = \text{constant}$.

The derivation in steps (a)-(c) assumes a parcel in hydrostatic equilibrium, where vertical accelerations and kinetic energy ($v^2/2$) are negligible. If we set $v \approx 0$, the Bernoulli equation reduces identically to $C_p T + \Phi = \text{constant}$. Thus, dry static energy conservation is simply a specific case of the Bernoulli principle applied to static or slow-moving flows.

**3. Derive the vorticity equation as in section 2.5, but for the Boussinesq equations.**

**A:** Start with the Boussinesq momentum equation, written with the material derivative expanded and viscosity ignored:

$$\frac{\partial \vec{u}}{\partial t} + (\vec{u}\cdot\vec{\nabla})\vec{u} + \vec{\nabla}\pi - b\hat{k} - f\hat{k}\times\vec{u} = 0$$

We apply the vector identity $(\vec{u}\cdot\vec{\nabla})\vec{u} = \vec{\nabla}(u^2/2) - \vec{u}\times\zeta$, where $\zeta = \vec{\nabla}\times\vec{u}$. Let absolute vorticity be $\zeta_a = \zeta + f\hat{k}$. Substitute and rearrange:

$$\frac{\partial \vec{u}}{\partial t} + \vec{\nabla}\left(\frac{u^2}{2} + \pi\right) + \zeta_a\times\vec{u} - b\hat{k} = 0$$

Take the curl ($\vec{\nabla}\times$) of the entire equation. Because the curl of any gradient is zero, $\vec{\nabla}\times\vec{\nabla}(u^2/2 + \pi)$ drops out entirely:

$$\frac{\partial \zeta_a}{\partial t} + \vec{\nabla}\times(\zeta_a\times\vec{u}) - \vec{\nabla}\times(b\hat{k}) = 0$$

Expand the middle term using the vector identity $\vec{\nabla}\times(\zeta_a\times\vec{u}) = \vec{\nabla}\cdot(\vec{u}\zeta_a - \zeta_a\vec{u})$. The final term is the baroclinic generation term. Rearranging gives the final vorticity equation:

$$\frac{\partial \zeta_a}{\partial t} + \vec{\nabla}\cdot(\vec{u}\zeta_a - \zeta_a\vec{u}) = \vec{\nabla}\times(b\hat{k})$$

**4. Show that the flux of the vertical component of absolute vorticity is purely horizontal. Hint: $F_{\zeta ij} = v_i \zeta_{aj} - v_j \zeta_{ai}$ is the flux of the $j$th component of the vorticity in the $i$ direction.**

**A:** The flux tensor of absolute vorticity is given by $F_{\zeta ij} = v_i \zeta_{aj} - v_j \zeta_{ai}$. This describes the flux of the $j$-th component of vorticity moving in the $i$-th direction

We want to isolate the flux of the vertical component of vorticity, so we set $j = 3$ (the $z$-direction). The flux vector for this component is defined by $(F_{\zeta 13}, F_{\zeta 23}, F_{\zeta 33})$.

To determine if this vector has a vertical direction, we evaluate its vertical component by setting $i = 3$:

$$F_{\zeta 33} = v_3 \zeta_{a3} - v_3 \zeta_{a3} = 0$$

Because $F_{\zeta 33}$ exactly cancels itself out, the flux vector has zero vertical movement. It only consists of $x$ ($F_{\zeta 13}$) and $y$ ($F_{\zeta 23}$) components, meaning it is purely horizontal.
